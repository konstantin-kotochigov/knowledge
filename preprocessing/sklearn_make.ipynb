```python
import matplotlib.pyplot as plt
import numpy as np

def visualize_dataset(X, y=None, title="Dataset"):

    plt.figure(figsize=(6, 6))

    if y is None:
        plt.scatter(X[:, 0], X[:, 1], s=15, alpha=0.7)
    else:
        y = np.asarray(y)
        # эвристика: регрессия если уникальных значений много
        if np.issubdtype(y.dtype, np.floating) and len(np.unique(y)) > 20:
            sc = plt.scatter(X[:, 0], X[:, 1], c=y, s=15, cmap="viridis", alpha=0.7)
            plt.colorbar(sc, label="y")
        else:
            for cls in np.unique(y):
                mask = y == cls
                plt.scatter(X[mask, 0], X[mask, 1], s=15, alpha=0.7, label=str(cls))
            plt.legend(title="class")

    plt.title(title)
    plt.xlabel("x₁")
    plt.ylabel("x₂")
    plt.grid(alpha=0.3)
    plt.show()
```

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=1500,
    centers=[(-3, -3), (0, 0), (3, 3)],  # фиксированные центры (или число центров)
    cluster_std=[1.0, 0.5, 1.3],         # разброс по кластерам
    n_features=2,
    random_state=0
)

visualize_dataset(X,y, title='make_blobs')
```

```python
from sklearn.datasets import make_classification
X, y = make_classification(
    n_samples=5000,
    n_features=2,              # всего признаков
    n_informative=2,            # информативные (влияют на класс)
    n_redundant=0,              # линейные комбинации информативных
    n_repeated=0,               # дубликаты
    n_classes=3,
    n_clusters_per_class=1,
    weights=[0.7, 0.2, 0.1],    # дисбаланс классов
    class_sep=1.5,              # разделимость классов
    flip_y=0.02,                # шум в метках
    random_state=42
)

visualize_dataset(X,y, title='make_classification')
```

```python
from sklearn.datasets import make_circles, make_moons
X1, y1 = make_circles(n_samples=1000, factor=0.4, noise=0.06, random_state=0)
X2, y2 = make_moons(n_samples=1000, noise=0.2, random_state=0)

visualize_dataset(X1,y1, title='make_circles')
```

```python
from sklearn.datasets import make_gaussian_quantiles
X, y = make_gaussian_quantiles(
    mean=None, cov=2.0, n_samples=2000, n_features=2, n_classes=3, random_state=0
)

visualize_dataset(X,y, title='make_gaussian_quantiles')
```

```python
from sklearn.datasets import make_hastie_10_2
X, y = make_hastie_10_2(n_samples=2000, random_state=0)  # y ∈ {-1, +1}

visualize_dataset(X,y, title='make_hastie_10_2')
```

```python
from sklearn.datasets import make_regression
X, y, coef = make_regression(
    n_samples=3000,
    n_features=2,
    n_informative=2,
    n_targets=1,
    bias=10.0,
    noise=15.0,          # гауссов шум в целевой
    coef=True,           # вернуть истинные коэффициенты
    effective_rank=None, # или задайте ранг для «почти низкоранговых» данных
    random_state=42
)

visualize_dataset(X,y, title='make_hastie_10_2')
```

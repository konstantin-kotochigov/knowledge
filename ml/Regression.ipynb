### Вывод формулы
Необходимое условие на оптимальность - чтобы производная $SS_{MODEL}$ равнялась нулю<Br>

$$L(\boldsymbol{\beta}) = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 = (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})^\top(\mathbf{y} - \mathbf{X}\boldsymbol{\beta}) \rightarrow \min$$

$$\frac{\partial S}{\partial \boldsymbol{\beta}} = -2\mathbf{X}^\top\mathbf{y} + 2\mathbf{X}^\top\mathbf{X}\boldsymbol{\beta} = \mathbf{0}$$

Это условие называется Normal Equations:<br>
$X^TX \beta = X^T y$

Или раскрывая матричную нотакцию:

$\underbrace{\begin{pmatrix} n & \sum x_{i1} & \cdots & \sum x_{ip} \\ \sum x_{i1} & \sum x_{i1}^2 & \cdots & \sum x_{i1}x_{ip} \\ \vdots & & \ddots & \vdots \\ \sum x_{ip} & \sum x_{ip}x_{i1} & \cdots & \sum x_{ip}^2 \end{pmatrix}}_{X^TX} \begin{pmatrix} \hat{\beta}_0 \\ \hat{\beta}_1 \\ \vdots \\ \hat{\beta}_p \end{pmatrix} = \underbrace{\begin{pmatrix} \sum y_i \\ \sum y_i x_{i1} \\ \vdots \\ \sum y_i x_{ip} \end{pmatrix}}_{X^Ty}$

Во-первых, получаем формулу<br>
$\beta = (X^TX)^{-1} X^T y$

Во-вторых, если вынесем $X^T$, получим запись необходимого условия через остатки:<br>
$X^T \underbrace{(y - X \beta)}_{\epsilon} = X \varepsilon = 0$

Из него следует необходимое подусловие МНК-регресии:

- $\sum \varepsilon = 0$<br>
  Почему: первая колонка в матрице $X$ - первое измерение всех наблюдений, то есть константа "1" (специально так выбрано для свободного коэффициента  $\beta_0$), поэтому из первого уравнения системы $X^T \varepsilon = 0$ сразу следует: сумма ошибок $\varepsilon$ равна нулю<br>
  $\sum_i 1 \cdot \varepsilon_i = \varepsilon_1 + \varepsilon_2 + ... + \varepsilon_n = 0$<br>
  Иными словами, МНК-регрессия обязана проходить через центроид $(\overline{X},\overline{Y})$<br>
  Остальные уравнения задают уже конкретный наклон прямой





### Coefficient of Determination
При наличии некоторой линейной моделиb
$SS_{TOT} = SS_{EXP} + SS_{RES}$

Откуда такое приятное соотношение?<br>
$$SS_{TOT} = \sum_i (y - \overline{y})^2 = \sum (y - \hat{y} + \hat{y} - \overline{y})^2 = \underbrace{\sum_i (y - \hat{y})^2}_{SS_{RES}} + \underbrace{2 \sum_i (y - \overline{y}) (\hat{y} - \overline{y})}_{0} + \underbrace{\sum_i (\hat{y} - \overline{y})^2}_{SS_{EXP}}$$

Средний член равен нулю так как 
$$\sum_i (y - \hat{y}) (\hat{y} - \overline{y}) = \sum (\epsilon_i (\hat{y} - \overline{y}) = \sum \epsilon \hat{y} - \overline{y} \cdot \underbrace{\sum \epsilon_i}_{=0} = 0$$

Здесь применили два свойства МНК-регресии
- сумма ошибок равна нулю (регрессия всегда идет через центроид)
- $X \varepsilon = 0$

Где еще это осотношение:
- ANOVA<br>тут дисперсия смеси складывается из дисперсий отдельных групп
- 

$$R^2=1 - \frac{\text{Var[y|x]}}{\text{Var[y]}}$$

where $\text{Var}[y|x]$ is a conditional variance

Это то же самое, что F-тест: тест Фишера для сравнения дисперсий "до" / "после"<br>статистически проверяем, что R^2 не нулевой

$$R^2 = Corr^2(y,f)$$

In simple linear regression<br>
$\beta = \text{Corr} \cdot \frac{\sigma_y}{\sigma_x}$

# Gauss-Markov

Формулировка:
> Пусть данные генерируются по линейному закону $\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ и при этом:
> 1. $\mathbf{X} \in \mathbb{R}^{n \times (p+1)}$ фиксирована
> 2. $\mathbb{E}[\boldsymbol{\varepsilon} \mid \mathbf{X}] = \mathbf{0}$
> 3. $\text{Var}(\boldsymbol{\varepsilon} \mid \mathbf{X}) = \sigma^2\mathbf{I}$
> 4. $\text{rank}(\mathbf{X}) = p+1$
> 
> Тогда МНК регрессия (OLS estimator) $\hat{\boldsymbol{\beta}} = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\mathbf{y}$ является BLUE (best linear unbiased estimator), то есть на всем множетсве линейных оценок именно у нее наименьшая возможная погрешность

Пусть есть множество Generating Laws. Среди них выделяем подмножество тех, которые линейные и у которых шум однородно случаный в соотвествии с Assumptions из теоремы. Далее вывод и утверждение теоремы ограничивается этим подможеством, в том числе 

О тех данных, к которым регрессию применяют, теорема ничего не говорит. Они вполне могут этим условиям не соотвествовать. Отвественность аналитика проверять на соотвествие этому классу

Еще важно не путать две "линейности"<br>
Мы рассамтриваем класс моделей, описывающих данные линейно: $y = X \beta + \varepsilon$<br>
Estimator линейный, если оценка $\hat{\beta}$ получена линейным преобразованием данных $\hat{\beta} = Ay$

### Доказательство эффективности

- <span style="color:darkblue">Опишем класс линейных несмщенных оценок</span><br>Допустим есть какой-то другой Linear Unbiased Estimator $\tilde{\beta} = AX$. Выразим его через наш OLS-ный:<br>
$\mathbf{A} = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top + \mathbf{D}$<br><br>
С учетом того, что он unbiased:<br>
$\mathbb{E}[\tilde{\boldsymbol{\beta}}] = \mathbf{A}\mathbf{X}\boldsymbol{\beta} = \mathbf{A}\mathbb{E}[\mathbf{y}] = \boldsymbol{\beta} \quad \forall \boldsymbol{\beta}$<br><br>
Значит для любой такой оценки A должно выполняться<br> $AX = I$<br><br>
Если подставим сюда A...<br> 
$\left[(\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top + \mathbf{D}\right]\mathbf{X} = \mathbf{I}$<br><br>
...и раскроем скобки, то получим необходимое условие для оценки, чтобы она была Linear Unbiased Estimator:<br>
$\boxed{DX = 0}$<br><br>
Итого, описали многообразие линейных несмешенных оценок, через смещение относительно $\beta_{OLS}$: такие матрицы, что $DX=0$<br>при $D = 0$ это OLS регрессия $(X^TX)^{-1}X$

- <span style="color:darkblue">Теперь выразим дисперсию $Var(\tilde{\beta})$ через дисперсию OLS оценки</span><br>
  $\text{Var}(\tilde{\boldsymbol{\beta}}) = \text{Var}(AX) = A\,\text{Var}(y)\,A^\top = \sigma^2\mathbf{A}\mathbf{A}^\top$<br><br>
  $\mathbf{A}\mathbf{A}^\top = \sigma^2 \left[(\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top + \mathbf{D}\right]\left[\mathbf{X}(\mathbf{X}^\top\mathbf{X})^{-1} + \mathbf{D}^\top\right]$<br><br>
  Поскольку DX = 0, два слагаемых посередине сокращаются и получаем выражение для дисперсии:<br>
  $= (\mathbf{X}^\top\mathbf{X})^{-1} + \mathbf{0} + \mathbf{0} + \mathbf{D}\mathbf{D}^\top$<br><Br>
  Итого:<br>
  $\text{Var}(\tilde{\boldsymbol{\beta}}) = \sigma^2(\mathbf{X}^\top\mathbf{X})^{-1} + \sigma^2\mathbf{D}\mathbf{D}^\top = \text{Var}(\hat{\boldsymbol{\beta}}) + \sigma^2\mathbf{D}\mathbf{D}^\top$

- <span style="color:darkblue">Надо показать, что $\text{Var}(\tilde{\boldsymbol{\beta}}) - \text{Var}(\hat{\boldsymbol{\beta}}) \succeq \mathbf{0}$</span><br>
  Поскольку у нас оценки многомерные $\boldsymbol{\beta} = (\beta_0 ... \beta_n)$, мы сравниваем не просто скалярные дисперсии $\hat{\beta}$ и $\tilde{\beta} $, а Covariance матрицы. Поэлементное сравнение не прокатит. Единственный норм способ сранвить матрицами - это проверить, что разница этих матриц positive semidefinite<br>
  - тут отсылка к Алгебре: матрица $A$ положительно полуопределена, если любая квадратичная норма $c^TAc \ge 0$
  - конус
  - в нашем случае матрица - это матрица ковариаций Cov, поэтому можно так записать $Cov \ge Cov$
  - геометрически требование можно проиллюстрировать так: эллипс разброса $\tilde{\beta}$ должен быть полностью вписанным в любой другой эллипс любой другой оценки $\beta$
  - нужно убедиться, что дисперсия любой линейной комбинации $\tilde{\beta}$ больше дисперсии динейной комбинации $\hat{\beta}$
  
  $\text{Var}(\tilde{\boldsymbol{\beta}}) - \text{Var}(\hat{\boldsymbol{\beta}}) \succeq \mathbf{0} \iff \forall \mathbf{c}: \mathbf{c}^\top\left[\text{Var}(\tilde{\boldsymbol{\beta}}) - \text{Var}(\hat{\boldsymbol{\beta}})\right]\mathbf{c} \geq 0$<br><br>
  Дисперсия становится<br>
  $\text{Var}(\mathbf{c}^\top\tilde{\boldsymbol{\beta}}) = \mathbf{c}^\top\text{Var}(\tilde{\boldsymbol{\beta}})\mathbf{c} = \mathbf{c}^\top\text{Var}(\hat{\boldsymbol{\beta}})\mathbf{c} + \sigma^2\mathbf{c}^\top\mathbf{D}\mathbf{D}^\top\mathbf{c}$<br>

  Поскольку $\sigma^2\|\mathbf{D}^\top\mathbf{c}\|^2 \geq 0$ квадратичная норма, выполняется всегда<br>
  $\text{Var}(\mathbf{c}^\top\tilde{\boldsymbol{\beta}}) = \text{Var}(\mathbf{c}^\top\hat{\boldsymbol{\beta}}) + \underbrace{\sigma^2\|\mathbf{D}^\top\mathbf{c}\|^2}_{\geq\, 0} \geq \text{Var}(\mathbf{c}^\top\hat{\boldsymbol{\beta}})$<br>
  
  Значит $D = 0$, что соотвествует $\tilde{\boldsymbol{\beta}} = \hat{\boldsymbol{\beta}}$. $\blacksquare$

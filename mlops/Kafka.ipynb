__Kafka__ - брокер сообщений, предназначенный для обеспечения асинхронной коммуникации между микросервисами / компонентами распределенного приложения. Появилась в 2010 году в Linkedin. В 2012 году попала в пул проектов Apache

<img src="img/kafka2.png" width=750>

Утверждение Kafka - это не совсем Message Processing, поскольку в отличие от многих MP систем, Kafka не удаляет прочитанные сообщения из очереди, что делает ситему ближе к классу KV-сторадж.

Работая с Kafka, работают с кластером. Кластер = набор Kafka нод.

__Топик__ - основная единица данных, аналог таблицы в СУБД или очереди в MQ системах

__Брокер__ = нода в кластере

Kafka - распределенная система, поэтому каждый топик распределен по брокерам кластера, единица такого распрделения - партиция. Каждая партиция заливается на свой кластер.

<img src="img/kafka.png" width=1000>

Запись и чтение всегда происходят через мастер-ноду партиции (leader), репликация на остальные нодцы выполняется асинхронно, соотвественно с некоторым лагом. Что представляет определенную сложность, когда процесс падает.

Структура данных = Append Log 

Данные хранятся ограниченное время, оно настраивается<br>
earlisest offset - первое доступное сообщение

Чтение аснихронно, соотвественно возникает вопрос, как гарантировать консистетнтонсть?

Так же как при любой потоковой обработке - запоминаем последнюю обработанную запись. Offset - номер записи, по которую все прочитали и обработали. Он существует для каждого потребителя.

Когда что-то ломается, могут быть ошибки. Варианты сохранения офсетов:
1. брокер комитит offset сразу после чтения<br>
не обеспечивается семантика - не понятно, успел потребитель обработать отданные записи или нет
2. брокер комитит по просьбе потрбителя <br>
так обеспечивается at least once семантика - могут быть дубликаты, если потребитель  не успел закомитить чтение
3. сохранять офсетвы на стороне потребителя <br>
самая строгая семантика exactly once

Inverntory - список хостов, сгруппированный по блокам

Все файлы формата yaml (.yml)

Можно запускать отдельные команды: 
ansible copy

Можно запустить на выбранном хосте<br>
Можно запустить на всех хостах<br>
Можно запустить на группе хостов<br>

Можно запускать более сложные функции (модули). На сайте есть описание 300 готовых модулей<br>
Ansible -m

Playbook - последовательность тасков<br>
имя<br>
Параметры

Роли - способ группировки тасков по функционалу

Пример из реального проекта (DMC обработка данных GorodRabot) 

Здесь site.yml - точка входа<br>
Внутри просто вызываются 2 роли

```
- name: deploy ml environment
 hosts: hadoop-oozie-client-hosts
 serial: 1
 roles:
   - { role: cleverdata-dmc-ml-python-venv, tags: 'ml-env' }
```

```
- name: deploy job
 hosts: hadoop-oozie-client-hosts
 serial: 1
 roles:
   - { role: cleverdata-dmpkit-oozie-job, tags: 'job' }
```

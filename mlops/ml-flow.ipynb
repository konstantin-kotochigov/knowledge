#### Tracking

Tracking = логирование параметров каждого запуска обучения модели

Имеет смысл, когда обучение долго выполняется (на сутки например). Тогда действительно важно ничего не пропустить и не дублировать запуски.

GridSearchCV делает ровно то же самое.

Интересно, можно ли использовать для логирования регулярных обновлений модели? По кол-ву кода по-моему тоже самое, что руками в файл или БД записывать.



#### Project
Project = архив-обертка с прописанными зависимостями окружения, bash-командой запуска питоновского кода и перечислением параметров, которые в эту команду должны передаваться. Все это прописывается в конфигурационном файле MLProject.

Запускается архив командой mlflow run x=1 y=2. При запуске стартует новое conda окружение.



#### Deploy
Deploy = разворачивает запускаемый проект в качестве веб-сервиса на mlflow сервере

mlflow models serve project/runnum/artifacts/model/ --host --port

Сервис можно протестировать curl-ом. В запросе передается строка с данными (или любой другой кейс для прогноза).

Можно сразу создать докер контейнер
mlflow models build-docker

и запускать его (с параметрами что ли?)
docker run …



Tagging runs = вообще какая-то хрень. Можно запускам назначать тэги. Ну заебись.

Storing results into DB (postgres) = можно записывтать. Достаточно настроить mlflow
mlflow server --baskend-store-uti postgres-connection

What challenges commonly arise while developing an ML model:
- it is hard to track results if there are many experiments
- it is hard to run code
- it is hard to maintain universal format
- it is hard to store

What is a typical case when MLFlow is useful?
There is a model that is used in production and requires multiple data scientists or even teams to continuously develop it.

Where run data can be stored?
- Locally (default ./mlruns)
- To any SQLALchemy-compatible database
- To HTTP-server

Tracking
Performance metrics can be tracked with `log()` functions. The logged metrics are automatically visualized in dashboard (mlflow UI).

Autologging does everything siltently. For example GridSearchCV logs cross-validated scores for each parameter set. Autologging option is available for most platforms (scikit-learn, keras, pytorch, tensorflow, spark, xgboost and others).

Storage types
- Backend storage (metadata)
- file
- database
- Artifact storage (data)

Projects
Logically an mlflow project is a package that contains everything to run it. Physically it is a directory with a central configuration file called MLProject.

Project can contain multiple entry points (functions to run). Default entry point is "main".
Each entry point contains:
run parameters
command

To run project:<br>
`mlflow run`

#### Models
All models are stored in a standard mlflow format. Each model has its own directory. Each directory contains MLFlow file - the model configuration.

Apart from necessary metadata each model can contain:
- Model signature 
- a simple JSON representing input & output parameters
- input example
- sample dataset with of a proper format

When logging trained models it is better to add signature and input example.

Models can be logged through
Model class
log()
save()
load()

Tool specific functions
mlflow.sklearn.log_model()
mlflow.sklearn.save_model()
mlflow.sklearn.load_model()

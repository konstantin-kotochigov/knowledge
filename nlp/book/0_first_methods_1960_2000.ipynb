# Pre Deep Learning Era (1950-2000)

## Две парадигмы

__Зеллиг Харрис__ в работе [«Distributional Structure»](https://www.tandfonline.com/doi/pdf/10.1080/00437956.1954.11659520) (1954) предложил описывать язык дистрибутивно — через окружение, в котором встречаются его единицы (слова, фонемы и т.д.), не апеллируя к значению напрямую. Если два слова систематически появляются в похожих контекстах, их значения близки. Его пример: *oculist* и *eye-doctor* встречаются почти в одинаковых окружениях — и означают почти одно и то же. 

Чуть позже __Джон Руперт Фёрс__ сжал ту же мысль в афоризм, который часто цитируют в NLP: «You shall know a word by the company it keeps» (Firth, 1957)

В те же 1950-е аспирант Харриса __Ноам Хомский__ развивал противоположный подход — [генеративную грамматику](https://ru.wikipedia.org/wiki/%D0%9F%D0%BE%D1%80%D0%BE%D0%B6%D0%B4%D0%B0%D1%8E%D1%89%D0%B0%D1%8F_%D0%B3%D1%80%D0%B0%D0%BC%D0%BC%D0%B0%D1%82%D0%B8%D0%BA%D0%B0). Его возражение: носитель свободно строит и понимает предложения, которых в корпусе никогда не было. Фраза «Colorless green ideas sleep furiously» статистически невозможна, но ощущается грамматически корректной — а значит, за языком стоит жесткая система порождающих правил, которую из одной сочетаемости не вывести

Различие позиций можно также описать тем, как строится модель языка: можно идти "сверху вниз" (Хомски): сначала постулировать систему правил (исходя из значений, логики, традиции) и затем накладывать её на материал. А можно "снизу вверх" (Харрис): не вводить категорий заранее, а позволить им вырасти из наблюдаемой сочетаемости

На несколько десятилетий научное сообщество предпочло подход Chomsky: он казался более перспективным, плюс для дистрибутивного подхода требовалось много вычислительных ресурсов, которых не хватало. И дистрибутивный подход отошел на обочину; ранний NLP во многом строился на созданных экспертами грамматиках и правилах. Реванш случился в 1990-е: считать сочетаемость стало проще и устойчивее на реальных текстах. Победил не более «правильный» взгляд на язык, а более инженерный

Победивший "дистрибутивный" подход часто называют статистическим, но стоит отметить, что полноценно статистическим он стал только начиная с 90-х. Сам Харрис частот не считал — он работал в режиме «возможно / невозможно»: пересекаются ли множества контекстов двух единиц. Это был скорее качественный анализ. Но именно переход от сочетаемости к конкретным частотам и превратил программу Харриса в рабочий инструмент

Дистрибутивняый взгляд на язык нашел отражение в большинстве современных NLP методов. [Bag-of-words](https://en.wikipedia.org/wiki/Bag-of-words_model) и [TF-IDF](https://en.wikipedia.org/wiki/Tf%E2%80%93idf) представляют документ через распределение слов в нём. N-граммные модели оценивают вер-ть слова по его контексту. Скрытые марковские модели размечают слово, опираясь на соседей. Латентно-семантический анализ и [кластеризация Брауна](https://en.wikipedia.org/wiki/Brown_clustering) прямо строят представления слов из матрицы «слово — контекст». Все они ничего не постулируют о грамматике заранее, а выводят обобщение из примеров — это Харрис, доведённый до машинного масштаба. Наконец в 2010-х появляется модель [word2vec](https://en.wikipedia.org/wiki/Word2vec), генерирующая контекстные эмбеддинги, и она наиболее прямо воплотила дистрибутивную гипотезу Харриса, пролежавшую к тому моменту шестьдесят лет.

Источники:

- Harris, Z. S. (1954). *Distributional Structure*. WORD, 10(2–3), 146–162. https://doi.org/10.1080/00437956.1954.11659520
- Firth, J. R. (1957). *A Synopsis of Linguistic Theory 1930–1955*. В сборнике Studies in Linguistic Analysis.
- Современный разбор: Sahlgren, M. (2008). *The Distributional Hypothesis*. Italian Journal of Linguistics, 20(1).

## Классические задачи

Исследования в этой области не были чисто академическими, была также потребность адаптировать их для решения практических задач. К классическим задачам, решаемым уже начиная с 1960 годов можно отнести:

- **Распознавание речи.**<br>Главный заказчик n-граммных языковых моделей. Акустический сигнал неоднозначен, и модель языка выбирала из похоже звучащих вариантов более правдоподобный («recognize speech» против «wreck a nice beach»). Отсюда же острый интерес к сглаживанию и перплексии
- **Машинный перевод.**<br>Статистический MT (модели IBM, 1990-е) разбивал перевод на модель перевода плюс языковую модель целевого языка, которая следила за гладкостью результата.
- **Информационный поиск.**<br>Векторная модель на TF-IDF: ранжирование документов по релевантности запросу. Прямой предок поисковых систем.
- **Классификация и фильтрация текста.**<br>Спам-фильтры, тематическая рубрикация новостей, ранний анализ тональности — обычно мешок слов плюс наивный Байес или SVM.
- **Разметка последовательностей.**<br>POS-теггинг (части речи), выделение именованных сущностей (NER), сегментация — задачи, где каждому токену нужно приписать метку; вотчина скрытых марковских моделей.
- **Исправление опечаток и предсказание ввода**<br>там же, где и распознавание речи, работала языковая модель, выбиравшая вероятное продолжение или исправление

## Первые языковые модели

Языковая модель в самом общем смысле — это распределение вероятностей над последовательностями слов: она должна уметь для любой цепочки слов сказать, насколько та правдоподобна. 
$$P(w)=P(w_1,w_2... w_n)$$

Эквивалентно, языковая модель предсказывает следующее слово по предыдущим. Это в точности та же задача, которую решают современные GPT: по предыдущему контексту оценить вероятное продолжение текста. Менялись только методы оценки этого распределения:
$$P(w_n|w_1 ... w_{n-1})$$

__Андрей Марков (1913)__ заложил математический фундамент. Анализируя чередование гласных и согласных в «Евгении Онегине», он показал, что текст можно описать как стохастический процесс, где вероятность следующего символа зависит от предыдущего (после гласной более вероятна согласная и т.п.). Так появились [цепи Маркова](https://en.wikipedia.org/wiki/Markov_chain) — формальная основа всех последующих моделей

__Клод Шеннон (1948)__, в работе [«A Mathematical Theory of Communication»](https://people.math.harvard.edu/~ctm/home/text/others/shannon/entropy/entropy.pdf), первым явно смоделировал язык вероятностно. Он строил приближения английского текста разного порядка (n-граммы), ввёл понятие энтропии языка и фактически описал теорию, на которой стоит вся область

Сам термин «языковая модель» в современном смысле закрепился позже — в 1970–80-х годах в сообществе распознавания речи, прежде всего в группе __Фреда Елинека__ (Fred Jelinek) в IBM. Там использовалась модель «шумного канала»: P(слова | звук) ∝ P(звук | слова) · P(слова), и вот эта априорная компонента P(слова) — оценка того, насколько правдоподобна сама фраза — и называлась language model. Елинеку приписывают полушутливую фразу о том, что качество распознавания растёт каждый раз, когда из команды уходит лингвист — это маркер перехода от правил Хомского к статистике.

Вероятность последовательности $w = (w_1...w_n)$ раскладывается в произведение условных вероятностей:

$$P(w_1, w_2, \dots, w_n) = P(w_i|w_{1..i-1}) \cdot P(w_2|w_{1..i-2}) = ... = \prod_{i=1}^{n} P(w_i \mid w_1, \dots, w_{i-1})$$

Проблема в том, что вероятность $P(w_i \mid w_1, \dots, w_{i-1})$ зависит от всей предыстории $w_1 ... w_{i-1}$, и таких предысторий бесконечно много — оценить их по данным невозможно. Здесь и вступает [марковское допущение](https://en.wikipedia.org/wiki/Markov_property): будем считать, что следующее слово зависит не от всей истории, а только от последних $n-1$ слов. Это и есть [n-граммная модель](https://en.wikipedia.org/wiki/Word_n-gram_language_model)

При $n = 2$ (биграммы) каждое слово зависит только от предыдущего: $P(w_i \mid w_1, \dots, w_{i-1}) \approx P(w_i \mid w_{i-1})$<br>
При $n = 3$ (триграммы) — от двух предыдущих, и так далее<Br>
При $n = 1$ мы получаем вырожденную модель, где слова считаются независимыми. Это сильное упрощение и на практике от такой модели будет мало пользы

Сами вероятности оцениваются простым подсчётом частот в обучающем корпусе (оценка максимального правдоподобия):

$$P(w_i \mid w_{i-1}) = \frac{C(w_{i-1}, w_i)}{C(w_{i-1})}$$

где $C(\cdot)$ — число вхождений соответствующей последовательности в корпус. То есть «вероятность слова *молоко* после слова *парное*» — это сколько раз встретилось «парное молоко», делённое на то, сколько раз встретилось «парное» вообще.

Плюсы n-gram моделей: 
- прозрачность: никакого обучения в современном смысле, только счётчики. На триграммах такие модели десятилетиями оставались рабочей лошадкой распознавания речи и машинного перевода и были на удивление сильны — настолько, что нейросетевым языковым моделям потребовались годы, чтобы их уверенно превзойти

Но есть и фундаментальные ограничения: 
- в каждый момент времени модель видит только узкое окно в $n-1$ слов. Это сильно ограничивает ее "контекстность"<br>Наивное решение - увеличить $n$, но тогда мы довольно быстро упираемся в потолок комбинаций, которые надо хранить (см следующий раздел)

Стоит отметить, что существуют и другие способы моделирования языка: например, диффузионные подходы, energy-based и прочее. Все они имеют практические воплощения, однако настолько широкую популярность не получили

## Curse of sparsity
Рассмотрим языковую модель $P(w_n|w_1, w_2 ... w_{n-1})$. При n >> 3 большинство грамматически правильных триграмм в любом конечном корпусе не встречаются ни разу и по формуле максимального правдоподобия получает вероятность ровно ноль. Это плохо: стоит в тестовом предложении появиться одной неизвестной триграмме, и вероятность всего предложения обнуляется

Для борьбы используют сглаживание (smoothing): часть вероятностной массы отбирается у наблюдённых событий и перераспределяется в пользу ненаблюдённых, чтобы нулей не осталось. Простейший вариант — сглаживание Лапласа (add-one): прибавить единицу ко всем счётчикам. На практике оно работает плохо — слишком грубая трансформация

За время развития направления было предложено целое семейство методов. Некоторые из них:
- **Сглаживание Гуда–Тьюринга** (Good, 1953)<br>оценивает вероятность ненаблюдённых событий через число событий, встретившихся ровно один раз; одна из первых принципиальных идей в этой области<br>$$c^* = (c + 1) \frac{N_{c+1}}{N_c}$$где с - исходная частота, $N_{c}$ и $N_{c+1}$ - количество уникальных событий, которые встретились в тексте ровно $c$ и $c+1$ раз<br><br>
- **Откат Каца** (Katz back-off, 1987) <br>если триграмма новая (частота в словаре 0), то «откатываемся» к биграмме, если и та нет — к униграмме<br><br>
- **Интерполяция Йелинека–Мерсера** <br>линейно смешивает оценки разных порядков (триграмму, биграмму, униграмму) с подобранными весами<br><br>
- **Сглаживание Кнезера–Нея** (Kneser & Ney, 1995) — вершина этой линии. Ключевая идея тонкая и красивая: вероятность слова в роли «отката» оценивается не по тому, как часто оно встречается вообще, а по тому, в скольких *разных* контекстах оно появляется. Классический пример — слово *Angeles*: оно частотно, но почти всегда идёт после *Los*, поэтому как самостоятельное продолжение произвольного контекста маловероятно. Его интерполированная (модифицированная) версия многие годы считалась state-of-the-art для дискретных n-граммных моделей.

Канонический обзор всего этого зоопарка — большое эмпирическое исследование Чена и Гудмана, где методы аккуратно сравниваются между собой и предлагается улучшенный вариант Кнезера–Нея:

- Chen, S. F. & Goodman, J. (1998/1999). *An Empirical Study of Smoothing Techniques for Language Modeling*. Harvard TR-10-98 / Computer Speech & Language 13(4), 359–394. https://dash.harvard.edu/handle/1/25104739
- Goodman, J. (2001). *A Bit of Progress in Language Modeling*. Computer Speech & Language 15(4), 403–434. https://arxiv.org/abs/cs/0108005 — дружелюбное продолжение, хорошая отправная точка для чтения.

Если вы запомните из этого раздела одну вещь, пусть это будет следующая: вся изобретательность статистической эпохи ушла не на саму модель (она тривиальна — это счётчики), а на борьбу с нулями. Современные нейросетевые модели решают ту же проблему обобщения на невиданные сочетания принципиально иначе — через плотные векторные представления, где близкие по смыслу слова автоматически делятся статистикой.

## Perplexity

Раз уж мы строим распределения над текстами, нужна метрика, которая говорит, насколько "хороша" модель. Такая метрика есть — называется **перплексия** ([perplexity](https://en.wikipedia.org/wiki/Perplexity)) 

Метрику впервые предложили использовать в 1977 году в IBM в проектах по распознаванию речи. Идея в том, что она показывает, насколько модель «удивлена» (perplexed) реальным текстом: если наблюдаемый токен имеет в рамках модели низкую вер-ть быть сгенерированным, то удивление большое.

Возьмем отдельный токен $w_i$ из какого-либо предложения тестового корпуса текстов. Подадим в языковую модель его предыдущий контекст $w_1...w_{i-1}$ и посмотрим, какую P(w) возвращает модель. Если вер-ть генерации правильного токена $w_i$ высока, то все хорошо, модель справляется. 

Если к этой вероятности применить отрицательный логарифм, то она трансформирует задачу максимизации в задачу минимизации:
$$-\ln P(w_i \mid w_1, \dots, w_{i-1})$$

Теперь вероятность близкая к 1 означает низкую перплексию, а к 0 высокую:

<img src="img/perplexity.png" width=500>

Таким образом, значение метрики зависит от того: 
1) насколько хорошо модель научилась предсказывать возможные продолжения в тексте
2) что у нас за текст и насколько он нестандартный
  
Для получения средней перплексии на множестве документов метрика усредняется. По всем токенам данного предложения, по всем предложениям в корпусе документов
$$L = -\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})$$

Получаемая величина (Average negative log-likelihood per token), это по сути и есть функция потерь (Loss function) оптимизруемая при обучении LLM. 

Она же кросс-энтропия $H(P^{model}, P^{true}) = \sum P^{true} \cdot \log P^{model}$, при этом $P^{true}$ вырождается в константу "1.0" и получается привычная форма

<img src="img/perplexity2.png" width=600>

Чтобы перевести значение метрики на более интрепретируемую шкалу от этого выражения еще берут экспоненту:

$$\text{PP}(W) = e^{L} = \exp\!\left(-\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})\right)$$

где $N$ — число слов в тесте

Можно $-1/N$ занести под логарифм, тогда экспонента сократится, а вероятности перемножатся в полную вероятность

$$\text{PP}(W) = P(w_1, \dots, w_N)^{-\frac{1}{N}}$$

На этой шкале у метрики есть интерпретация. Перплексия — это число равновероятных вариантов, между которыми модель в среднем выбирает на каждом шаге генерации. Перплексия 100 означает, что модель в среднем колеблется так, как если бы на каждом шаге угадывала равновероятный исход из ста. Перплексия, равная размеру словаря, соответствует модели, которая не выучила вообще ничего (равномерное распределение). Чем ниже — тем лучше: модель уверенно сужает выбор.

Перплексия обычно падает к концу предложения, так как в конце накоплен конекста больше
<img src="img/perplexity_1.png" width=500>

Несколько важных оговорок, без которых перплексией легко злоупотребить:

- Перплексия зависит от словаря. Сравнивать перплексии двух моделей корректно только при одинаковом словаре и одинаковой разбивке на токены<br><br>
- Перплексия - <u>внутренняя</u> (intrinsic) метрика модели. Она говорит, как хорошо модель моделирует распределение текста, но не говорит, как хорошо она решает прикладную задачу (перевод, распознавание речи). Важно смотреть и на внешние (extrinsic) метрики — например, частоту ошибок распознавания слов (WER) в речевых системах<br><br>
- Перплексия тесно связана с кросс-энтропией: $\text{PP} = 2^{H}$ (или $e^H$, смотря какое основание логарифма), где $H$ — кросс-энтропия в битах (натах) на слово. Поэтому в статьях того времени качество часто отчитывали именно в битах на слово

Перплексию удобно испольовать для детекции сгенерированного текста. У ИИ тексты стабильно низкой перплексией, в то время как перпексия более взрывная

## Document representation

Языковые модели — это про порядок слов и предсказание следующего. Но огромный пласт задач (классификация текстов, информационный поиск, фильтрация спама) порядок слов как раз игнорирует и смотрит только на то, *какие* слова встретились. Базовое представление здесь — **мешок слов** (bag-of-words, BoW): документ описывается вектором по словарю, где каждая компонента — сколько раз соответствующее слово встретилось в документе. Порядок полностью отбрасывается, остаётся только частотный профиль. «Кошка укусила собаку» и «собака укусила кошку» становятся одним и тем же вектором — и для многих задач этого, как ни странно, достаточно.

Сырые частоты, однако, обманчивы: самые частые слова («и», «в», «не») встречаются повсюду и почти ничего не говорят о содержании документа. Отсюда взвешивание **TF-IDF** (term frequency — inverse document frequency). Идея в двух множителях:

- **TF** — частота слова в данном документе: чем чаще слово в документе, тем оно для него важнее.
- **IDF** — обратная документная частота: чем в большем числе документов коллекции слово встречается, тем меньше его вес. Обычно $\text{idf}(t) = \log \frac{N}{df(t)}$, где $N$ — число документов, $df(t)$ — число документов со словом $t$.

Произведение $\text{tf} \cdot \text{idf}$ даёт высокий вес словам, которые часты в конкретном документе, но редки в коллекции в целом, — то есть характерным, различающим словам. На таких векторах строился весь классический информационный поиск: запрос и документы представлялись векторами, релевантность измерялась косинусной близостью между ними (векторная модель поиска, восходящая к работам Солтона). TF-IDF до сих пор остаётся крепкой базовой линией для поиска и классификации — настолько крепкой, что его не стыдно попробовать первым делом перед любыми нейросетями.

Стоит заметить связь с дистрибутивной гипотезой: если мешок слов строить не для документа, а для контекстного окна вокруг слова, и сложить такие профили по всему корпусу, получится представление слова через его окружение. Это прямой предок матричных дистрибутивных моделей (LSA и подобных), которые мы упомянем в следующей главе.

## HMM (Hidden Markov Models)

Не все задачи — это классификация целого документа. Многие требуют разметить каждый токен: приписать каждому слову часть речи (POS-tagging), выделить именованные сущности (NER), сегментировать речь. Доминирующим инструментом для таких задач последовательной разметки в эту эпоху была **скрытая марковская модель** (Hidden Markov Model, HMM).

Есть матрица переходов между токенами. Есть переменная (скрытая, скаляр, дискретная: категории описаны), которая влияет на матрицу переходов. 

Идея: наблюдаемая последовательность (слова) порождается ненаблюдаемой («скрытой») последовательностью состояний (например, частей речи). Модель описывается двумя наборами вероятностей:

- Вероятности переходов — насколько вероятно, что за состоянием «прилагательное» идёт состояние «существительное». Здесь снова марковское допущение: следующее состояние зависит только от текущего.
- Вероятности эмиссии — насколько вероятно, что состояние «существительное» порождает конкретное слово «стол».

<img src="img/hmm1.png" width=550>

Задача разметки — найти наиболее вероятную последовательность скрытых состояний для данной последовательности слов. Решается она эффективно динамическим программированием — [алгоритмом Витерби](https://en.wikipedia.org/wiki/Viterbi_algorithm). Параметры оцениваются либо по размеченному корпусу (простым подсчётом, как у n-грамм), либо без разметки — алгоритмом [Баума–Велча](https://en.wikipedia.org/wiki/Baum%E2%80%93Welch_algorithm) (частный случай EM).

HMM хорошо работали для POS-теггинга и были естественным выбором, пока их в начале 2000-х не стали теснить дискриминативные модели — прежде всего условные случайные поля __Conditional Random Fields__ [(Lafferty et al, 2001](https://daiwk.github.io/assets/Conditional%20random%20fields%20Probabilistic%20models%20for%20segmenting%20and%20labeling%20sequence%20data.pdf)). CRF снимали несколько структурных ограничений HMM (в частности, позволяли свободно использовать произвольные, в том числе пересекающиеся, признаки) и стали стандартом для разметки последовательностей на следующее десятилетие — вплоть до прихода нейросетевых моделей. Но логика «скрытые состояния порождают наблюдения, ищем лучшую цепочку Витерби» осталась важной частью образования любого, кто работает с последовательностями.

## SVM for classification

Типичный конвейер классификации текста в конце 1990-х выглядел так: текст превращается в вектор признаков (мешок слов или TF-IDF), а поверх этих векторов обучается классификатор.

Королём классификаторов в эту эпоху была **машина опорных векторов** (Support Vector Machine, SVM). Причины её популярности именно в текстовых задачах:
- Тексты дают чрезвычайно высокоразмерные и разреженные векторы, а SVM с этим справляется отлично: она ищет разделяющую гиперплоскость с максимальным зазором и хорошо обобщает даже когда признаков больше, чем примеров.
- Линейное ядро на текстах работает прекрасно — текстовые классы часто линейно разделимы в пространстве мешка слов, так что сложные ядра обычно и не нужны.

Каноническая работа, показавшая, что SVM — почти идеальный инструмент для классификации текстов «Text Categorization with Support Vector Machines» [(Joachims, 1998)](https://www.cs.cornell.edu/people/tj/publications/joachims_98a.pdf). Связка «TF-IDF + линейный SVM» стала на годы стандартным рецептом для спам-фильтров, тематической классификации, анализа тональности. Из той же эпохи и наивный байесовский классификатор — ещё проще, ещё быстрее, удивительно конкурентоспособен как базовая линия.

Ключевое, что отличает всю эту эпоху от современной, — признаки придумывал человек. Качество системы определялось не столько алгоритмом обучения, сколько изобретательностью инженера в feature engineering: какие n-граммы взять, как нормализовать, какие лингвистические маркеры добавить. Революция представлений (эмбеддинги, а затем end-to-end обучение) состояла именно в том, что конструирование признаков отдали самой модели. Но интуиция «хорошие признаки решают больше, чем хитрый классификатор» полезна и сегодня.

## Другие важные методы

Чтобы картина была честной, упомянем несколько вещей, которые тоже относятся к этому периоду и пригодятся дальше:

- **Латентно-семантический анализ (LSA / LSI)** (Deerwester и др., 1990; Landauer & Dumais, 1997) — попытка реализовать дистрибутивную гипотезу через сингулярное разложение (SVD) матрицы «слова × документы». LSA — это прямой мост от TF-IDF к плотным векторным представлениям слов и идейный предшественник эмбеддингов; о нём подробнее в следующей главе.
- **Статистический машинный перевод** (модели IBM, Brown и др., 1993) — отдельная большая ветвь, где n-граммные языковые модели играли центральную роль (отсюда, кстати, особый интерес к сглаживанию Кнезера–Нея).
- **Кластеризация слов по Брауну** (Brown и др., 1992) — иерархическая кластеризация слов по контекстам; ещё одна ранняя дистрибутивная идея, которая давала «дешёвые» классы слов как признаки и продержалась в продакшене очень долго.
- Понятия **точности, полноты и F-меры** — внешние метрики для классификации и разметки, которые тоже сложились в эту эпоху и остаются стандартом.

## Первая нейросетевая модель

В 2003 году Josua Benjio презентовал работу "A Neural Probabilistic Language Model" [(Benjio et al, 2003)](https://jmlr.org/papers/volume3/tmp/bengio03a.pdf) - с этого момента языковое моделирование стало нейросетевым

N-граммные модели страдали от проклятия размерности:
- Во-первых, число возможных сочетаний слов астрономическое (`|V|ⁿ`), поэтому большинство реальных n-грамм просто ни разу не встречается в обучающих данных и получает нулевую вероятность: приходилось делать сглаживание и откаты (backoff)
- Во-вторых, слова в таких моделях - атомарные дискретные символы без семантики («кот» и «собака» для них различаются ровно так же, как «кот» и «пылесос», просто разные индексы в словаре). Из-за этого модель не умеет нормально обобщать

Бенжио предложил сопоставить каждому слову словаря обучаемый вещественный вектор признаков (в их экспериментах — порядка 30–100 измерений), то есть распределённое представление. И дальше — выразить вероятность последовательности через эти векторы, причём векторы и саму вероятностную функцию учить совместно, одной нейросетью.

<img src="img/benjio.png" width=400>

Это мгновенно дало обобщение: если в ходе обучения «кот» и «собака» получают близкие векторы, то модель, видевшая «кот сидел на полу», автоматически назначит разумную вероятность фразе «собака сидела на полу», даже если её в данных не было. Похожие слова → близкие векторы → переносимое знание. Вот так непрерывное пространство признаков лечит проклятие размерности.

__Архитектура__<br>
Обычная полносвязная сеть (MLP) с фиксированным окном контекста. На схеме пример для n=3

Представления слов из n-грамма конкатенирутся. Есть также опцииональный residual connection мимо основного tanh преобразования ($Wx$):

$$P(wₜ | \text{контекст}) = \text{softmax}(b + W·x + U·tanh(d + H·x))\text{, где } x = [C(w_{t-3}), C(w_{t-2}), C(w_{t-1})]$$

Здесь `C` — матрица эмбеддингов размера `|V|×m`, общая для всех входных позиций; `H` и `U` — веса скрытого и выходного слоёв; 

Модель авторегрессионная и генеративная. На каждом шаге генерации на вход подается сгенерированные к этому моменту окно текста. На выходе к вектору размера |V| применяется softmax для получения вероятностей . На обучении оптимизируется лог-правдоподобие наблюдаемых данных. Позиция слова никак не учитывается.

На корпусах Brown и AP News нейросестевая модель по перплексии стала лучше n-граммных больше чем на 10%. Но главное была доказана идея: обучаемые распределённые представления работают и побеждают подсчёт частот

Ограничения модели:
- softmax по всему словарю на каждом шаге — самая дорогая операция: при словаре в десятки тысяч слов это доминирует в вычислениях<br>Бенжио боролся с этим распараллеливанием, а последующие работы придумали способы обойти полный softmax: иерархический softmax (Морен и Бенжио, 2005), важностную выборку, а позже noise-contrastive estimation и negative sampling — последнее и сделало возможным быстрый word2vec<br><Br>
- главное ограничение языковой модели: фиксированный и небольшой контекст<br>позже это ограничение обошли с помощью рекурсивных сетей RNN, LSTM

Из идеи обучения распределенного представления позднее выросли word2vec (Миколов, 2013), GloVe а также обучаемые эмбеддинги во всех современных трансформерех

## Итог главы

Что важно унести из этой главы перед тем, как двигаться к нейросетям:

Во-первых, задача не изменилась. Языковая модель и в 1995-м, и сегодня — это распределение над последовательностями слов, предсказание следующего слова по контексту. Менялись только средства оценки этого распределения.

Во-вторых, главным врагом была разреженность данных, и почти вся инженерная мысль эпохи ушла на борьбу с нулями — на сглаживание. Это объясняет, почему плотные представления, где близкие слова автоматически делятся статистикой, оказались таким прорывом: они атакуют ровно эту проблему обобщения.

В-третьих, перплексия — мера того, насколько модель «удивляется» реальному тексту, или эффективное число вариантов на каждом шаге, — родилась здесь и останется с нами до конца курса как сквозная мера прогресса языковых моделей.

И, наконец, дистрибутивная гипотеза Харриса и Фёрса — «слово познаётся по его окружению» — была сформулирована за полвека до того, как появились средства её по-настоящему реализовать. Вся дальнейшая история, к которой мы переходим в следующей главе, — это история о том, как эту идею наконец-то удалось воплотить: сначала через матричные методы и эмбеддинги, а потом и через контекстные представления внутри нейросетей.

```python
from transformers import GPT2LMHeadModel, GPT2TokenizerFast
from accelerate import Accelerator

device = Accelerator().device
model_id = "openai-community/gpt2-large"
model = GPT2LMHeadModel.from_pretrained(model_id).to(device)
tokenizer = GPT2TokenizerFast.from_pretrained(model_id)

---

from datasets import load_dataset

test = load_dataset("wikitext", "wikitext-2-raw-v1", split="test")
encodings = tokenizer("\n\n".join(test["text"]), return_tensors="pt")

---

import torch
from tqdm import tqdm

max_length = model.config.n_positions
stride = 512
seq_len = encodings.input_ids.size(1)

nll_sum = 0.0
n_tokens = 0
prev_end_loc = 0


for begin_loc in tqdm(range(0, seq_len, stride)):
    end_loc = min(begin_loc + max_length, seq_len)
    trg_len = end_loc - prev_end_loc  # may be different from stride on last loop
    input_ids = encodings.input_ids[:, begin_loc:end_loc].to(device)
    target_ids = input_ids.clone()
    target_ids[:, :-trg_len] = -100

    # 
    with torch.no_grad():
        outputs = model(input_ids, labels=target_ids)

        # loss is calculated using CrossEntropyLoss which averages over valid labels
        # N.B. the model only calculates loss over trg_len - 1 labels, because it internally shifts the labels
        # to the left by 1.
        neg_log_likelihood = outputs.loss

    # Accumulate the total negative log-likelihood and the total number of tokens
    num_valid_tokens = (target_ids != -100).sum().item()  # number of valid tokens in target_ids
    batch_size = target_ids.size(0)
    num_loss_tokens = num_valid_tokens - batch_size  # subtract batch_size due to internal label shift
    nll_sum += neg_log_likelihood * num_loss_tokens
    n_tokens += num_loss_tokens

    prev_end_loc = end_loc
    if end_loc == seq_len:
        break

avg_nll = nll_sum / n_tokens  # average negative log-likelihood per token
ppl = torch.exp(avg_nll)
```

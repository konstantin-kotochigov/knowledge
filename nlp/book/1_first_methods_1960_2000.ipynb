# Глава 1. Первые модели

## История

Язык — древнейшая и, возможно, самая важная технология человечества. Когда именно он появился, неизвестно и, по-видимому, не станет известно никогда: звучащая речь не оставляет свидетельств. Мы можем судить только по косвенным признакам — строению голосового тракта, появлению генов, ассоциированных с речевой способностью (например, FOXP2), следам символического поведения вроде украшений, погребальных обрядов и наскальной живописи. Скорее всего развитый язык существовал уже 50–100 тысяч лет назад. Некоторые исследователи связывают его возникновение с самим появлением вида Homo Sapiens 200–300 тысяч лет назад.

#### Язык - данные
Почти всю свою историю язык существовал исключительно в устной форме. Письменность появилась лишь около 5 тысяч лет назад. Шумерская клинопись, около 3200 года до н. э. и почти одновременно в Египте, а затем была независимо изобретена ещё как минимум дважды, в Китае и центральной америке. Письменность превратила язык из акустической волны в данные - то, что можно хранить, копировать, сопоставлять и анализировать.

Шумерские гиняные таблички примером клинописи. Иероглифы. Узелковое письмо Майа. Первый алфавит, ставший прародителем всех современных алфавитов, датируется X веком донашей эры и был основой в Финикиийского языка.

В V веке до н. э. индийский грамматист Панини описал санскрит системой из четырёх тысяч коротких правил. Он по сути создал первую формальную грамматику. Греческая традиция, от софистов до Аристотеля, связала язык с логикой: высказывание стало объектом, к которому применимы формальные операции.

Первый документированный случай, когда количественная модель языка решила практическую задачу, относится к IX веку. Арабский учёный Аль-Кинди описал *частотный анализ* - метод взлома шифров простой замены, основанный на том, что буквы естественного языка встречаются с устойчивыми, характерными частотами: достаточно сопоставить распределение символов шифртекста с распределением букв языка. К слову этот принцип использовался потом на протяжении больше тысячи лет. Так в XVI веке с помощью частотного анализа разоблачили заговор Марии Стюарт против королевы Елизаветы. А в годы Второй мировой войны во главе с Аланом Тъюрингом взломали немецкую шифровальную машину Энигма.

#### Язык - случайный процесс
Систематическое приложение теории вероятностей к тексту началось в 1913 году, когда Андрей Андреевич Марков взял первые 20 000 букв «Евгения Онегина» и показал, что чередование гласных и согласных хорошо описывается случайным процессом, в котором распределение следующего символа зависит только от текущего. Так возникло понятие марковской цепи и мысль, что текст можно рассматривать как случайную последовательность с локальными зависимостями. 

К 1930–40-м годам накопились достаточно данных для исследований и Джордж Ципф заметил, что если упорядочить слова корпуса текста по убыванию их частоты, то частота слова оказывается примерно обратно пропорциональна его рангу (номеру в отсортированной последовательности). Что ещё удивительнее. закон Ципфа воспроизводится во всех языках, независимо от жанра текста. Так, например, самая редкая буква русского алфавита "й" встречается в среднем в 33 раза реже самой популярной буквы "о". Иными словами, небольшой набор слов покрывает большую часть любого текста, а «хвост» из редких слов практически бесконечен.

Клод Шеннон занимался исследованием кодирования. В «Математической теории связи» (1948) он предложил рассматривать источник сообщений — в том числе человека, порождающего текст, как случайный процесс и ввёл числовую меру его непредсказуемости, энтропию.

Приближая английский текст моделями возрастающего порядка — от независимых букв к парам, тройкам и целым словам, — Шеннон фактически ввёл n-граммные модели языка, а в опытах с людьми оценил энтропию английского примерно в один бит на букву, впятеро меньше, чем $\log_2 27 \approx 4{,}75$ бита для равновероятных символов. Язык, стало быть, глубоко избыточен, и именно эту избыточность — предсказуемость продолжения — и улавливает языковая модель. 

Из теории Шеннона следовал принципиальный вывод: чтобы эффективно сжимать и передавать сообщения, нужно уметь предсказывать их продолжение, то есть располагать вероятностной моделью источника; хорошая языковая модель и хороший архиватор — математически одна и та же задача. Не случайно и криптоанализ Второй мировой войны, от Блетчли-парка до работ самого Шеннона по секретной связи, держался на статистических свойствах языка.

#### Язык - задача для машины
Появление компьютеров дало толчок развитию языкового моделирования. В 1949 году Уоррен Уивер разослал меморандум, в котором предложил смотреть на перевод как на задачу дешифровки: русский текст закодирован английским и наоборот. Это в свою очеедь значит, что машина в принципе способна его раскодировать. В 1950 году Алан Тьюринг предложил считать критерием машинного интеллекта способность вести неотличимый от человеческого разговор - так язык оказался в самом определении искусственного интеллекта. А в январе 1954 года состоялся Джорджтаунский эксперимент: компьютер IBM 701 перевёл с русского на английский около шестидесяти тщательно подобранных предложений, и организаторы пообещали решить проблему машинного перевода за три–пять лет. Обещание не сбылось, на самом деле потребовалось больше 60 лет, но именно тогда родилась дисциплина, которую сегодня называют обработкой естественного языка (natural language processing, NLP)

#### Язык - набор правил
Ноам Хомский («Синтаксические структуры», 1957) построил теорию формальных грамматик и их иерархию — от регулярных языков к контекстно-свободным и далее — и решительно выступил против статистического взгляда: его знаменитый пример «Colorless green ideas sleep furiously» грамматически безупречен, хотя имеет нулевую частоту в любом корпусе, а значит, рассуждал Хомский, грамматичность не сводится к вероятности. 

Под влиянием этих идей NLP на два десятилетия стало по преимуществу инженерией правил. Системы вроде ELIZA (1966), пародировавшей психотерапевта, и SHRDLU (1971), понимавшей команды в игрушечном мире кубиков, производили сильное впечатление, но упирались в общую стену: вручную составленных правил всегда оказывалось недостаточно, а их взаимодействие — хрупким. Отчёт комиссии ALPAC (1966), констатировавший, что машинный перевод не оправдал вложений, надолго заморозил финансирование и остудил область.

#### Язык - статистика
Возрождение пришло оттуда же, откуда началась вся история, — из вероятностного мира Маркова и Шеннона. В 1970–80-е годы группа Фредерика Елинека в IBM показала, что распознавание речи лучше всего работает как задача вероятностного вывода, а на рубеже 1990-х тот же аппарат был перенесён на перевод. В канонической модели зашумлённого канала наилучший перевод $\hat{e}$ для входной фразы $f$ ищется как

$$\hat{e} = \arg\max_{e} P(f \mid e)\, P(e),$$

и вторым сомножителем здесь стоит именно языковая модель $P(e)$. Совместная вероятность текста по цепному правилу раскладывается в произведение условных вероятностей, которые на практике усекаются марковским приближением порядка $k$:

$$P(w_1, \dots, w_n) = \prod_{i=1}^{n} P(w_i \mid w_1, \dots, w_{i-1}) \approx \prod_{i=1}^{n} P(w_i \mid w_{i-k}, \dots, w_{i-1}).$$

N-граммные модели, оценённые по большим корпусам, на двадцать лет стали рабочей лошадкой области; появились размеченные корпуса вроде Penn Treebank, а девиз эпохи можно передать так: данные важнее правил.

#### Язык - непрерывные представления
Нейросетевая революция выросла из старой дистрибутивной гипотезы (Зеллиг Харрис, 1954; Джон Фёрс, 1957: «you shall know a word by the company it keeps» — о слове судят по его окружению). Если значение слова определяется контекстами его употребления, то слово можно представить плотным вектором, выученным из этих контекстов. Нейронная языковая модель Бенджио (2003) и векторные представления word2vec (2013) сделали эту идею практичной; рекуррентные сети, архитектура «кодировщик–декодировщик» и механизм внимания (2014) позволили работать с целыми последовательностями; 

В 2017 случилась еще одна революция, радикально изменившая ландшафт языкового моделирования. Появлились Трансформеры, архитектура, которая сделала обучение масштабируемым и дала возможность использовать предобученные модели. 

Интересно, что при всем многообразии методов и моделей, которое мы увидим, в основе лежит все та же классическая постановка, известная еще Маркову и Шеннону. Максимизируется правдоподобие следующего слова:

$$\mathcal{L}(\theta) = \sum_{t} \log p_{\theta}(w_t \mid w_{<t}),$$

просто теперь распределение $p_{\theta}$ задаётся сетью с миллиардами параметров, а обучается на существенной доле всего когда-либо написанного человечеством.

Всю дорогу моделирование языка было одной из прикладных задач, искусстивенный интеллект. Сейчас эти два понятия слились. Языковые модели стали синонимом интеллекта.

Обработка естественного языка (Natural Language Processing, или коротко NLP) - дисциплина на стыке лингвистики, математики и информатики, которая строит вычислительные модели языка.

## Обзор прикладных задач NLP
Посмотрим, как используется в задачах, исторических и сегодняшнего дня.

Вы проснулись и выключили будильник голосом — сработало распознавание речи. Набрали сообщение, и клавиатура молча исправила две опечатки и подсказала окончание фразы. Открыли почту: двадцати писем в ней нет, они уже в спаме, вы их никогда не увидите, а оставшиеся разложены по категориям и ждут ответа в одно нажатие. Попросили колонку поставить таймер на десять минут — она поняла и чего вы хотите, и на сколько.

По дороге спросили у поисковика «кофейня рядом открыта сейчас»: он сообразил, что «рядом» относится к вашей геопозиции, а не к слову на странице, нашёл «кофейни» и «кофейню» по запросу «кофейня» и не показал тридцать копий одного и того же меню. Навигатор произнёс название улицы вслух, правильно поставив ударение в слове, которое читается двумя способами. Вы включили видео с автоматическими субтитрами, навели камеру на вывеску на незнакомом языке и получили перевод поверх изображения. Пролистали новостной агрегатор, где материалы уже разложены по разделам, а тридцать перепечаток одной новости схлопнуты в один сюжет. Заодно выбрали наушники: «87% положительных отзывов» — это число никто не считал вручную.

На работе вас ждала расшифровка вчерашнего созвона с готовым списком поручений. Вы поискали что-то во внутренней базе и получили не список ссылок, а готовый ответ со ссылками на источники. Написали комментарий к функции, и редактор дописал за вас тело. Отправили договор в систему, которая за секунду вытащила из него стороны, суммы и сроки. Ответили в чате поддержки боту, который с третьей попытки понял, чего вы хотите. И между делом пролистали ленту, не увидев в ней сотни комментариев, удалённых до того, как их кто-либо прочитал.

Ещё не наступил полдень, а вы воспользовались двумя десятками языковых моделей и ни одну из них не назвали бы моделью. Это нормальное состояние области: NLP почти нигде не является продуктом, но встроено почти во всё.

## Распределенная модель текста

__Зеллиг Харрис__ в работе [«Distributional Structure»](https://www.tandfonline.com/doi/pdf/10.1080/00437956.1954.11659520) (1954) предложил описывать язык дистрибутивно — через окружение, в котором встречаются его единицы (слова, фонемы и т.д.), не апеллируя к значению напрямую. Если два слова систематически появляются в похожих контекстах, их значения близки. Его пример: *oculist* и *eye-doctor* встречаются почти в одинаковых окружениях — и означают почти одно и то же. 

Чуть позже __Джон Руперт Фёрт__ сжал ту же мысль в афоризм, который часто цитируют в NLP: «You shall know a word by the company it keeps» (Firth, 1957)

Параллельно с Харрисом его аспирант __Ноам Хомский__ развивал в некотором смысле противоположный подход — [генеративную грамматику](https://ru.wikipedia.org/wiki/%D0%9F%D0%BE%D1%80%D0%BE%D0%B6%D0%B4%D0%B0%D1%8E%D1%89%D0%B0%D1%8F_%D0%B3%D1%80%D0%B0%D0%BC%D0%BC%D0%B0%D1%82%D0%B8%D0%BA%D0%B0). Его критика сводилась к тому что носитель свободно строит и понимает предложения, которых в корпусе никогда не было (например, фраза «Colorless green ideas sleep furiously» статистически невозможна, но ощущается грамматически корректной), а значит, за языком должна стоять жесткая система порождающих правил, которую из одной сочетаемости не вывести

Ро Хомски язык описывается "сверху вниз": от сначала постулировать систему правил (исходя из значений, логики, традиции) и затем накладывать её на материал. А можно "снизу вверх" (Харрис): не вводить категорий заранее, а позволить им вырасти из наблюдаемой сочетаемости

На несколько десятилетий научное сообщество предпочло подход Chomsky: он казался более перспективным, плюс для дистрибутивного подхода требовалось много вычислительных ресурсов, которых не хватало. И дистрибутивный подход отошел на обочину; ранний NLP во многом строился на созданных экспертами грамматиках и правилах. Реванш случился в 1990-е: считать сочетаемость стало проще и устойчивее на реальных текстах. Победил не более «правильный» взгляд на язык, а более инженерный

Победивший "дистрибутивный" подход часто называют статистическим, но стоит отметить, что полноценно статистическим он стал только начиная с 90-х. Сам Харрис частот не считал — он работал в режиме «возможно / невозможно»: пересекаются ли множества контекстов двух единиц. Это был скорее качественный анализ. Но именно переход от сочетаемости к конкретным частотам и превратил программу Харриса в рабочий инструмент

Дистрибутивняый взгляд на язык нашел отражение в большинстве современных NLP методов. [Bag-of-words](https://en.wikipedia.org/wiki/Bag-of-words_model) и [TF-IDF](https://en.wikipedia.org/wiki/Tf%E2%80%93idf) представляют документ через распределение слов в нём. N-граммные модели оценивают вер-ть слова по его контексту. Скрытые марковские модели размечают слово, опираясь на соседей. Латентно-семантический анализ и [кластеризация Брауна](https://en.wikipedia.org/wiki/Brown_clustering) прямо строят представления слов из матрицы «слово — контекст». Все они ничего не постулируют о грамматике заранее, а выводят обобщение из примеров — это Харрис, доведённый до машинного масштаба. Наконец в 2010-х появляется модель [word2vec](https://en.wikipedia.org/wiki/Word2vec), генерирующая контекстные эмбеддинги, и она наиболее прямо воплотила дистрибутивную гипотезу Харриса, пролежавшую к тому моменту шестьдесят лет.

## Информационный поиск
После Второй мировой войны произошел взрывной рост научной и технической литературы, которую необходимо было оцифровывать и классифицировать. Задача подтолкнула к созданию целого научного направления, которое в 1950 году с легкой руки __Кельвина Муэрса__ назвали "информационным поиском" (Information Retrieval).

Информационный поиск (information retrieval, IR) решает задачу: дана коллекция документов $D = \{d_1, \dots, d_N\}$ и запрос $q$, требуется упорядочить документы по убыванию их полезности для пользователя, сформулировавшего этот запрос. Формально система вычисляет функцию ранжирования $s(q, d) \in \mathbb{R}$ и возвращает $k$ документов с наибольшим значением.

Появлявшиеся в это время инструменты были по сути первыми поисковыми системами, на замену которым позже пришли всем известные сервисы типа Yahoo, Google, Яндекс, Bing и прочих.

Для решения этой задачи важно уметь описывать каждый документ в виде компактного представления, кодирующего его содержание, а также сравнивать документы между собой и документ с запросом.

### Векторная модель текста
Среди ранних работ наиболее цитируемыми были работы Джерарда Салтона, которого называют отцом информационного поиска. Он ввел термин векторная модель текста __Vector Space Model__ [(Salton et al, 1975)](https://dl.acm.org/doi/epdf/10.1145/361219.361220), согласно которой любой текстовый документ описывается как вектор частот (term frequency) слов, а близость двух документов измеряется углом между этими векторами.

Множество слов из словаря образует пространство, в котором каждая координата - отдельное слово из словаря. Тексты живут в этом многомерном пространстве. Множество документов составляют Document-Term матрицу, где по строкам отложены документы. а по столюцам слова.

Модели текста, не учитыващие порядок составляющих его слов, собирательно называют __Bag-of-Words__ моделями. относится именно к таким представлениям. В рамках этой модели представления текстов («Cat bites Dog» и «Dog bites Cat») не отличимы. Хотя семн Для сравнения, рекуррентные модели текста или Трансформеры, о которых будем говорить в слежующих глаахе, порядок уже учитывают. Для многих задач, где порядок не так важен (например, задача сравнения документов), этого достаточно.

Векторную модель текста иногда называют алгебраической моделью в том смысле, что она впервые позволила думать о тексте, как о математическом объекте. Действительно, у нас есть непрерывное пространство текстов, на нем заданы операции: сложение векторов, умножение (скалярное) векторов. И также задаются отношения между векторами.

В качестве основной меры близости между двумя векторами была выбрана __косинусная близость__, то есть скларяное произведение, нормированное на длину каждого из векторов. Нормировка, гарантирует что мы сравниваем именно слоавр состав документов, но не их масштаб.

$$\text{cosine\_similarity}(d, q) = 
\frac{\mathbf{d} \cdot \mathbf{q}}{\|\mathbf{d}\| \cdot \|\mathbf{q}\|}
= \frac{\sum_{i=1}^{n} d_i \cdot q_i}
{\sqrt{\sum_{i=1}^{n} d_i^2} \cdot \sqrt{\sum_{i=1}^{n} q_i^2}}$$

Если угол равен 0° (косинус равен единице) — векторы сонаправлены, документы максимально похожи (идеальная релевантность). Если угол равен 90° (косинус равен нулю) — векторы перпендикулярны, документы абсолютно не связаны.

При всем удобстве, у векторного представления Салтона есть несколько очевидных минусов. Давайте их перечислим. 

Во-первых, такие вектора огромны. Они по определению соразмерны количеству слов в словаре, а у нас только в английском языке насчитывается более 1 млн словоформ.

Во-вторых, такие представления почти полностью состоят из нулей. Их неэффективны с точки зрения хранения и обработки. Решить можно инженерно, например, использовать для хранения сжатые представления (sparse matrix). Суть в том, что хранятся только ненулевые значения веткора.

В третьих, это представление заточено искоючительно под лексику. Семантика (смысл) вообще никак не учитывается. Для поиска по словам это норм, для задач, требующих глубокого понимания языка, это слишком поверхностно. При прямом сравнении двух представлений вообще не учитывается синонимию: два слова с близким смыслом могут почти не пересекаться по конкретным контекстам.

### Предобработка текста
Решить проблему большого количества словоформ может в том числе грамотный препроцессинг. Во-первых, он уменьшает размер словаря. Во-вторых,  кол-ва учитываемых словоформ в словаре. 

__Стеминг__ - оставляем только корень слова. 

__Лемматизация__ - приводим слово к его канонической форме (для существительного это единственное число именительный падеж, для глагола настоящее время и так далее). 

Также можно поставить лимит по частоте - в словарь добавляем только самые частотные слова. Эффективность этого процесса отличается от языка к языку, но в среднем на 30% можно сократить размер словаря. но это скорее купирует проблему, а не решает ее фундаментально.

### Взвешиваение по TF-IDF
Ещё одна проблема - неравномерность измерений в векторном представлении текста. Если мы попробуем моделировать реальные документы, первое, с чем мы сталкнемся - самые частотные слова («и», «в», «не») встречаться повсюду и почти ничего не говорят о содержании документа. Несмотря на свою неинформативность, в рамках стандартной модели они получают очень большой вес, что не совсем логично. В 1972 году Карен Спарк Джонс предложила следующую идею: давайте при описании текста использовать не просто частоты слов (TF), а будем домножать каждую частоту на некоторый нормирующий коэффицент, который будет выполнять роль штрафа за слишком большую распространенность слова. Так мы изабвимся от влияния априорной частоты слова, а совпадение редких слов в двух документах будет гораздо более сильным сигналом об их схожести, чем совпадение предлогов и союзов.

Такая логика присвоения весов получила название **TF-IDF** по двум её компонентам, которые пееремножаются: "term frequency" (частоте токена в документе), и "inverse document frequency" (обратной чатсоте токена в корпусе). Общая формула выглядит так:

$$\text{TF-IDF} = \text{TF}(t, d) \cdot \text{IDF}(t)$$

**TF(t,d)** — частота слова $t$ в данном документе $d$. Чем чаще слово употреюляется в документе, тем больше оснований полагать, что это слово важно для этого текста:
$$\text{tf}(t, d) = \frac{f_{t,d}}{\sum_{t' \in d} f_{t',d}}$$

**IDF(t)** — обратная документная частота слова $t$. Чем в большем числе документов коллекции слово встречается, тем меньше его вес. В целом TF-IDF допускает разные способы подсчета своих компонентов, но приведем классическую формулу, которая встречается чаще всего. Пусть N - это число документов в корпусе текстов, тогда IDF отражает логарифм от обратной доли документов, в которых хотя бы раз используется данный токен $t$:

$$\text{idf}(t, D) = \log \frac{N}{|\{d \in D : t \in d\}|}$$

В 1973 году Салтон провел масштабное исследование со сравнением реализаций вариантов TF и IDF компонентов на примере своей поисковой системы SMART и ввел нотацию. Любопыно, что в оригинальной нотации компонент IDF называлась специфичностью (Term specificity).

Роль логарифма в формуле для IDF - смягчать взрывной рост обратной функции на частотах близких к 0.

### Взвешиваение по BM25
После своего появления подход TF-IDF довольно быстро стал стандартом в информационном поиске. Как следствие, выросло и количество попыток сконструировать другие способы взвешивания, учитывающие большее количество факторов. Наиболее успешной модификацеий стала метрика __BM25__, которую британские исследователи разработали в 1994 году для своей поисковой системой __Okapi__ (система получила название в честь Окапи - экзотичекого животного, смеси жирафа и лошади). Несмотря на свою громоздкость, формула BM25 имеет ту же структуру, что и TF-IDF, просто в нее входит большее количество факторов:

$$\text{BM25} = \text{IDF}(t) \times 
\frac{f_{t,d} \cdot (k_1 + 1)}
      {f_{t,d} + k_1 \cdot \left(1 - b + b \cdot \frac{|d|}{\text{avgdl}}\right)}$$
BM25 содержит три ключевых изменения по сравнению с TF-IDF. Во-первых, вклад комопнента TF перестал быть линейным, теперь он уменьшается с ростом количества упоминаний. Скорость насыщения задается параметром $k_1$. При больших $k_1$ влкад частоты перестает расти бычтро, а при $k_1 = 0$ мы откатываемся к обычной частоте TF: $$\frac{tf \cdot (k_1+1)}{tf + k_1}$$
Во-вторых, компонент TF сделали зависящим от длины документа $$b \cdot \bigg(1-\frac{|D|}{avgdl}\bigg)$$

В третьих, авторы внесли изменения в расчет обратной частоты компонент IDF. Теперь она выглядит так: $$\text{IDF}(t) = \log \frac{N - n_t + 0.5}{n_t + 0.5}$$
Вместо доли используют вариант из вероятностного варианта соотношение частот. Прибавка 0.5 нужна, чтобы не делить на ноль, а прибавка +1, чтобы логарифм был гарантированно положительным.

Любопытна история названия, которая по сути повторяет ставшую хрестоматйиной в среде маркетологов историю создания очитителя WD-40 - модель назвали BM25 в честь номера версии формулы, которая оказалась наиболее успешной. А BM это акроним для "Best Matching".

Метод быстро опробовали на соревнованиях по информационному поиску (TREC), где он успешно себя показал и постепенно стал заменой для TF-IDF.

### Латентно-семантическое индексирование
К концу 1980-х точный поиск по словам и фразам стал стандартом, но оставалась нерешенной проблема синонимичности: пользователь мог написать в запросе "автомобиль", но документы, содержащие слово "машина" полностью игнорировались. Группа исследователей из комапнии Bellcore пыталсь обойти это ограничение и искали способ представления текста, как векторов в непрерывном пространстве.

И такой способ был найден, его назвали термином латентно-семантический анализ (LSA), или синонимично латентно-семантическое индексированием (LSI) [(Deerwester et al., 1990)](https://doi.org/10.1002/%28SICI%291097-4571%28199009%2941:6%3C391::AID-ASI1%3E3.0.CO;2-9). Термин LSI больше прижился в информационном поиске, а LSA в когнитивной науке и психолингвистике.

Суть метода в том, что они взяли частотное описания набора текстов («Term-Document») $X$ размера $m \times n$, содержащую частоту каждого из $m$ слов в каждом из $n$ документов (возможно взвешенные по TF-IDF) и применили к ней сингулярное разложение (SVD), который декомпозирует произвольную прямоуглольную матрицу в произведение двух ортогональных матриц U и V и диагонального масштабирования $\sigma$:

$$X = U \Sigma V^{\top}$$

Стоит отметить, что в алгебре существует целый класс разложений, решающих подобную задачу приближения произвкедения . К ним относятся, например QR разложение, LU разложение, Cholesky, NMF. Но SVD получило особенную популярность, поскольку обладает сразу несколькими хорошими свойствами. Во-первых, оно гарантирует оптимальность приближения в терминах матричной нормы. Во-вторых, выбирая размер, мы легко регулируем сложность отдельных компонентов. В третьих, SVD универсален и может работать для прямоугольных матриц.

Замечательное свойство данного разложения в том, что диагональная матрица $\Sigma$ состоит из сингулярных значений, упорядоченных по убыванию. Это значит, что оставив только первые $k$ из них (обычно несколько сотен), мы получим приближенное усечённое разложение (__Truncated SVD__). Разложение - наилучшее в смысле нормы приближение оригиниальной матрицы $X$ матрицей ранга $k$:

$$X \approx X_k = U_k \Sigma_k V_k^{\top}$$

Строки матрицы $U_k \Sigma_k$ можно рассматривать как $k$-мерные описания слов, а столбцы матрицы $\Sigma_k V_k^{\top}$ как векторы документов. Зануляя измерения с наименьшими сингулярнми занчениями в U и V, мы как бы выбрасываем весь малозначимый шум и заставляет слова, встречающиеся в похожих документах, получить близкие представления, даже если они ни разу не встретились в одном документе напрямую. 

Сжатые описания образуют непрерывное пространство описаний. Оно абсолютно неинтерпретируемое, но можно точно сказать, что те самые "автотмобиль" и "машина" в этом пространстве будут располагаться где-то близко [объяснить почему]. Вообще концепция непрерывных представлений  - одна из центральных для машинного обучения и нам ещё будет не раз встречаться. Когда мы "запихиваем" разреженный сигнал в вектор, мы тренируем машину "обобщать", без этого не будет работать. Обощение это ключ к абстрактному мышлению.

Первые попытки ещё в 1960-ых применение факторного анализа для автоматической классификации документов (Borko & Bernick, 1963)
В 1992 году LSI впервые применили в системе автоматического назначения рецензентов на научные статьи<br>нужно было сопоставить текстов с экспертами в данной област. В 1994 году выдан патент на применение мультиязычного LSI: сопоставлять по смыслу написанных на разных языках. В 1995 году LSI впервые задействовали для автоматической оценки сочинений, система измеряла семантическую близость студенческого текста с эталонным ответом. К концу 1990-х метод стал повсеместно использоваться на практике.

К минусам подхода можно отнести то, что он остается «Bag-of-Words» моделью, то есть порядок слов никак не учитывается. Плохо справляется с многозначностью, посольку все смыслы слова смешиваются в один вектор. Требует пересчёта разложения при добавлении новых данных.


## Языковые модели

Языковая модель в самом общем смысле — это распределение вероятностей над последовательностями слов: она должна уметь для любой цепочки слов сказать, насколько та правдоподобна.
$$P(w)=P(w_1,w_2... w_n)$$

Эквивалентно, языковая модель предсказывает следующее слово по предыдущим. Ровно эту задачу решают современные GPT модели: по предыдущему контексту определить наиболее вероятное продолжение текста. Менялись только методы оценки этого распределения:
$$P(w_n|w_1 ... w_{n-1})$$

__Андрей Марков (1913)__ заложил математический фундамент. Анализируя чередование гласных и согласных в «Евгении Онегине», он показал, что текст можно описать как стохастический процесс, где вероятность следующего символа зависит от предыдущего (после гласной более вероятна согласная и т.п.). Так появились [цепи Маркова](https://en.wikipedia.org/wiki/Markov_chain) — формальная основа всех последующих моделей

__Клод Шеннон (1948)__, в работе [«A Mathematical Theory of Communication»](https://people.math.harvard.edu/~ctm/home/text/others/shannon/entropy/entropy.pdf), первым явно смоделировал язык вероятностно. Он строил приближения английского текста разного порядка (n-граммы), ввёл понятие энтропии языка и фактически описал теорию, ставшую основой всей области. Тот самый Клод, в честь которого назван чат-бот Claude от компании Antropic.

Сам термин «языковая модель» в современном смысле закрепился позже — в 1970–80-х годах в сообществе распознавания речи, прежде всего в группе __Фреда Елинека__ (Fred Jelinek) в IBM. Там использовалась модель «шумного канала»: P(слова | звук) ∝ P(звук | слова) · P(слова), и вот эта априорная компонента P(слова) — оценка того, насколько правдоподобна сама фраза — и называлась language model. Елинеку приписывают полушутливую фразу о том, что качество распознавания растёт каждый раз, когда из команды уходит лингвист — это маркер перехода от правил Хомского к статистике.

Вероятность последовательности $w = (w_1...w_n)$ раскладывается в произведение условных вероятностей:

$$P(w_1, w_2, \dots, w_n) = P(w_i|w_{1..i-1}) \cdot P(w_2|w_{1..i-2}) = ... = \prod_{i=1}^{n} P(w_i \mid w_1, \dots, w_{i-1})$$

Проблема в том, что вероятность $P(w_i \mid w_1, \dots, w_{i-1})$ зависит от всей предыстории $w_1 ... w_{i-1}$, и таких предысторий бесконечно много — оценить их по данным невозможно. Здесь и вступает [марковское допущение](https://en.wikipedia.org/wiki/Markov_property): будем считать, что следующее слово зависит не от всей истории, а только от последних $n-1$ слов. Это и есть [n-граммная модель](https://en.wikipedia.org/wiki/Word_n-gram_language_model)

Так например, в биграммных моделях (n=2) каждое слово зависит только от предыдущего: $P(w_i \mid w_1, \dots, w_{i-1}) \approx P(w_i \mid w_{i-1})$. В триграммных ($n = 3$) от двух предыдущих, и так далее. При $n = 1$ мы имеем вырожденную модель, где слова считаются независимыми от конекста и их вероятность оценивается просто частотой. Это сильное упрощение и на практике от такой модели будет мало пользы.

Вероятность $P(w_i \mid w_{i-1})$ оцениваются простым подсчётом частот в обучающем корпусе (оценка максимального правдоподобия):

$$P(w_i \mid w_{i-1}) = \frac{C(w_{i-1}, w_i)}{C(w_{i-1})}$$

где $C(\cdot)$ — число вхождений соответствующей последовательности в корпус. То есть «вероятность слова *молоко* после слова *парное*» — это сколько раз встретилось «парное молоко», делённое на то, сколько раз встретилось «парное» вообще.

К сильным сторонам n-грамных моделей можно отнести их прозрачность. Градиентного обучения не требуется, все вероятности считаются детеминированно, через подсчет частот. Триграммные модели довольно долго были основным рабочим инструментом в задачах распознавания речи и машинного перевода. Нейросетевым языковым моделям потребовались годы, чтобы их превзойти.

К слабым сторонам можно отнести то, что они очень "локальны". В каждый момент времени модель видит только узкое окно в $n-1$ слов, что сильно ограничивает ее "контекстность". Попытка увеличить длину окна $n$ приводит к тому, что мы быстро упираемся в потолок комбинаций, которые надо хранить. Кроме того, такие модели чисто лексические, они не учитывают семантику, не могут эффективно работать с омонимами.

Стоит отметить, что существуют и другие способы моделирования языка: например, более современные диффузионные подходы, подходы основанные на минимизации энергии и прочее. Все они имеют практические воплощения, однако настолько широкую популярность не получили.

## Сглаживание в n-граммные модели
Помимо плохой масштабируемости второй проблемой является дискретность вероятности. Какой бы большой не была обучающая выборка, большинство комбинаций слов не встретятся в ней ни разу. Значит ли это, что у комбинации нулевая вероятность? Нет. Cтоит в тестовом предложении появиться одной неизвестной триграмме, и вероятность всего предложения обнуляется.

Для борьбы с этим эффектом используют различные приемы сглаживания (smoothing), которые в общих чертах можно описать так, что часть вероятностной массы отбирается у тех событий, которые мы наблюдали в выборке, и перераспределяется в пользу ненаблюдаемых. 

Простейший вариант сглаживания, который используется во многих приложениях и известен еше с 18 века, это __сглаживание Лапласа__. Идея в том, чтобы прибавить небольшой вес ко всем частотам, и в числитель, и в знаменатель. Сегодня используется, например, в Наивном Баейсовском классификаторе в задачах детекции спама или определении тональности. Пусть $c_i$ - частота слова $i$, $\alpha$ - сила сглаживания, а |V| - размерность словаря, тогда:

$$\hat{P}(x=i) = \frac{c_i + \alpha}{N + \alpha |V|}$$

В знаменателе мы суммируем скорректированные частоты всех слов, поэтому $\alpha$ умножается на $|V|$. На практике, однако, сглаживание в языковых моделях работает довольно плохо, поскольку реализует слишком грубую трансформацию. Ниже рассмотрим ряд более гибких методов.

В 1953 году Алан Тьюринг работал со своим ассистентом Ирвином Гудом над проектом расшифровки немецкой шифровальной машины Энигма. В процессе работы ему потребовалось оценить вероятность встретить редкие сочетания символов. В результате (Good et al., 1953) появился способ оценивания, в рамках которого вероятность ненаблюдаемых в обучающей выборке слов оценивалось как вероятности слов, встретившихся ровно один раз. Такой способ оценвания получил название __сглаживания Гуда-Тьюринга__. Пусть $с$ - частота слова в обучающей выборке, $N_{c}$ и $N_{c+1}$ - количество слов, встречающихся ровно $c$ и $c+1$ раз, $N_{0}$ количество встречающихся слов. Тогда скорректированная частота слова будет считается по формуле:

$$c^* = (c + 1) \frac{N_{c+1}}{N_c}$$

У этого подхода есть простая геометрическая интерпретация, распишем ее по шагам. Предположим, мы построили гистограмму частотности слов. Первый стоблик - сколько слов встречается по одному разу, второй - сколько по два раза и т.д. Если хотим узнать общее количество слов частоты $c$, оно будет считаться как $c \times N_c$. Чтобы оценить количество слов которые не встречались, мы просто копируем первый стоблик на место нулевого - задача решена, вероятность посчитана. Но остальные стоблики нужно уменьшить, чтобы это была полноценная гистограмма. Уменьшить суммарно на высоту первого стоблика. Распределяем массу правого стоблика $(с+1)N_{c+1}$ по $N_c$ элементам левого столбика. Чтобы частоты сконвертировать в веротяности, достаточно поделить частоты на общее количество символов $N$.

<img src="img/good_turing.png" width=400>

В 1987 году Слава Кац из группы Елинека в IBM предложил еще одну механику корректировки частот, основанную на сглаживании Гуда-Тьюринга. Метод назвали __Katz back-off__. Суть в том, что если n-грамма встречаласб в корпусе текстов, то мы корректируем её дисконтом Гуда–Тьюринга. А если не встречалась, то откатываемся к соотвествующей $(n−1)$-грамме и домножаем вероятность на вес $\alpha$, который считается отдельно для каждого контекста и подобран так, чтобы сумма вероятностей оставалась равной единице.

$$P_{BO}(w_i \mid w_{i-1}) = \begin{cases} d_r \dfrac{c(w_{i-1}w_i)}{c(w_{i-1})}, & c(w_{i-1}w_i) > 0 \\[2ex] \alpha(w_{i-1})\, P_{BO}(w_i), & c(w_{i-1}w_i) = 0 \end{cases}$$

В 1980 в той же команде предлагали альтернативный подход, который назвали интерполяцией __Йелинека–Мерсера__. В данном подходе оценки вероятностей разных порядков (триграммы, биграммы, униграммы) смешиваются всегда, а не только при необходимости "отката" до более короткого окна. 

$$P_{int}(w_i \mid w_{i-n+1}^{i-1}) = \lambda\, P_{ML}(w_i \mid w_{i-n+1}^{i-1}) + (1-\lambda)\, P_{int}(w_i \mid w_{i-n+2}^{i-1})$$

Здесь корректировка не нужна, так как вероятнсоти всегда остаются ненулевыми.

В итоге методы уступили популярность сглаживанию __Кнезера–Нея__, описанному в 1995 году. Вероятность слова в роли «отката» оценивается не по тому, как часто оно встречается вообще, а по тому, в скольких *разных* контекстах оно появляется. Классический пример — слово *Angeles*: оно частотно, но почти всегда идёт после *Los*, поэтому как самостоятельное продолжение произвольного контекста маловероятно. Его интерполированная (модифицированная) версия многие годы считалась SOTA методом для дискретных n-граммных моделей.

### Метрика перплексии
Нужна метрика, которая говорит, насколько "хороша" модель. Такая метрика есть — называется **перплексия** ([perplexity](https://en.wikipedia.org/wiki/Perplexity)) 

Метрику впервые предложили использовать в 1977 году в IBM в проектах по распознаванию речи. Идея в том, что она показывает, насколько модель «удивлена» (perplexed) реальным текстом: если наблюдаемый токен имеет в рамках модели низкую вер-ть быть сгенерированным, то удивление большое.

Возьмем отдельный токен $w_i$ из какого-либо предложения тестового корпуса текстов. Подадим в языковую модель его предыдущий контекст $w_1...w_{i-1}$ и посмотрим, какую P(w) возвращает модель. Если вер-ть генерации правильного токена $w_i$ высока, то все хорошо, модель справляется. 

Если к этой вероятности применить отрицательный логарифм, то она трансформирует задачу максимизации в задачу минимизации:
$$-\ln P(w_i \mid w_1, \dots, w_{i-1})$$

Теперь вероятность близкая к 1 означает низкую перплексию, а к 0 высокую:

<img src="img/perplexity.png" width=400>

Таким образом, значение метрики зависит от того: 
1) насколько хорошо модель научилась предсказывать возможные продолжения в тексте
2) что у нас за текст и насколько он нестандартный
  
Для получения средней перплексии на множестве документов метрика усредняется. По всем токенам данного предложения, по всем предложениям в корпусе документов
$$L = -\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})$$

Получаемая величина (Average negative log-likelihood per token), это по сути и есть функция потерь (Loss function) оптимизруемая при обучении LLM. 

Она же кросс-энтропия $H(P^{model}, P^{true}) = \sum P^{true} \cdot \log P^{model}$, при этом $P^{true}$ вырождается в константу "1.0" и получается привычная форма

<img src="img/perplexity2.png" width=400>

Чтобы перевести значение метрики на более интрепретируемую шкалу от этого выражения еще берут экспоненту:

$$\text{PP}(W) = e^{L} = \exp\!\left(-\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})\right)$$

где $N$ — число слов в тесте

Можно $-1/N$ занести под логарифм, тогда экспонента сократится, а вероятности перемножатся в полную вероятность

$$\text{PP}(W) = P(w_1, \dots, w_N)^{-\frac{1}{N}}$$

На этой шкале у метрики есть интерпретация. Перплексия — это число равновероятных вариантов, между которыми модель в среднем выбирает на каждом шаге генерации. Перплексия 100 означает, что модель в среднем колеблется так, как если бы на каждом шаге угадывала равновероятный исход из ста. Перплексия, равная размеру словаря, соответствует модели, которая не выучила вообще ничего (равномерное распределение). Чем ниже — тем лучше: модель уверенно сужает выбор.

Перплексия обычно падает к концу предложения, так как в конце накоплен конекста больше

<img src="img/perplexity_1.png" width=400>

Несколько важных оговорок, без которых перплексией легко злоупотребить:

- Перплексия зависит от словаря. Сравнивать перплексии двух моделей корректно только при одинаковом словаре и одинаковой разбивке на токены<br><br>
- Перплексия - <u>внутренняя</u> (intrinsic) метрика модели. Она говорит, как хорошо модель моделирует распределение текста, но не говорит, как хорошо она решает прикладную задачу (перевод, распознавание речи). Важно смотреть и на внешние (extrinsic) метрики — например, частоту ошибок распознавания слов (WER) в речевых системах<br><br>
- Перплексия тесно связана с кросс-энтропией: $\text{PP} = 2^{H}$ (или $e^H$, смотря какое основание логарифма), где $H$ — кросс-энтропия в битах (натах) на слово. Поэтому в статьях того времени качество часто отчитывали именно в битах на слово

Перплексию удобно испольовать для детекции сгенерированного текста. У ИИ тексты стабильно низкой перплексией, в то время как перпексия более взрывная

### Кластеризация Брауна
В 1992 году [(Brown et al.)](https://aclanthology.org/J92-4003.pdf) предложил небольшую модификацию n-граммной языковой модели. Выше мы выяснили, что такие модели требуют расчета и хранения гигантского кол-ва вероятностей, а именно $V^n$ при окне $n$ и размере словаря $|V|$, что является их главным ограничением. Идея Брауна заключалась в том, чтобы вместо полного расчета всех веротяностей $P(w_t \mid w_{t-1}, w_{t-2})$, уменьшить размерность словаря, группируя слова в небольшое кол-во кластеров (несколько сотен) и тем самым сократить кол-во рассчитываемых вероятностей:

$$P(w_t \mid w_{t-1}) \;=\; P\bigl(c_t \mid c_{t-1}\bigr) P\bigl(c_{t-1} \mid c_{t-2}\bigr) \cdot P\bigl(w_t \mid c_t\bigr)$$

Переход с уровня классов на уровень слов происходит только на последнем шаге мультпликативной цепочки. В результате общее число параметров сокращается с $|V|^n$ до $C^n + |V|$. Все условные вероятности считаются точно так же как обычные n-грамные модели, по статистике корпуса.

Преимущество подхода не столько в экономии места под таблицу вероятностей, сколько в появившейся возможности учитывать семантику слов. Согласно алгоритму, слова часто употребляющиеся в одном контексте, оказыватся в одном классе. То есть мы моделируем бутылочное горлышко, заставляет модель больше учитывать семантику и учит обощению. Можно рассматривать модель как дальнего предка модели Word2Vec, о которой подробно поговорим в следующей главе. Ведь и тут, и там мы пытаемся найти пододходящее представление слова, ориентируясь на его контекстную близость с другими словами. Разница лишь в том, что здесь представление дискретно.

Вопрос только в том, как именно делать сегементацию $\pi$. Если считать $\pi$ параметром модели, естественно искать $\pi$, максимизирующее правдоподобие корпуса. Подстановка оценок в логарифмическое правдоподобие и группировка слагаемых дают тождество:

$$\frac{1}{T}\,\mathcal{L}(\pi) \;=\; \underbrace{\sum_{c,\,c'} p(c, c')\,\log \frac{p(c, c')}{p(c)\,p(c')}}_{\text{Mutual Information}(C_{t-1};\, C_t)} \;-\; \underbrace{\vphantom{\sum_{c,c'}} H(W)}_{\text{не зависит от } \pi}$$

Второе слагаемое от разбиения не зависит. А вот в первом слагаемом узнается классическая метрика из теории вероятностей - взаимная информация (Mutual Information), имеющая интерпретацию универсальной меры зависимости между двумя случайными величинами, в данном случае между классами соседних слов $c'$ и $c$. То есть идеальная сегментация это сгруппировать слова так, чтобы получились максимально независимые классы.

Полная оптимизация очень ресурсозатратна, поэтому на практике применяется следующая жадная агломеративная процедура: вначале мы считаем каждое слово отедльным классом, а потом среди всех пар классов объединяем ту, слияние которой влечёт наименьшую потерю взаимной информации. Так повторяем до достижения требуемого количества классов *C*.

То есть у нас каждое слово словаря теперь ассоциируется не просто номером класса, но и его позицией в дереве (код - бинарное представление пути по дереву до слова). Это уже не просто скалярная метка, но приближение непрерывного представления, которое можно использовать как признак в задачах. Классический эмбединг появится через 25 лет, но данный веткор уже учитывает семантику. Причем выбирая количество первых битов, можно легко регулировать гранулярность этого представления, от грубого до детального.

Жёсткость разбиения не позволяет моделировать многозначность; 
контекст ограничен одним соседом; 
между классами не определено расстояние, 
поэтому операции над представлениями невозможны.

## Скрытые Марковские Модели

Не все задачи — это классификация целого документа. Многие требуют разметить каждый токен: приписать каждому слову часть речи (POS-tagging), выделить именованные сущности (NER), сегментировать речь. Доминирующим инструментом для таких задач последовательной разметки в эту эпоху была **скрытая марковская модель** (Hidden Markov Model, HMM).

Есть матрица переходов между токенами. Есть переменная (скрытая, скаляр, дискретная: категории описаны), которая влияет на матрицу переходов. 

Идея: наблюдаемая последовательность (слова) порождается ненаблюдаемой («скрытой») последовательностью состояний (например, частей речи). Модель описывается двумя наборами вероятностей:

- Вероятности переходов — насколько вероятно, что за состоянием «прилагательное» идёт состояние «существительное». Здесь снова марковское допущение: следующее состояние зависит только от текущего.
- Вероятности эмиссии — насколько вероятно, что состояние «существительное» порождает конкретное слово «стол».

<img src="img/hmm1.png" width=550>

Задача разметки — найти наиболее вероятную последовательность скрытых состояний для данной последовательности слов. Решается она эффективно динамическим программированием — [алгоритмом Витерби](https://en.wikipedia.org/wiki/Viterbi_algorithm). Параметры оцениваются либо по размеченному корпусу (простым подсчётом, как у n-грамм), либо без разметки — алгоритмом [Баума–Велча](https://en.wikipedia.org/wiki/Baum%E2%80%93Welch_algorithm) (частный случай EM).

HMM хорошо работали для POS-теггинга и были естественным выбором, пока их в начале 2000-х не стали теснить дискриминативные модели — прежде всего условные случайные поля __Conditional Random Fields__ [(Lafferty et al, 2001](https://daiwk.github.io/assets/Conditional%20random%20fields%20Probabilistic%20models%20for%20segmenting%20and%20labeling%20sequence%20data.pdf)). CRF снимали несколько структурных ограничений HMM (в частности, позволяли свободно использовать произвольные, в том числе пересекающиеся, признаки) и стали стандартом для разметки последовательностей на следующее десятилетие — вплоть до прихода нейросетевых моделей. Но логика «скрытые состояния порождают наблюдения, ищем лучшую цепочку Витерби» осталась важной частью образования любого, кто работает с последовательностями.

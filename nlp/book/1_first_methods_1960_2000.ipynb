# Глава 1. Первые модели

## Предыстория

Язык — древнейшая и, возможно, самая важная технология человечества. Когда именно он появился, неизвестно и, по-видимому, не станет известно никогда: звучащая речь не оставляет свидетельств. Мы можем судить только по косвенным признакам — строению голосового тракта; появлению генов, ассоциированных с речевой способностью (например, FOXP2), следам символического поведения вроде украшений, погребальных обрядов и наскальной живописи. Скорее всего развитый язык существовал уже 50–100 тысяч лет назад. Некоторые исследователи связывают его возникновение с самим появлением вида Homo Sapiens 200–300 тысяч лет назад.

Почти всю свою историю язык существовал только в устной форме. Письменность появилась лишь около 5,5 тысячи лет назад — в Шумере (клинопись, ок. 3200 г. до н. э.) и почти одновременно в Египте, а затем была независимо изобретена ещё как минимум дважды, в Китае и Мезоамерике. Письменность превратила язык из акустической волны в данные - то, что можно хранить, копировать, сопоставлять и, что для нас важнее всего, анализировать

В V веке до н. э. индийский грамматист Панини описал санскрит системой из четырёх тысяч коротких правил - по сути создал первую формальную грамматику. Греческая традиция, от софистов до Аристотеля, связала язык с логикой: высказывание стало объектом, к которому применимы формальные операции

Первый документированный случай, когда количественная модель языка решила практическую задачу, относится к IX веку. Арабский учёный Аль-Кинди описал *частотный анализ* - метод взлома шифров простой замены, основанный на том, что буквы естественного языка встречаются с устойчивыми, характерными частотами: достаточно сопоставить распределение символов шифртекста с распределением букв языка. К слову этот принцип использовался потом на протяжении больше тысячи лет. Так в 15 веке с помощью частотного анализа разоблачили заговор Марии Стюарт против королевы Елизаветы. А в годы второй мировой войны взломали немецкую шифровальную машину Энигма

Систематическое приложение теории вероятностей к тексту началось в 1913 году, когда Андрей Андреевич Марков взял первые 20 000 букв «Евгения Онегина» и показал, что чередование гласных и согласных хорошо описывается случайным процессом, в котором распределение следующего символа зависит только от текущего:

$$P(x_{n+1} | x_n, x_{n-1} \dots x_1) = P(x_{n+1} | x_n)$$

Марковское Так возникло понятие марковской цепи и мысль, что текст можно рассматривать как случайную последовательность с локальными зависимостями. К 1930–40-м годам накопились достаточно данных для исследований и Джордж Ципф заметил, что если упорядочить слова корпуса текста по убыванию их частоты, то частота слова оказывается примерно обратно пропорциональна его рангу (номеру в отсортированной послежовательности)

$$f(r) \approx \frac{C}{r^{\alpha}}, \qquad \alpha \approx 1$$

Что ещё удивительнее. закон Ципфа воспроизводится во всех языках, независимо от жанра текста. Так, например, самая редкая буква русского алфавита "й" встречается в среднем в 33 раза чаще самой популярной буквы "о". Иными словами, небольшой набор слов покрывает большую часть любого текста, а «хвост» из редких слов практически бесконечен

Клод Шеннон занимался исследованием . В «Математической теории связи» (1948) он предложил рассматривать источник сообщений — в том числе человека, порождающего текст, как случайный процесс и ввёл числовую меру его непредсказуемости, энтропию:

$$H = -\sum_{x} p(x) \log_2 p(x)$$

Приближая английский текст моделями возрастающего порядка — от независимых букв к парам, тройкам и целым словам, — Шеннон фактически ввёл n-граммные модели языка, а в опытах с людьми оценил энтропию английского примерно в один бит на букву, впятеро меньше, чем $\log_2 27 \approx 4{,}75$ бита для равновероятных символов. Язык, стало быть, глубоко избыточен, и именно эту избыточность — предсказуемость продолжения — и улавливает языковая модель. 

Из теории Шеннона следовал принципиальный вывод: чтобы эффективно сжимать и передавать сообщения, нужно уметь предсказывать их продолжение, то есть располагать вероятностной моделью источника; хорошая языковая модель и хороший архиватор — математически одна и та же задача. Не случайно и криптоанализ Второй мировой войны, от Блетчли-парка до работ самого Шеннона по секретной связи, держался на статистических свойствах языка.

Появление компьютеров дало толчок развитию языкового моделирования. В 1949 году Уоррен Уивер разослал меморандум, в котором предложил смотреть на перевод как на задачу дешифровки: русский текст закодирован английским и наоборот. Это в свою очеедь значит, что машина в принципе способна его раскодировать. В 1950 году Алан Тьюринг предложил считать критерием машинного интеллекта способность вести неотличимый от человеческого разговор - так язык оказался в самом определении искусственного интеллекта. А в январе 1954 года состоялся Джорджтаунский эксперимент: компьютер IBM 701 перевёл с русского на английский около шестидесяти тщательно подобранных предложений, и организаторы пообещали решить проблему машинного перевода за три–пять лет. Обещание не сбылось, на самом деле потребовалось больше 60 лет, но именно тогда родилась дисциплина, которую сегодня называют обработкой естественного языка (natural language processing, NLP)

Ноам Хомский («Синтаксические структуры», 1957) построил теорию формальных грамматик и их иерархию — от регулярных языков к контекстно-свободным и далее — и решительно выступил против статистического взгляда: его знаменитый пример «Colorless green ideas sleep furiously» грамматически безупречен, хотя имеет нулевую частоту в любом корпусе, а значит, рассуждал Хомский, грамматичность не сводится к вероятности. 

Под влиянием этих идей NLP на два десятилетия стало по преимуществу инженерией правил. Системы вроде ELIZA (1966), пародировавшей психотерапевта, и SHRDLU (1971), понимавшей команды в игрушечном мире кубиков, производили сильное впечатление, но упирались в общую стену: вручную составленных правил всегда оказывалось недостаточно, а их взаимодействие — хрупким. Отчёт комиссии ALPAC (1966), констатировавший, что машинный перевод не оправдал вложений, надолго заморозил финансирование и остудил область.

Возрождение пришло оттуда же, откуда началась вся история, — из вероятностного мира Маркова и Шеннона. В 1970–80-е годы группа Фредерика Елинека в IBM показала, что распознавание речи лучше всего работает как задача вероятностного вывода, а на рубеже 1990-х тот же аппарат был перенесён на перевод. В канонической модели зашумлённого канала наилучший перевод $\hat{e}$ для входной фразы $f$ ищется как

$$\hat{e} = \arg\max_{e} P(f \mid e)\, P(e),$$

и вторым сомножителем здесь стоит именно языковая модель $P(e)$. Совместная вероятность текста по цепному правилу раскладывается в произведение условных вероятностей, которые на практике усекаются марковским приближением порядка $k$:

$$P(w_1, \dots, w_n) = \prod_{i=1}^{n} P(w_i \mid w_1, \dots, w_{i-1}) \approx \prod_{i=1}^{n} P(w_i \mid w_{i-k}, \dots, w_{i-1}).$$

N-граммные модели, оценённые по большим корпусам, на двадцать лет стали рабочей лошадкой области; появились размеченные корпуса вроде Penn Treebank, а девиз эпохи можно передать так: данные важнее правил.

Последний поворот сюжета — нейросетевой — вырос из старой дистрибутивной гипотезы (Зеллиг Харрис, 1954; Джон Фёрс, 1957: «you shall know a word by the company it keeps» — о слове судят по его окружению). Если значение слова определяется контекстами его употребления, то слово можно представить плотным вектором, выученным из этих контекстов. Нейронная языковая модель Бенджио (2003) и векторные представления word2vec (2013) сделали эту идею практичной; рекуррентные сети, архитектура «кодировщик–декодировщик» и механизм внимания (2014) позволили работать с целыми последовательностями; трансформер (2017) сделал обучение масштабируемым и открыл дорогу предобученным моделям. Существенно, однако, что при всей смене аппарата целевая функция современных больших языковых моделей осталась бы понятна и Маркову, и Шеннону — максимизируется правдоподобие следующего слова:

$$\mathcal{L}(\theta) = \sum_{t} \log p_{\theta}(w_t \mid w_{<t}),$$

просто теперь распределение $p_{\theta}$ задаётся сетью с миллиардами параметров, а корпусом служит заметная доля всего написанного человечеством.

Оглядываясь на этот путь, легко увидеть, что потребность моделировать язык возникала снова и снова из самых разных источников: из нужды сохранять канонические тексты без искажений, из криптографии, из инженерии связи, из геополитики и мечты о машинном переводе, из желания понять устройство человеческого мышления и, наконец, из простого практического обстоятельства — язык остаётся главным носителем человеческого знания и самым естественным интерфейсом между человеком и машиной. Обработка естественного языка и есть дисциплина на стыке лингвистики, математики и информатики, которая строит вычислительные модели языка; их устройству и посвящено дальнейшее изложение.

#### Машинный перевод
Есть текст на оригинальном языке и нужно сгенерировать его перевод на другом языке, как можно точнее передав смысл и по возможности сохранив стиль и регистр. Трудность в том, что соответствие между языками не пословное: меняется порядок слов, часть грамматических категорий одного языка в другом попросту отсутствует, а выбор эквивалента зависит от контекста.

Это одна из первых по настоящему прикладных задач, которая в середине 20 века дала гигантский толочок развитию всей области. Хотя по мнению некоторых экспертов, в своей базовой постановке задача уже решена, она сохраняет актуальность в некоторых аспектах. Например, остается ещё много языков с небольшим объемом оцифрованных текстов. В длинных документах (например, книги) сложно удерживать контекст и, например, отсылка к герою, главами ранее, может считаться моделью неверно. Моделям первода далеко до идеала в плане считывания культурного контекста, юмора, игры слов, иронии, локальных идиом или сохранения уникального авторского стиля, это все вектор развития на ближайшие годы. Сейчас активно развивается перевод в реальном времени, в скором будущем достаточно будет надеть гарнитуру и представлители любых национальностей смогут свободно общаться между собой, не зная ни слова на языке друг друга. Не так давно появилась возможность за пару часов заменять голосовые аудиодорожки в фильмах, для чего ранее требовалась работа команды переводчиков и мастеров дубляжа. Все это заслуга прорыва в сфере машинного перевода.

Как развивалось направление. Насалась в 1949 году, когда Уоррен Уивер отправил коллегам открытое письмо, озаглавленное "Translation", где персппективы научить маины пеерводу. Среди своих тезисов, он провел аналогию аналогию между задачей перевода и задачу дешифрования. Считается, что этот текст заложил фундамент для индустрии машинного перевода. В 1954 году в IBM провели уже упоминаемый выше Джорджтаунский эксперимент, в рамках которого в машину загрузили записанные на перфокартах 60 предложений на русском языке и машина перевела их на английский. Это была скорее демонстрация возможностей, чем реальный прорыв, однако породила всплекс интереса и привлекло финансирование в область. Спустя 10 лет в 1966 году собрали комитет ALPAC и подвели промежуточные итоги предыдущих десяти лет развития индустрии и вяснилось, что по всем параметрам человеческие переводчики оказались лучше машинного перевода.

Если попытаться уложить историю развития, в первые годы пытались делать системы, которые основывались на правилах грамматики (RBMT). В конце 1980-х они уступили место статистическому подходу (SMT), где перевод раскладывался на модель соответствия фраз и языковую модель целевого языка, следившую за гладкостью. Но решить задачу смогу только нейронный перевод, появившийся после 2000 года. Вехой стало появление в 2016 году сервисов Google Translate и модели перевода GMNT, а также появление в 2017 году модели Трансформер, революционизировавший работу с текстом.

#### Суммаризация текстов
Задача суммаризации очень похожа на задачу машинного перевода. Есть длинный текст (например статья, переписка, комментарий, стенограмма совещания), но сгененрировать нужно его сжатую версию на том же языке. Эта уменьшенная версия текста должна сохранять содержание, ключевые факты, но опускать незначащие детали.

Первая система автоматического реферирования появилась в IBM в 1958 году. Логика ее работы заключалась в том, что она оценивала значимость каждого предложения из оригинального текста, основываясь по частоте входящих в него слов. А затем наиболее весомые предложения объединялись и получалась суммаризация. Такой подход называется __экстрактивной суммаризацией__, поскольку ничего не генерирует, а просто достает из оригинала нужные куски. Подход стал стандартом почти полвека, менялся только критерий оценки "значимости" предложения.

В 2020-х с развитием языковых моделей подход уступил первенство __абстрактивной суммаризации__ - подходу, когда суммаризация полностью генерируется с нуля. Молель куда более гибкая, но вместе с тем и рискованная она может ошибочно включить в реферат утверждение, которого в источнике не было, и такую ошибку читатель, не видевший оригинала, обнаружить не в состоянии.

#### Генерация контента
Одна из самых широких задач. На входе мы имеем текстовое задание, а от системы нужен связный, уместный по стилю и фактически корректный сгенерированный текст.

Если в старых моделях ограничивались прогнозы погоды, отчёты о биржевых торгах, репортажи о спортивных матчах, описания товаров в каталогах, то с появлением больших языковых моделей генерация стала универсальной и породила целую индустрию: маркетинговые тексты, черновики писем и документов, сценарии, локализация зарубежного контента. 

Первая — фактическая ненадёжность: правдоподобный по форме текст может содержать выдуманные детали. 
Вторая — стилистическая усреднённость, узнаваемая интонация «текста, написанного моделью». 
Третья интереснее для нашей темы: интернет наполняется сгенерированными текстами, которые попадают в обучающие выборки следующих поколений моделей, и вопрос о том, к чему приводит обучение на собственных выходах, стал самостоятельным предметом исследований.

#### Написание кода
Есть выскоуровневое описание продукта или ПО, содержащее набор требований на естественном языке и нужно по этому описанию сгенерировать работающий код на каком-то из языков программирования, а также объяснить, дополнить или исправить уже существующий. 

Удобно, что код это тоже текст, хоть и не считается *естественным* языком. Это значит, что и все описанные механизмы работы с языком к нему также применимы. А ещё у задачи есть приятное свойство: корректность генерации можно легко проверить: код либо компилируется и проходит тесты, либо нет.

Кодовые ассистенты существовали давно, но их помощь сводилась в основном к автодополнению по статистике или по шаблонам. Появившаяся в 2021 году модель Codex от Microsoft и построенный на ней Copilot показали, что предобученная на коде языковая модель способна генерировать довольно сложные проекты, основываясь только на промптах на естественном языке. Кроме того, код нужен не только для разработки, это еще и ключевой инструмент аналитики (SQL запросы, макросы Excel), поэтому задача text-to-SQL сейчас это также один из самых востребованных корпоративных сценариев.

#### Информационный поиск
Есть большая коллекция документов (в базовом варианте текстовых) и есть короткий запрос (query), отражающий текущую потребность пользователя. Нужно отобрать документы, отвечающие этой потребности, и упорядочить их по убыванию релевантности. Задача лежит в основе работы поисковых систем.

Вклад задачи поиска в развитие области был тоже колоссальный. Ради нее во многом создавалась векторная модель текста, которая представила работу с текстами как математическую задачу (мы разберем ниже в этой главе). Проложила дорогу таким сервисам как Google, Яндекс, Perplexity и т.п.

Задача продолжает быть актуальной. Последних лет значительно выросла роль семантики. С развитием языковых моделей после 2020 года двинулась в сторону объединения поиска (Retrieval) и генерации и возникла схема RAG (retrieval-augmented generation), где найденные классическими методами документы подаются языковой на вход модели как часть контекста.

#### Вопросно-ответные системы
Есть фактологический запрос от пользвателя, сформулированный на языке, и нужно дать ответ на заданный вопрос, в текстовом формате, по возможности короткий и проверяемый. То есть не только найти релевантный источник информации, нужно ещё извлечь из него точный фрагмент и возможно сформулировать ответ самостоятельно. Такие системы можно рассматривать как промежуточное звено между обычным поиском и современными чат-ботами.

Первые системы работали над структурированными базами: BASEBALL (1961) отвечала на вопросы о результатах бейсбольных матчей, LUNAR (1971) о химическом составе лунного грунта, обе переводили вопрос в формальный запрос по грамматическим правилам в жёстко ограниченной предметной области. Систематическое развитие началось с трека QA на конференции TREC в конце 1990-х. Демонстрацией первых успехов стала победа системы IBM Watson в телевикторине Jeopardy (в России больше известная как Своя Игра) в 2011 году. Ответы на вопросы искались в реальном времени по большому корпусу текстов ансамблем из сотен эвристик и классификаторов.

#### Персональные ассистенты
Есть произвольно сформулированная просьба от пользователя, в голосовом или текстовом формате (например, "запиши меня к зубному на своодную дату"). Нужно понять, что именно хочет польщователь, выполнить действие и отчитаться по результату. Тут возникает сразу пайплайн из задач, это и распознавание речи, и определение намерения, и поиск информации, и генерация ответа.

Идея разговаривать с машиной на естественном языке появилась вместе с самой областью: одна из первых разговорных систем ELIZA (1966) имитировала психотерапевта. Система работала на простой подстановке шаблонов, но даже этого было достаточно, чтобы производить на людей сильное впечатление. Массовая волна таких систем пришла с ростом популярности мобильных устройств — Siri (2011), затем Google Assistant, Alexa, Алиса. 

Классический ассистент устроен как конвейер: распознавание речи, определение интента и извлечение слотов, менеджер диалога с прописанными сценариями, генерация ответа по шаблону, синтез речи. Каждый компонент разрабатывается и обучается отдельно. Ошибки также накапливаются по цепочке.

#### Определение интента пользователя
Есть запрос или реплика пользователя «стиральные машины, дешево, сравнение», нужно по этой реплике понять, что нужно пользователю. запрашивает изменение брони, и заодно извлечь параметры — дату, номер заказа. Технически это связка из классификации всего высказывания и заполнения слотов (slot filling), которое решается как задача разметки последовательности — тем же аппаратом, что и распознавание именованных сущностей, о котором речь пойдёт ниже. Эталонным корпусом долгие годы служил ATIS с запросами о расписании авиарейсов, собранный ещё в начале 1990-х.

Слабое место классической схемы — закрытый список интентов. Система умеет ровно то, что в неё заложили, а всё остальное падает в корзину «не понял, переформулируйте». Именно этим объясняется ощущение ограниченности ранних голосовых ассистентов и ботов техподдержки.

Следующая группа задач сводится к одной постановке: приписать документу метку из заранее заданного набора. Именно здесь мешок слов и наивный байесовский классификатор десятилетиями показывали неприлично хорошее качество при почти нулевой вычислительной стоимости.

#### Анализ тональности
Есть текст, нужно определить, какое отношение автора выражено в тексте, положительное, отрицательное или нейтральное, А ещё к какому конкретно аспекту это отношение относится (aspect-based sentiment analysis). Например, в отзыве на отель еда может быть отличной, а номер грязным.

В отдельное напрвление задача оформилась в 2002 году, и тогда же начал использоваться термин Sentiment Analysis. Это произогло в том числе потому, что появилось много размеченных датасетов.

Основной потребитель — маркетинг и системы анализа рынка: отслеживание отзывов, реакции на рекламные кампании, сравнение с конкурентами. А также финансовый мониторинг: тональность новостного потока и сообщений в социальных сетях используют как признак в торговых стратегиях системы трейдинга.

К сложностям относится наличие иронии, сарказма, отрицания, культурный контекст. Фразы «ну спасибо, очень помогло».

#### Детекция спама
Есть входящее сообщение. Нужно понять, нормальное это письмо или спам.

Фильтрация нежелательной почты стала первым массовым применением машинного обучения к тексту, с которым столкнулся обычный пользователь. В 1998 году Microsoft работу про Байесовский подход к классикаиции. В 2002 году Эссе Пола Грэма «A Plan for Spam» (2002) популяризовало байесовскую фильтрацию: письмо разбивается на слова, для каждого слова оценивается вероятность встретиться в спаме и в обычной почте, вероятности комбинируются. Метод оказался неожиданно эффективным и, что важнее, персонализируемым — фильтр обучался на почте конкретного человека.

Особенность задачи в том, что она соревновательная: по ту сторону находится противник, который активно адаптируется. Спамеры начали заменять буквы похожими символами, вставлять случайные фрагменты нейтрального текста для «отравления» байесовского фильтра, переносить текст в картинки. Поэтому детекция спама быстро перестала быть чисто текстовой задачей и превратилась в комбинацию текстовых признаков, репутации отправителя, поведенческих сигналов и графового анализа. Тот же контур сегодня работает в детекции фишинга, где цена ошибки существенно выше.

#### Системы модерации
Есть сгенерированный пользователями контент. Требуется найти в потоке то, что нарушает правила площадки, и желательно раньше, чем нарушение увидят остальные пользователи. Под этой формулировкой скрывается несколько разных задач: детекция токсичности и оскорблений, выявление призывов к насилию, поиск мошеннических схем, отсев накрученных отзывов и дубликатов объявлений. Промышленный интерес к теме возник вместе с ростом социальных платформ; типичный ориентир — набор данных и сервис Perspective API от Jigsaw (2017), задавший стандарт оценки токсичности комментариев.

Практические сложности здесь специфические. Во-первых, сильный дисбаланс классов: нарушения составляют доли процента потока, и метрика accuracy бесполезна. Во-вторых, всё та же состязательность — пользователи быстро учатся обходить фильтры через замену символов и эвфемизмы. В-третьих, решающую роль играет контекст: одно и то же слово может быть оскорблением, цитатой или самоназванием, и модель, не видящая ветку обсуждения, ошибается систематически. Поэтому автоматическая модерация почти всегда работает в связке с людьми: модель приоритизирует поток, финальные решения по спорным случаям принимает человек.

#### Рубрикация и тематическое моделирование

Нужно приписать документу тему — либо из заранее заданного списка рубрик, либо, если списка нет, выделить темы самостоятельно. Первая постановка выросла из библиотечного дела и новостных агентств; классическим набором данных здесь долго была коллекция новостей Reuters. Вторая, тематическое моделирование, ставится без учителя: рубрики заранее неизвестны и должны сами проявиться из данных. Линия развития идёт от латентно-семантического анализа (1990) через вероятностный LSA к латентному размещению Дирихле (LDA, 2003), которое на полтора десятилетия стало стандартным инструментом разведочного анализа текстовых коллекций.

Прикладной смысл — навигация по большим массивам: о чём вообще пишут в этих ста тысячах обращений в поддержку, какие темы доминируют в отзывах, как менялась повестка изданий за десять лет. Сегодня ту же работу чаще делают через кластеризацию плотных векторных представлений, а называние получившихся кластеров поручают языковой модели.

#### Распознавание именованных сущностей
Есть текст, в нем нужно найти упоминания людей, организаций, географических объектов, дат, денежных сумм и т.п. Выделить из текста список конкретных объектов, о которых в нём говорится. 

Задача известна как NER (named entity recognition). Термин закрепился на конференции MUC-6 в 1995 году, где задачу впервые формализовали как отдельное соревнование. Формально, это разметка последовательности: каждому токену приписывается метка вроде «название организации», «страна» или «ничего». Классификатор.

Методы развивались от словарей и правил, через скрытые марковские модели и условные случайные поля (CRF) к двунаправленным LSTM моделям и BERT. 

На NER держится обработка документов в банках и страховых компаниях, обезличивание медицинских карт перед передачей исследователям, разбор резюме, мониторинг упоминаний компаний в новостях, извлечение фактов для построения графов знаний.

#### Определение частей речи
Есть текст, нужно провести его морфологическую разметку: каждому слову грамматическую категорию: существительное, глагол, предлог и так далее, а для флективных языков вроде русского ещё и падеж, число, род, время. Задача называется POS-разметка (Part-of-Speech Tagging).

Задача выглядит тривиально ровно до первого омонима: «стекло» может быть существительным и глаголом, «печь» — тоже, и разрешить неоднозначность можно только по контексту. Первые теггеры работали на базе наборов правил (система TAGGIT для Brown Corpus в 1971 году. На смену пришли скрытые марковские модели и трансформационный теггер Брилла (1992), а с появлением размеченных корпусов вроде Penn Treebank точность на английском поднялась до 97% — уровня разметчиков.

Сегодня POS-теггинг почти не встречается как отдельный продукт, скорее это инструмент препроцессинга, и остаётся внутри библиотек, систем синтеза речи, в исследованиях, и поисковых системах для нормализации словоформ. Это хороший пример задачи, которая была решена настолько хорошо, что перестала быть интересной.

#### Исправление опечаток и грамматических ошибок
Есть текст с ошибками, нужно получить исправленную версию этого текста. Система должна во-первых заметить ошибку, во-вторых понять, как лучше исправить.

Первые проверки орфографии появились в 1970-х, а массовым инструментом стали T9 в мобильных телефонах и подчёркивание красной волнистой линией в текстовых редакторах. В 1965 году для оценки "величины" ошибки предложили использовать метрику Левенштейна (1965) - минимальное число правок, превращающих одну строку в другую. так и называют "Редакторское расстояние". В контексте информационного поиска (если опечатка в запросе), иногда используют инженерный: связать ошибочный запрос и следующий за ним исправленный.

Более сложный вариант задачи — исправление грамматических и стилистических ошибок (grammatical error correction). Классические системы собирали для этого отдельные классификаторы под каждый тип ошибки: артикли, предлоги, согласование времён. Современный подход трактует задачу как частынй случай задачи перевода - нам нужно перевести с «неправильного» на «правильный».

#### Распознавание речи
Есть аудиозапись, на выходе нужно получить текст произнесённого. Именно распознавание речи было главным заказчиком n-граммных языковых моделей, которые критически важны для разрешения большого объема неоднозначностей, встречающиеся в работе с аудио сигналом (представления фраз «recognize speech» и «wreck a nice beach» почти идентичны). 

Обратная задача, синтез речи, тоже требует языкового анализа: чтобы прочитать текст, нужно как минимум снять омографию и расставить ударения.

Распознавание речи сегодня — неотъемлемый элемент для целого класса продуктов: голосовые ассистенты, автоматические субтитры, транскрипция звонков в колл-центрах, расшифровка совещаний с последующей суммаризацией и извлечением поручений.

#### Антиплагиат и поиск дубликатов
Есть документ, нужно установить, содержит ли документ фрагменты, заимствованные из чужих текстов, и указать источник заимствования. 

Технически это сводится к поиску пересечений между документом и большой коллекцией. Базовый приём — шинглирование: текст режется на перекрывающиеся цепочки слов, от каждой берётся хеш, и близость документов оценивается по доле совпавших хешей. 

Считать это честно на всей коллекции невозможно, поэтому применяют вероятностные схемы вроде MinHash (Бродер, 1997) и локально-чувствительное хеширование, позволяющие искать похожие документы за приемлемое время. Тот же аппарат используется для дедупликации веб-страниц, объявлений и, что немаловажно для нашей темы, обучающих корпусов языковых моделей.

Слабость точного сопоставления очевидна: достаточно перефразировать текст, и совпадений не останется. Отсюда переход к семантическому сравнению через векторные представления — но вместе с ним и рост числа ложных срабатываний, поскольку близкие по смыслу тексты на общую тему не обязательно являются заимствованием. Отдельный сюжет последних лет — попытки отличить текст, написанный человеком, от сгенерированного моделью. Надёжного решения этой задачи на сегодня не существует, а публичные детекторы демонстрируют высокую долю ошибок, что не мешает им применяться в образовании.

#### Автоматизированное оценивание
Есть работа учащегося (сочинение, развёрнутый ответ на вопрос, решение задачи), и нужно выставить оценку, так как это сделал бы преподаватель. 

Первые такие системы, например, Project Essay Grade появились в 1966 году и оценивала сочинения по набору косвенных признаков - длине текста и предложений, разнообразию лексики, доле сложных слов. К 1990-м начала учитваться семантика. Многие подобные системы (например, e-rater от ETS) применяются в массовом тестировании до сих пор, часто в паре с реальным экспертом.

Поскольку оценка строится на поверхностных признаках, её можно накрутить, например, написав длинный связный текст ни о чём. Сегодня языковые модели стали использовать для оценки других языковых моделей — подход LLM-as-a-judge, который сегодня лежит в основе большинства пайплайнов оценки качества генерации.

#### Разметка и генерация обучающих данных

Требуется получить размеченный набор примеров для обучения другой модели: приписать имеющимся текстам метки или породить сами примеры с нуля. В классическом списке задач такой не было, но сегодня она одна из самых экономически значимых. Ручная разметка — главная статья расходов в обучении с учителем, и естественным шагом стало поручить её модели: сильная модель размечает данные или порождает синтетические примеры, на которых обучается модель поменьше. На этой схеме построены многие открытые инструктивные наборы данных, а сама она получила название дистилляции знаний. Подробнее мы вернёмся к этому в главе про дообучение.

Ну и в целом в любых задачах, где текст выступает признаком или целевой переменной. А с учетом того, что количество данных растет, это почти все задачи. Например, оценка кредитного риска по описанию бизнеса, прогноз оттока по обращениям в поддержку, рекомендации по описаниям товаров, поиск похожих вакансий.

## Распределенная модель текста

__Зеллиг Харрис__ в работе [«Distributional Structure»](https://www.tandfonline.com/doi/pdf/10.1080/00437956.1954.11659520) (1954) предложил описывать язык дистрибутивно — через окружение, в котором встречаются его единицы (слова, фонемы и т.д.), не апеллируя к значению напрямую. Если два слова систематически появляются в похожих контекстах, их значения близки. Его пример: *oculist* и *eye-doctor* встречаются почти в одинаковых окружениях — и означают почти одно и то же. 

Чуть позже __Джон Руперт Фёрт__ сжал ту же мысль в афоризм, который часто цитируют в NLP: «You shall know a word by the company it keeps» (Firth, 1957)

Параллельно с Харрисом его аспирант __Ноам Хомский__ развивал в некотором смысле противоположный подход — [генеративную грамматику](https://ru.wikipedia.org/wiki/%D0%9F%D0%BE%D1%80%D0%BE%D0%B6%D0%B4%D0%B0%D1%8E%D1%89%D0%B0%D1%8F_%D0%B3%D1%80%D0%B0%D0%BC%D0%BC%D0%B0%D1%82%D0%B8%D0%BA%D0%B0). Его критика сводилась к тому что носитель свободно строит и понимает предложения, которых в корпусе никогда не было (например, фраза «Colorless green ideas sleep furiously» статистически невозможна, но ощущается грамматически корректной), а значит, за языком должна стоять жесткая система порождающих правил, которую из одной сочетаемости не вывести

Ро Хомски язык описывается "сверху вниз": от сначала постулировать систему правил (исходя из значений, логики, традиции) и затем накладывать её на материал. А можно "снизу вверх" (Харрис): не вводить категорий заранее, а позволить им вырасти из наблюдаемой сочетаемости

На несколько десятилетий научное сообщество предпочло подход Chomsky: он казался более перспективным, плюс для дистрибутивного подхода требовалось много вычислительных ресурсов, которых не хватало. И дистрибутивный подход отошел на обочину; ранний NLP во многом строился на созданных экспертами грамматиках и правилах. Реванш случился в 1990-е: считать сочетаемость стало проще и устойчивее на реальных текстах. Победил не более «правильный» взгляд на язык, а более инженерный

Победивший "дистрибутивный" подход часто называют статистическим, но стоит отметить, что полноценно статистическим он стал только начиная с 90-х. Сам Харрис частот не считал — он работал в режиме «возможно / невозможно»: пересекаются ли множества контекстов двух единиц. Это был скорее качественный анализ. Но именно переход от сочетаемости к конкретным частотам и превратил программу Харриса в рабочий инструмент

Дистрибутивняый взгляд на язык нашел отражение в большинстве современных NLP методов. [Bag-of-words](https://en.wikipedia.org/wiki/Bag-of-words_model) и [TF-IDF](https://en.wikipedia.org/wiki/Tf%E2%80%93idf) представляют документ через распределение слов в нём. N-граммные модели оценивают вер-ть слова по его контексту. Скрытые марковские модели размечают слово, опираясь на соседей. Латентно-семантический анализ и [кластеризация Брауна](https://en.wikipedia.org/wiki/Brown_clustering) прямо строят представления слов из матрицы «слово — контекст». Все они ничего не постулируют о грамматике заранее, а выводят обобщение из примеров — это Харрис, доведённый до машинного масштаба. Наконец в 2010-х появляется модель [word2vec](https://en.wikipedia.org/wiki/Word2vec), генерирующая контекстные эмбеддинги, и она наиболее прямо воплотила дистрибутивную гипотезу Харриса, пролежавшую к тому моменту шестьдесят лет.

## Информационный поиск
После Второй мировой войны произошел взрывной рост научной и технической литературы, которую необходимо было оцифровывать и классифицировать. Задача подтолкнула к созданию целого научного направления, которое в 1950 году __Кельвин Муэрс__ назвал "информационным поиском" (Information Retrieval). Появившиеся инструменты были по сути первых поисковыми системами, на замену которым позже пришли Yahoo, Google, Яндекс, Bing и прочие.

Информационный поиск — это процесс нахождения нужного документа в крупных коллекциях документов (в основном текстовых), который максимально удовлетворяет информационную потребность пользователя, выраженную в виде поискового запроса (query). Соотвественно для решения этой задачи важно уметь описывать каждый документ в виде компактного представления, кодирующего его содержание, а также сравнивать документы между собой и документ с запросом.

### Векторная модель текста

Наиболее цитируемыми были работы Джерарда Салтона, которого называют отцом информационного поиска. Он ввел термин векторная модель __Vector Space Model__ [(Salton et al, 1975)](https://dl.acm.org/doi/epdf/10.1145/361219.361220), согласно которой текстовый документ описывается как вектор частот (term frequency) слов, а близость двух документов измеряется углом между этими векторами.

Множество слов из словаря образует пространство, в котором каждая координата - отдельное слово из словаря. Тексты живут в этом многомерном пространстве. Множество документов составляют Document-Term матрицу, где по строкам отложены документы. а по столюцам слова.

Модели текста, не учитыващие порядок составляющих его слов, собирательно называют __Bag-of-Words__ моделями. относится именно к таким представлениям. В рамках этой модели представления текстов («Cat bites Dog» и «Dog bites Cat») не отличимы. Хотя семн Для сравнения, рекуррентные модели текста или Трансформеры, о которых будем говорить в слежующих глаахе, порядок уже учитывают. Для многих задач, где порядок не так важен (например, задача сравнения документов), этого достаточно. 

<img src="img/vector_space.png" width=350>

Векторную модель текста иногда называют алгебраической моделью в том смысле, что она впервые позволила думать о тексте, как о математическом объекте. Действительно, у нас есть непрерывное пространство текстов, на нем заданы операции: сложение векторов, умножение (скалярное) векторов. И также задаются отношения между векторами.

В качестве основной меры близости между двумя векторами была выбрана __косинусная близость__, то есть скларяное произведение, нормированное на длину каждого из векторов. Нормировка, гарантирует что мы сравниваем именно слоавр состав документов, но не их масштаб.

$$\text{cosine\_similarity}(d, q) = 
\frac{\mathbf{d} \cdot \mathbf{q}}{\|\mathbf{d}\| \cdot \|\mathbf{q}\|}
= \frac{\sum_{i=1}^{n} d_i \cdot q_i}
{\sqrt{\sum_{i=1}^{n} d_i^2} \cdot \sqrt{\sum_{i=1}^{n} q_i^2}}$$

Если угол равен 0° (косинус равен единице) — векторы сонаправлены, документы максимально похожи (идеальная релевантность). Если угол равен 90° (косинус равен нулю) — векторы перпендикулярны, документы абсолютно не связаны.

При всем удобстве, у векторного представления Салтона есть несколько очевидных минусов. Давайте их перечислим. 

Во-первых, такие вектора огромны. Они по определению соразмерны количеству слов в словаре, а у нас только в английском языке насчитывается более 1 млн словоформ. Решать эту проблему может грамотный препроцессинг текста, нацеленный в том числе на уменьшение кол-ва учитываемых словоформ в словаре. __Стеминг__ - оставляем только корень слова. __Лемматизация__ - приводим слово к его канонической форме (для существительного это единственное число именительный падеж, для глагола настоящее время и так далее). Также можно поставить лимит по частоте - в словарь добавляем только самые частотные слова. Эффективность этого процесса отличается от языка к языку, но в среднем на 30% можно сократить размер словаря. но это скорее купирует проблему, а не решает ее фундаментально.

Во-вторы, такие представления почти полностью состоят из нулей. Их неэффективны с точки зрения хранения и обработки. Решить можно инженерно, например, использовать для хранения сжатые представления (sparse matrix). Суть в том, что хранятся только ненулевые значения веткора.

В третьих, это представление заточено искоючительно под синтаксис. Семантика (смысл) вообще никак не учитывается. Для поиска по словам это норм, для задач, требующих глуюокого понимания языка, это слишком поверхностно. При прямом сравнении двух представлений вообще не учитывается синонимию: два слова с близким смыслом могут почти не пересекаться по конкретным контекстам.

Ну и наконец, не все оси несут одинаковую полезную нагрузку. В векторном представлении бывает много шума, слов, по которым невозмонжо восстановить содержание документа. Нужно научиться их как-то обходить. Проблему шума частично решает взвешивание частот по TF-IDF, о котором поговорим в следующем разделе.

### Взвешиваение по TF-IDF
На практике одной частоты в качестве описания токена не достаточно. Первое с чем мы сталкнемся, что самые частые слова («и», «в», «не») встречаются повсюду и почти ничего не говорят о содержании документа, тем не менее получают большой вес. в 1972 году Краен Спарк Джонс предложил следующую идею: давайте при описании текста использовать не просто частоты слов (TF), а будем домножать частоту на некоторый нормирующий коэффицент, который будет выполнять роль штрафа за слишком большую распространенность слова. Так мы изабвимся от влияния априорных частот слова текстового вектора, а совпадение двух документов по редким словам "" будет гораздо более сильным сигналом, чем совпадение. 

Такая логика присвоения весов получила название **TF-IDF** по двум её компонентам, которые пееремножаются: "term frequency" (частоте токена в документе), и "inverse document frequency" (обратной чатсоте токена в корпусе). Общая формула выглядит так:

$$\text{TF-IDF} = \text{TF}(t, d) \cdot \text{IDF}(t)$$

**TF(t,d)** — частота слова $t$ в данном документе $d$. Чем чаще слово употреюляется в документе, тем больше оснований полагать, что это слово важно для этого текста:
$$\text{tf}(t, d) = \frac{f_{t,d}}{\sum_{t' \in d} f_{t',d}}$$

**IDF(t)** — обратная документная частота слова $t$. Чем в большем числе документов коллекции слово встречается, тем меньше его вес. В целом TF-IDF допускает разные способы подсчета своих компонентов, но приведем классическую формулу, которая встречается чаще всего. Пусть N - это число документов в корпусе текстов, тогда IDF отражает логарифм от обратной доли документов, в которых хотя бы раз используется данный токен $t$:

$$\text{idf}(t, D) = \log \frac{N}{|\{d \in D : t \in d\}|}$$

Роль логарифма в формуле для IDF - смягчать взрывной рост обратной функции на частотах близких к 0.

### Взвешиваение по BM25
Подход TF-IDF был настолько удобен, что быстро стал стандартом в информационном поиске. Как следствие, выросло и количество попыток сконструировать другие способы взвешивания, учитывающие большее количество факторов. Наиболее успешной модификацеий стала метрика __BM25__, которую британские исследователи разработали в 1994 году для своей поисковой системой __Okapi__ (система получила название в честь Окапи - экзотичекого животного, смеси жирафа и лошади). Несмотря на свою громоздкость, формула BM25 имеет ту же структуру, что и TF-IDF, просто в нее входит большее количество факторов:

$$\text{BM25} = \text{IDF}(t) \times 
\frac{f_{t,d} \cdot (k_1 + 1)}
      {f_{t,d} + k_1 \cdot \left(1 - b + b \cdot \frac{|d|}{\text{avgdl}}\right)}$$
BM25 содержит три ключевых изменения по сравнению с TF-IDF. Во-первых, вклад комопнента TF перестал быть линейным, теперь он уменьшается с ростом количества упоминаний. Скорость насыщения задается параметром $k_1$. При больших $k_1$ влкад частоты перестает расти бычтро, а при $k_1 = 0$ мы откатываемся к обычной частоте TF: $$\frac{tf \cdot (k_1+1)}{tf + k_1}$$
Во-вторых, компонент TF сделали зависящим от длины документа $$b \cdot \bigg(1-\frac{|D|}{avgdl}\bigg)$$

В третьих, авторы внесли изменения в расчет обратной частоты компонент IDF. Теперь она выглядит так: $$\text{IDF}(t) = \log \frac{N - n_t + 0.5}{n_t + 0.5}$$
Вместо доли используют вариант из вероятностного варианта соотношение частот. Прибавка 0.5 нужна, чтобы не делить на ноль, а прибавка +1, чтобы логарифм был гарантированно положительным.

Любопытна история названия, которая по сути повторяет ставшую хрестоматйиной в среде маркетологов историю создания очитителя WD-40 - модель назвали BM25 в честь номера версии формулы, которая оказалась наиболее успешной. А BM это акроним для "Best Matching".

Метод быстро опробовали на соревнованиях по информационному поиску (TREC), где он успешно себя показал и постепенно стал заменой для TF-IDF.

### Латентно-семантическое индексирование
К концу 1980-х точный поиск по словам и фразам стал стандартом, но оставалась нерешенной проблема синонимичности: пользователь мог написать в запросе "автомобиль", но документы, содержащие слово "машина" полностью игнорировались. Группа исследователей из комапнии Bellcore пыталсь обойти это ограничение и искали способ представления текста, как векторов в непрерывном пространстве.

И такой способ был найден, его назвали термином латентно-семантический анализ (LSA), или синонимично латентно-семантическое индексированием (LSI) [(Deerwester et al., 1990)](https://doi.org/10.1002/%28SICI%291097-4571%28199009%2941:6%3C391::AID-ASI1%3E3.0.CO;2-9). Термин LSI больше прижился в информационном поиске, а LSA в когнитивной науке и психолингвистике.

Суть метода в том, что они взяли частотное описания набора текстов («Term-Document») $X$ размера $m \times n$, содержащую частоту каждого из $m$ слов в каждом из $n$ документов (возможно взвешенные по TF-IDF) и применили к ней сингулярное разложение (SVD), который декомпозирует произвольную прямоуглольную матрицу в произведение двух ортогональных матриц U и V и диагонального масштабирования $\sigma$:

$$X = U \Sigma V^{\top}$$

Стоит отметить, что в алгебре существует целый класс разложений, решающих подобную задачу приближения произвкедения . К ним относятся, например QR разложение, LU разложение, Cholesky, NMF. Но SVD получило особенную популярность, поскольку обладает сразу несколькими хорошими свойствами. Во-первых, оно гарантирует оптимальность приближения в терминах матричной нормы. Во-вторых, выбирая размер, мы легко регулируем сложность отдельных компонентов. В третьих, SVD универсален и может работать для прямоугольных матриц.

Замечательное свойство данного разложения в том, что диагональная матрица $\Sigma$ состоит из сингулярных значений, упорядоченных по убыванию. Это значит, что оставив только первые $k$ из них (обычно несколько сотен), мы получим приближенное усечённое разложение (__Truncated SVD__). Разложение - наилучшее в смысле нормы приближение оригиниальной матрицы $X$ матрицей ранга $k$:

$$X \approx X_k = U_k \Sigma_k V_k^{\top}$$

Строки матрицы $U_k \Sigma_k$ можно рассматривать как $k$-мерные описания слов, а столбцы матрицы $\Sigma_k V_k^{\top}$ как векторы документов. Зануляя измерения с наименьшими сингулярнми занчениями в U и V, мы как бы выбрасываем весь малозначимый шум и заставляет слова, встречающиеся в похожих документах, получить близкие представления, даже если они ни разу не встретились в одном документе напрямую. 

Сжатые описания образуют непрерывное пространство описаний. Оно абсолютно неинтерпретируемое, но можно точно сказать, что те самые "автотмобиль" и "машина" в этом пространстве будут располагаться где-то близко [объяснить почему]. Вообще концепция непрерывных представлений  - одна из центральных для машинного обучения и нам ещё будет не раз встречаться. Когда мы "запихиваем" разреженный сигнал в вектор, мы тренируем машину "обобщать", без этого не будет работать. Обощение это ключ к абстрактному мышлению.

Первые попытки ещё в 1960-ых применение факторного анализа для автоматической классификации документов (Borko & Bernick, 1963)
В 1992 году LSI впервые применили в системе автоматического назначения рецензентов на научные статьи<br>нужно было сопоставить текстов с экспертами в данной област. В 1994 году выдан патент на применение мультиязычного LSI: сопоставлять по смыслу написанных на разных языках. В 1995 году LSI впервые задействовали для автоматической оценки сочинений, система измеряла семантическую близость студенческого текста с эталонным ответом. К концу 1990-х метод стал повсеместно использоваться на практике.

К минусам подхода можно отнести то, что он остается «Bag-of-Words» моделью, то есть порядок слов никак не учитывается. Плохо справляется с многозначностью, посольку все смыслы слова смешиваются в один вектор. Требует пересчёта разложения при добавлении новых данных.

## Языковые модели

Языковая модель в самом общем смысле — это распределение вероятностей над последовательностями слов: она должна уметь для любой цепочки слов сказать, насколько та правдоподобна.
$$P(w)=P(w_1,w_2... w_n)$$

Эквивалентно, языковая модель предсказывает следующее слово по предыдущим. Ровно эту задачу решают современные GPT модели: по предыдущему контексту определить наиболее вероятное продолжение текста. Менялись только методы оценки этого распределения:
$$P(w_n|w_1 ... w_{n-1})$$

__Андрей Марков (1913)__ заложил математический фундамент. Анализируя чередование гласных и согласных в «Евгении Онегине», он показал, что текст можно описать как стохастический процесс, где вероятность следующего символа зависит от предыдущего (после гласной более вероятна согласная и т.п.). Так появились [цепи Маркова](https://en.wikipedia.org/wiki/Markov_chain) — формальная основа всех последующих моделей

__Клод Шеннон (1948)__, в работе [«A Mathematical Theory of Communication»](https://people.math.harvard.edu/~ctm/home/text/others/shannon/entropy/entropy.pdf), первым явно смоделировал язык вероятностно. Он строил приближения английского текста разного порядка (n-граммы), ввёл понятие энтропии языка и фактически описал теорию, на которой стоит вся область

Сам термин «языковая модель» в современном смысле закрепился позже — в 1970–80-х годах в сообществе распознавания речи, прежде всего в группе __Фреда Елинека__ (Fred Jelinek) в IBM. Там использовалась модель «шумного канала»: P(слова | звук) ∝ P(звук | слова) · P(слова), и вот эта априорная компонента P(слова) — оценка того, насколько правдоподобна сама фраза — и называлась language model. Елинеку приписывают полушутливую фразу о том, что качество распознавания растёт каждый раз, когда из команды уходит лингвист — это маркер перехода от правил Хомского к статистике.

Вероятность последовательности $w = (w_1...w_n)$ раскладывается в произведение условных вероятностей:

$$P(w_1, w_2, \dots, w_n) = P(w_i|w_{1..i-1}) \cdot P(w_2|w_{1..i-2}) = ... = \prod_{i=1}^{n} P(w_i \mid w_1, \dots, w_{i-1})$$

Проблема в том, что вероятность $P(w_i \mid w_1, \dots, w_{i-1})$ зависит от всей предыстории $w_1 ... w_{i-1}$, и таких предысторий бесконечно много — оценить их по данным невозможно. Здесь и вступает [марковское допущение](https://en.wikipedia.org/wiki/Markov_property): будем считать, что следующее слово зависит не от всей истории, а только от последних $n-1$ слов. Это и есть [n-граммная модель](https://en.wikipedia.org/wiki/Word_n-gram_language_model)

При $n = 2$ (биграммы) каждое слово зависит только от предыдущего: $P(w_i \mid w_1, \dots, w_{i-1}) \approx P(w_i \mid w_{i-1})$<br>
При $n = 3$ (триграммы) — от двух предыдущих, и так далее<Br>
При $n = 1$ мы получаем вырожденную модель, где слова считаются независимыми. Это сильное упрощение и на практике от такой модели будет мало пользы

Сами вероятности оцениваются простым подсчётом частот в обучающем корпусе (оценка максимального правдоподобия):

$$P(w_i \mid w_{i-1}) = \frac{C(w_{i-1}, w_i)}{C(w_{i-1})}$$

где $C(\cdot)$ — число вхождений соответствующей последовательности в корпус. То есть «вероятность слова *молоко* после слова *парное*» — это сколько раз встретилось «парное молоко», делённое на то, сколько раз встретилось «парное» вообще.

Плюсы n-gram моделей: 
- прозрачность: никакого обучения в современном смысле, только счётчики. На триграммах такие модели десятилетиями оставались рабочей лошадкой распознавания речи и машинного перевода и были на удивление сильны — настолько, что нейросетевым языковым моделям потребовались годы, чтобы их уверенно превзойти

Но есть и фундаментальные ограничения: 
- в каждый момент времени модель видит только узкое окно в $n-1$ слов. Это сильно ограничивает ее "контекстность"<br>Наивное решение - увеличить $n$, но тогда мы довольно быстро упираемся в потолок комбинаций, которые надо хранить (см следующий раздел)

Стоит отметить, что существуют и другие способы моделирования языка: например, диффузионные подходы, energy-based и прочее. Все они имеют практические воплощения, однако настолько широкую популярность не получили

## Приемы сглаживания
Рассмотрим языковую модель $P(w_n|w_1, w_2 ... w_{n-1})$. При n >> 3 большинство грамматически правильных триграмм в любом конечном корпусе не встречаются ни разу и по формуле максимального правдоподобия получает вероятность ровно ноль. Это плохо: стоит в тестовом предложении появиться одной неизвестной триграмме, и вероятность всего предложения обнуляется

Для борьбы используют сглаживание (smoothing): часть вероятностной массы отбирается у наблюдённых событий и перераспределяется в пользу ненаблюдённых, чтобы нулей не осталось. Простейший вариант — сглаживание Лапласа (add-one): прибавить единицу ко всем счётчикам. На практике оно работает плохо — слишком грубая трансформация.

За время развития направления было предложено целое семейство методов. Некоторые из них:
- Сглаживание Гуда–Тьюринга (Good, 1953)<br>оценивает вероятность ненаблюдённых событий через число событий, встретившихся ровно один раз; одна из первых принципиальных идей в этой области<br>$$c^* = (c + 1) \frac{N_{c+1}}{N_c}$$где с - исходная частота, $N_{c}$ и $N_{c+1}$ - количество уникальных событий, которые встретились в тексте ровно $c$ и $c+1$ раз<br><br>
- Откат Каца (Katz back-off, 1987) <br>если триграмма новая (частота в словаре 0), то «откатываемся» к биграмме, если и та нет — к униграмме<br><br>
- Интерполяция Йелинека–Мерсера <br>линейно смешивает оценки разных порядков (триграмму, биграмму, униграмму) с подобранными весами<br><br>
- Сглаживание Кнезера–Нея (Kneser & Ney, 1995) — вершина этой линии. Ключевая идея тонкая и красивая: вероятность слова в роли «отката» оценивается не по тому, как часто оно встречается вообще, а по тому, в скольких *разных* контекстах оно появляется. Классический пример — слово *Angeles*: оно частотно, но почти всегда идёт после *Los*, поэтому как самостоятельное продолжение произвольного контекста маловероятно. Его интерполированная (модифицированная) версия многие годы считалась state-of-the-art для дискретных n-граммных моделей.

### Метрика перплексии

Раз уж мы строим распределения над текстами, нужна метрика, которая говорит, насколько "хороша" модель. Такая метрика есть — называется **перплексия** ([perplexity](https://en.wikipedia.org/wiki/Perplexity)) 

Метрику впервые предложили использовать в 1977 году в IBM в проектах по распознаванию речи. Идея в том, что она показывает, насколько модель «удивлена» (perplexed) реальным текстом: если наблюдаемый токен имеет в рамках модели низкую вер-ть быть сгенерированным, то удивление большое.

Возьмем отдельный токен $w_i$ из какого-либо предложения тестового корпуса текстов. Подадим в языковую модель его предыдущий контекст $w_1...w_{i-1}$ и посмотрим, какую P(w) возвращает модель. Если вер-ть генерации правильного токена $w_i$ высока, то все хорошо, модель справляется. 

Если к этой вероятности применить отрицательный логарифм, то она трансформирует задачу максимизации в задачу минимизации:
$$-\ln P(w_i \mid w_1, \dots, w_{i-1})$$

Теперь вероятность близкая к 1 означает низкую перплексию, а к 0 высокую:

<img src="img/perplexity.png" width=400>

Таким образом, значение метрики зависит от того: 
1) насколько хорошо модель научилась предсказывать возможные продолжения в тексте
2) что у нас за текст и насколько он нестандартный
  
Для получения средней перплексии на множестве документов метрика усредняется. По всем токенам данного предложения, по всем предложениям в корпусе документов
$$L = -\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})$$

Получаемая величина (Average negative log-likelihood per token), это по сути и есть функция потерь (Loss function) оптимизруемая при обучении LLM. 

Она же кросс-энтропия $H(P^{model}, P^{true}) = \sum P^{true} \cdot \log P^{model}$, при этом $P^{true}$ вырождается в константу "1.0" и получается привычная форма

<img src="img/perplexity2.png" width=400>

Чтобы перевести значение метрики на более интрепретируемую шкалу от этого выражения еще берут экспоненту:

$$\text{PP}(W) = e^{L} = \exp\!\left(-\frac{1}{N}\sum_{i=1}^{N} \ln P(w_i \mid w_1, \dots, w_{i-1})\right)$$

где $N$ — число слов в тесте

Можно $-1/N$ занести под логарифм, тогда экспонента сократится, а вероятности перемножатся в полную вероятность

$$\text{PP}(W) = P(w_1, \dots, w_N)^{-\frac{1}{N}}$$

На этой шкале у метрики есть интерпретация. Перплексия — это число равновероятных вариантов, между которыми модель в среднем выбирает на каждом шаге генерации. Перплексия 100 означает, что модель в среднем колеблется так, как если бы на каждом шаге угадывала равновероятный исход из ста. Перплексия, равная размеру словаря, соответствует модели, которая не выучила вообще ничего (равномерное распределение). Чем ниже — тем лучше: модель уверенно сужает выбор.

Перплексия обычно падает к концу предложения, так как в конце накоплен конекста больше
<img src="img/perplexity_1.png" width=400>

Несколько важных оговорок, без которых перплексией легко злоупотребить:

- Перплексия зависит от словаря. Сравнивать перплексии двух моделей корректно только при одинаковом словаре и одинаковой разбивке на токены<br><br>
- Перплексия - <u>внутренняя</u> (intrinsic) метрика модели. Она говорит, как хорошо модель моделирует распределение текста, но не говорит, как хорошо она решает прикладную задачу (перевод, распознавание речи). Важно смотреть и на внешние (extrinsic) метрики — например, частоту ошибок распознавания слов (WER) в речевых системах<br><br>
- Перплексия тесно связана с кросс-энтропией: $\text{PP} = 2^{H}$ (или $e^H$, смотря какое основание логарифма), где $H$ — кросс-энтропия в битах (натах) на слово. Поэтому в статьях того времени качество часто отчитывали именно в битах на слово

Перплексию удобно испольовать для детекции сгенерированного текста. У ИИ тексты стабильно низкой перплексией, в то время как перпексия более взрывная

### Кластеризация Брауна
Возьмем авторегрессионную языковую модель. Выше мы выяснили, что н-граммные языковые модели требуют хранение гигантского кол-ва парамеров. Идея метода в том, чтобы сократить кол-во вычислений, заменив работу с токенами, на работу с классом токена. Метод предложенный в 1992 году.

Если с помощью отображения $\pi$ мы каждый токен относим в свой класс, тогда авторегрессинная гернерация следующего токена декомпозируется на два шага - выбор класса и выбор слова внутри класса:

$$P(w_t \mid w_{t-1}) \;=\; \underbrace{P\bigl(c_t \mid c_{t-1}\bigr)}_{\text{переход между классами}} \cdot \underbrace{P\bigl(w_t \mid c_t\bigr)}_{\text{выбор слова в классе}}$$

Обе условные вероятности считаются точно так же как обычные н-грамные модели, по статистике корпуса. Общее число параметров сокращается с $|V|^2$ до $C^2 + |V|$.

Вопрос как считать разбиение $\pi$. Будем считать $\pi$ параметром модели. Тогда естественно искать $\pi$, максимизирующее правдоподобие корпуса. Подстановка оценок в логарифмическое правдоподобие и группировка слагаемых дают тождество:

$$\frac{1}{T}\,\mathcal{L}(\pi) \;=\; \underbrace{\sum_{c,\,c'} p(c, c')\,\log \frac{p(c, c')}{p(c)\,p(c')}}_{I(C_{t-1};\, C_t)} \;-\; \underbrace{\vphantom{\sum_{c,c'}} H(W)}_{\text{не зависит от } \pi}$$

Второе слагаемое от разбиения не зависит. А вот первое взаимной информации между классами соседних слов. Оптимальным оказывается такое разбиение, при котором знание класса левого слова максимально информативно относительно класса правого. Это переводит задачу из области языкового моделирования в чисто теоретико-информационную.

Полное решение сложно. На практике применяется жадная агломеративная процедура:
1. Сначала каждое слово считается за отедльный класс.
2. На каждом шаге среди всех пар классов выбирается та, слияние которой влечёт наименьшую потерю взаимной информации (mutual information)
3. Пара сливается; шаг повторяется до достижения *C* классов

Последовательность слияний образует бинарное дерево, листьями которого являются слова. Путь от корня к листу задаёт битовую строку — код слова. Префикс длины *k* определяет класс на соответствующем уровне детализации, что позволяет использовать несколько гранулярностей одновременно:

| Слово | Код | Префикс (4 бита) |
|---|---|---|
| *Monday* | `01101100` | `0110` |
| *Tuesday* | `01101101` | `0110` |
| *January* | `01110010` | `0111` |

Как языковая модель конструкция уступает сглаженным пословным *n*-граммам и на практике применялась лишь в интерполяции с ними. Основную ценность представляет побочный результат — само разбиение. Битовые коды слов, полученные на неразмеченном корпусе, используются как признаки в дискриминативных моделях (распознавание именованных сущностей, морфологическая разметка, синтаксический анализ). Незнакомое в обучающей выборке слово всё равно принадлежит известному классу, что даёт модели информацию о его поведении. Это делает метод одним из первых практических примеров обучения представлений слов без учителя — предшественником плотных векторных представлений, в которых та же дистрибутивная гипотеза реализована в непрерывной форме.

Жёсткость разбиения не позволяет моделировать многозначность; контекст ограничен одним соседом; между классами не определено расстояние, поэтому операции над представлениями невозможны.


## Скрытые Марковские Модели

Не все задачи — это классификация целого документа. Многие требуют разметить каждый токен: приписать каждому слову часть речи (POS-tagging), выделить именованные сущности (NER), сегментировать речь. Доминирующим инструментом для таких задач последовательной разметки в эту эпоху была **скрытая марковская модель** (Hidden Markov Model, HMM).

Есть матрица переходов между токенами. Есть переменная (скрытая, скаляр, дискретная: категории описаны), которая влияет на матрицу переходов. 

Идея: наблюдаемая последовательность (слова) порождается ненаблюдаемой («скрытой») последовательностью состояний (например, частей речи). Модель описывается двумя наборами вероятностей:

- Вероятности переходов — насколько вероятно, что за состоянием «прилагательное» идёт состояние «существительное». Здесь снова марковское допущение: следующее состояние зависит только от текущего.
- Вероятности эмиссии — насколько вероятно, что состояние «существительное» порождает конкретное слово «стол».

<img src="img/hmm1.png" width=550>

Задача разметки — найти наиболее вероятную последовательность скрытых состояний для данной последовательности слов. Решается она эффективно динамическим программированием — [алгоритмом Витерби](https://en.wikipedia.org/wiki/Viterbi_algorithm). Параметры оцениваются либо по размеченному корпусу (простым подсчётом, как у n-грамм), либо без разметки — алгоритмом [Баума–Велча](https://en.wikipedia.org/wiki/Baum%E2%80%93Welch_algorithm) (частный случай EM).

HMM хорошо работали для POS-теггинга и были естественным выбором, пока их в начале 2000-х не стали теснить дискриминативные модели — прежде всего условные случайные поля __Conditional Random Fields__ [(Lafferty et al, 2001](https://daiwk.github.io/assets/Conditional%20random%20fields%20Probabilistic%20models%20for%20segmenting%20and%20labeling%20sequence%20data.pdf)). CRF снимали несколько структурных ограничений HMM (в частности, позволяли свободно использовать произвольные, в том числе пересекающиеся, признаки) и стали стандартом для разметки последовательностей на следующее десятилетие — вплоть до прихода нейросетевых моделей. Но логика «скрытые состояния порождают наблюдения, ищем лучшую цепочку Витерби» осталась важной частью образования любого, кто работает с последовательностями.

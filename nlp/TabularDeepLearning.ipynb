# SAINT (2021)
---
[[paper]](https://arxiv.org/pdf/2106.01342)

Model data
- Tokens = features (categorical or numeric)
- Input = batch of $b$ points
- Output = [CLS] embedding

Tokens are embedded with encoders into d-dimensional vectors. So the embedding matrix is $b \times N \times d$

Architecture:
1. self-attention layer<br>standard Transformer attention
2. intersample self-attention layer<br>used to explore point neighbourhood, for example to better treat missing data
<img src="img/saint2.png" width=750>

Pre-training = self-supervised contrastive learning<br>
"Good" pairs = slightly modified versions of data point<br>
Modification strategies: Cutmix (2019), mixup (2017)

(Optional) fine-tuning = adjust encoding to match specific downstream classification task

# Tabular Text Transformer (2024)
---
[[paper]](https://aclanthology.org/2024.findings-acl.87.pdf)

Architecture that allows to handle both text and tabular data

Key features:
- Two separate streams (text + data) with their joint optimization during training<br>to enhance model reprenetational power
- 2 two-way cross-attentions<br>attention in the "text" stream attends its embeddings to both text + data<br>attention in the "data" stream  attends its embeddings to both text + data
- quantile embeddings for numeric data
- handles uncertainty by applying SHAP values

# TabTransformer (2020)
---

Architecture inspired by Transformers to process tabular (numeric and categorical) data.

Categorical data is embedded using standard embeddings. Numeric data is processed via MLP

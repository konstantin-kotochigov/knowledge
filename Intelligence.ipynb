# Терминология

### Интеллект
В среде экспертов нет единого определния понятия Intellignce, оно разнится в зависимости от области науки его изучающей

<u>По Спирмену (1900)</u><br><i>Интеллект - навык решения интеллектуальных задач (cognitive tasks)</i><br>

То есть умный = успешный (в решении сложных задач)

Определение одновременно маскимально широкое, и при этом наименее противоречивое - мы отталкиваемся сразу от результата. Оно правда зацикленное, но для того времени было достаточно. К когнитивным задачам, как и сейчас, относили всякие тесты на память, логический вывод, итд. Эти тесты удобно использовать как апроксимацию задач из реального мира

Такое определение задает нужные рамки и даже обхясняет направление развития, но оно все же слишком грубое. Чтобы изучать концепт детальнее, его желательно конкретизировать

<u>По APA:</u><br><i>Интеллект - способность понимать сложные концепции, адаптироваться к меняющимся условиям, обучаться, аргументированно обдумывать, обходить препятствия</i>

В этом определении перешли к списку навыков без выделения какого-то основного

<u>По Gottfredson (1994)</u><br><i>Интеллект - умение планировать, аргументированно рассуждать, решать задачи, абстрактно мыслить, понимать сложные концепции, понимать новое и накапливать опыт</i>

Примерно то же, но добавляется планирование и абстрактное мышление

<u>По CHC</u> (модель Cattell-Horn-Caroll):<br><i>Интеллект - система связанных в иерархию интеллектуальных навыков</i>

<u>По Resnick</u><br><i>Интеллект - не просто решение задач, а умение создавать новые задачи</i>

Тут акцент на субъектность, жажду познания

<u>По Legg & Hutter (2007)</u>:<br><i>
Умение достигать цели в широком спектре вводных</i> (акцент на универсальность)

Видно, что разные эксперты фокусируются на разных аспектах. Если попытаться вывести общей знаменатель, это скорее всего будет что-то про:
- обучаемость
- адаптируемость
- умение строить логический вывод
- навык решения задач в новых условиях

Два наблюдения:
1) если начиналось все с определения интеллекта, как некой скалярной величины, то сейчас все согласны, что это сложная многоаспектная система, которую надо изучать с разных углов. можно иметь высокие навыки по одному асекту и низкие по другому
2) не уйти от привязки к человеческой деятельности как центральному критерию (problem solving skills), интеллект - это совокупность навыков, которая позволила равиваться Homo Sapience как виду

К размышлению<br>
- животные также обладают большей частью способностей<br>
- а если ты не решаешь задачи, значит глупый? Или решаешь, не прибегая к интеллектуальным способностям (интуицией или грубой силой)

Наук, связанных с изучением интеллекта, довольно много:
- психология 
  - когнитивная психология<br>восприятие
  - эволюционная<br>как интеллект развивается через поколения
  - дифференциальная<br>исследуются индивидуальные особенности интеллекта
  - образовательная / развивающая<br>как интеллект развивается в процессе обучения
- нейронаука<br>биологические аспекты процесса мышления
  - когнитивная нейронаука
- антропология / генетика
- комтьютерная когнитивная наука<br>моделирование человеческого интеллекта

## Искусственный интеллект

Искусственный интеллект (Artificail Intellignece) - термин, известный с 1950 годов, и получивший очередной виток популярности в начале 2020-х годов, когда в широкое употребление вышли продукты, достигшие качества максимально близкого в сравнении с человеком 

Термин вытеснил два других термина "машинное обучение" и "глубокое обучение"<br>Глубокое обучение, потому что нейросети показали свою эффективность как математическая модель и стали просто использоваться везде по умолчанию - термин стал не нужен<br>Машинное обучение, так как появилась возможность решать более широкие и практико-ориентированные задачи (чат-боты, генерация видео, аудио и тд). В итоге стали больше говорить о продуктах, а не о моделях, и логично стало называть такие системы зонтичным термином "искусственный интеллект"

Так как нет консенсуса относительно того, как определять естественный интеллект, произодные термины типа AI еще более неоднозначны. Кроме того по мере реализации идей он понемногу ээволюционировал

Философские вопросы были всегда. Они активизировались в 1940-х, когда . В частности их формулировал Тьюринг в работе "Can Machines Think"

<u>Дартмутская конференция</u>

Первое использование термина AI было на Дартмутской конференции (1956), Великобритания. Авторы доклада (Маккарти, Мински) попытались заложить основы будущего AI, сформулировав гипотезу, что любой мыслительный процесс может быть формализован, а значит машина может научиться его повторять. Основной посыл: "Машины должны научиться решать задачи, которые интеллектуально доступны только человеку"

Доклад в каком-то смысле сформулировал масштабный roadmap на будущие исследования. В него в частности входила оценка потенциала использования нейросетей, желание научить машину естественному языку, вывод закономенрностей, распознавание образов, владение абстркациями. 

Значительная часть этого roadmap-а воплотилась: модели действительно активно используют язык, выигрывают в интеллектуальных играх, нейросети стали повсеместными, обучение - неотъемлимая часть построения любой модели, самосовершенствование (доступна самокритика) 

Что-то не воплолитлось. Те методы, которыми задачи в итоге решались, оказались более приземленными - вместо вывода формальной модели всего на свете машины полагаются на статистический вывод. Хотя идеи на тот момент выглядели разумными, быстрого роста не получилось из-за того что скорее всего уперлись в ограничения масштаба. Нейросети только спустя 50 лет стали более или менее эффекинво функционировать

<i>вместо мышления модедирование человеческого мышления</i>
Это дло фундамент для самого простого определния: искусственный интеллект = попытка повторить человеческий интеллект

<u>По Мински</u><br>
<i>Искусственный интеллект - способность решать задач, которые требуют человеческого интеллекта, когда решаются людьми</i>

<u>По Норвигу</u><br>
<i>AI изучает агентов, которые взаимодействуют с внешней средой и реагируют, оказывая на нее влияние</i><br>
Тут видно желание сделать определение более общим и наделить интеллект агентностью (идеи из мира reinforcment learning)

Можно попробовать выделить общий знаменатель: явную привязка к естественному интеллекту. Искусственный интеллект - это способность моделировать естестнвенный интеллект

Однако для AI человеческий интеллект - ориентир, но не цель. Во-первых многие произовдители AI ставят задачу превзойти человеческий интеллект, Во-вторых Нет потребности мимикрировать под человеческого мозга / мышления (есть много компаний, которые пытаются воспроизвести биологически достоверную модель мышления, но пока не очень успешно). Инженерно довольно сильно отличается. Например, технически AI априори будет лучше по скорости signal processing, по short0term memory и прочее.

## Теории интеллекта

Психометрия - наука о численной оценке когнитивных способностей

#### Spearman (1903)
Да, это тот Спирммен, который придумал ранговую корреляцию. В целом изучение моделей интеллекта дало мощный буст развития статистике и факторного анализа в частности

Одно из первых исследований. Спирмен исследовал результаты выполнения когнитивных тестов и заметил, что оценки по разным тестам довольно сильно коррелируют => это говорит о существовании общего показателя, который он назвал G-фактор (general intelligence factor)

Этот факт с оговорками переносится на бытовой уровень - у отличника хорошие отметки будут скорее всего по всем предметам

Главное предположение: мозг - не модульная конструкция (блок речи, блок математики и тд), а все же единая среда и какие-то механические факторы влияют на решение отдельных задач

В то же время невозможно отрицать наличие людей по-разному способных к разным задачам - нескотря на общее, есть и различия<br>
Экстремальный пример - саванты, у которых непропорционально развивается какая-то одна способность

Понимая это, Спирмен все же выделил в интеллекте отдельные способности - s-факторы (specific factors). В его двухуровневой иерархии были: 
- общий интеллект G
- частные способности s

<img src="img/s_factors.jpg" width=300>

Все последующие исследования были про попытку найти какой-то независимый базис из отдельных когнитивных способностей. Спойлер: полностью независимый найти не получилось, так что Спирмен был прав насчет существования G-фактора

есть гипотеза, что составление сложносочиненных предложений сродни пространственным манипуляциям с объектами, возможно работают одни и те же нейроны. Это могло бы объяснить буст качества искусственного интеллекта, когда появились LLM-ки - типа в воспроизведении речи участвуют те же моделирующие механизмы

#### Thurstone (1930)
В начале полемировал со Спирменом. Он также исследовал факторным анализом, что влияет на интеллект и вместо одного фактора выделил набор факторов

В частности он выделял отдельные способности: вербальные, числовые, пространственные, память, беглость, рассуждение

Вскоре однако их позиции сблизились - есть и общий фактор, и индивидуальные

С возрастом влияние фактора g уменьшается, так как интеллект становится более разносторонним

#### Гилфорд (1950)
Он решил пойти своим путем, не копался в результатах тестов, а предложил кубическую теорию интеллекта. В ее рамках он выделил 3 измерения:
- __сигнал__<br>с чем работаем
- __операции__<br>что нужно делать
- __форма результата__<br>что получается

Всего получается 120 вариантов интеллекта

Главная проблема: не понятно как это обосновывать, где гарантия независимости факторов

Кроме того её сразу стали уточнять/дополнять, по модальности сигнала и прочее

#### Айзенк ()
Его теория во многом повторяла теорию Гилфорда, но с меньшим кол-вом измерений

Но известен он тем, что предложил использовать IQ-тесты для определния интеллекта

Он выделил такую способность, как "скорость интеллекта". А также "настойчивость"



#### Гарднер (1983)
У него был творческий бэкграунд, поэтому он критиковал холистический подход к интеллекту и особенно IQ тесты. 

Он предложил теорию множественного интеллекта, в рамках которой выделил целый набор интеллектов (часто творческих)
- __вербальный__<br>
- __логическо-математический__<br>
- __музыкальный__<br>наличие слуха
- __тактильный__<br>танцы, спорт, актерство
- __межличностный__<br>социальный, умение коммуницировать
- __внутрилинчостный__<br>осознанность, саморазвитие
- __природный__<br>Беар Грилз
- __экзистанципальный__<br>умение философствовать
- __педагогический__<br>талант объяснять

Почему нет древоколотельного? Выглядит список странно, многое их этого похоже просто на личные качества, но он сыграл важную роль в критике IQ тестов и подтолкнул к более всесторонней оценке индивидума

#### Стернберг (1985)
Первая разработка - триархическая модель мышления. В ней он перешел на чуть более высокий уровень навыков и выделил базис из трех моделей поведения:
- __аналитическая система__<br>проблемно-ориентированный подход - планирование, принятие решений
- __практическая__<br>умение решать задачи в ежедневной жизни
- __творческая__<br>условные soft-skills, умение действовать как в незнакомых ситуациях, так и интуитивно в знакомых ситуациях

Также он предложил концепцию "успешного интеллекта". 

Можно смотреть на интеллект немного под другим углом (чуть ближе к биологии) - как процесс
Отвел особую роль планированию действий. Это противоречит такому фактору как скорость
Отводил особую роль новизне задачи

#### Catell (1940)
Ученик Спирмена. Известен прежде всего 16-факторная модель личности, но интеллект также изучал

Выделил две состовляющих: 
- Gf Fluid Intellect<br>умение решать задачи
- Gc Crystalized Intellect<br>полученные знания, соответствие культурной среде

Удобная прокси к кристализованному интеллекту (Gc) - это средний балл за учебу. Мигранты заведомо хуже.

#### Horn (1960)
Дополнил $g_f$ и $g_c$ еще 10 факторами. Общего фактора (g) не было

#### Caroll (1993)
Сфокусировался на мета-исследовании существующих работ, отсмотрел 400 штук. В итоге сформулировал стрехстатную модель интеллкта

#### CHC (1990)
Общая модель, которая на текущий момент считается наиболее употребимой на практике. Появилась в 1990-х, когда смопилировали результаты трех исследований Кейтелла и факорную модель Хорну-Кэроллу. Модель назвали CHC, которая используется до сих пор при оценке интеллекта

Она хороша тем, что подключать новые факторы

Типовой набор факторов:
- $G_f$ - fluid
- $G_c$ - crystalized
- $G_{rw}$ - read / write
- $G_{wm}$ - working memory
- $G_v$ - visual
- $G_f$ - fetching
- $G_a$ - audial
- $G_s$ - processing speed
- $G_t$ - decision time

внизу эмпирические рценки конкретных задач (умение повторять ряд задом наперед)

#### Когнитивные навыки
Разные работы предлагают свои таксономии интеллектуальных навыков, но если выделять те, которые используются чаще всего в IQ тестах:
- вербальный интеллект (значения слов)
- гибкость (умение находить паттерны)
- геометрическое мышление (3D восприятие)
- краткосрочная память (умение запоминать)
- скорость реакции
- арифметика
- аудио

Общая эрудиция

#### Эмоциональный интеллект 
EI - умение считывать и объяснять эмоции. Термин из 1960го, но получил популярностсь в 1990-х годах.  Считается важным навыком для развития Homo Sapience, так как полезен для кооперации внутри сообщества

#### IQ тесты
Появились в 1905 году. К 1930 году их доработали к виду, известному по сегодняшний день



#### Креативность 
Умение создавать что-то одновременно новое (original) и полезное

# Подходы к Evaluation

Как и любой термин, вводившийся для обозначения существующего явления, он допускает широкую трактовку

До сих пор нет консенсусного определения сильного искусственного интеллекта. Те, что есть, довольно широкие (местами ближе к философии), что затрудняет их практическое применение - не понятно, как измерить соотвествие системы этому криетерию. 

Понятно почему: нет общепринятого опредления искусственного интеллект и более того определения интеллекта человеческого очень сильно разнятся

Похоже на термин тест Тьюринга - когда он вводился, он выглядел достаточно четким криетрием. Когда нужно использовать на практике, приходится делать много оговорок - какие-то вариации теста проходятся, какие-то нет

Более или менее все согласны, что мы должны отталкиваться от единственного доступного нам примера полноценного интеллекта - челеовеческого (что конечно слегка по-снобски, но раз мы искусственный интеллект создаем, то логично)

### AGI evaluation (2025)
[[paper]](https://arxiv.org/pdf/2510.18212)

__Идея:__ существующие системы оценки генеративных AI моделей оценивают все вместе, без фокуса на отдельные навыки. В то время как в тестах оценки человеческого интеллекта (экзамены общего интеллекта, IQ тесты и тому подобное) давно используется иерархическая модель - вопросы подораны так, чтобы иожно было независимо оценить разные аспекты интеллекста, которых несколько десятков штук

Авторы предлагают оценивать AGI-ность системы так же, как это делают для естественного интеллекта и заимствуют CHC модель интеллекта

Зачем  при оценке машины ориентироваться на тест, созданный для человека? 
- Краткий аргумент: других ground truth примеров интеллекта нет
- Почему именно CHC - потому что это результат примерно 100 ста ислледования интеллекта

Они оценивают по 10 критериям, каждый из которых берется с 1 весом

Авторы применили метод оценки к существующим моделям (ChatGPT 4, ChatGPT 5) и отмечают, что навыки распределены очень неравномерно (Jagged). Что очеидно
Например, модель не запоминает Memory Storage score = 0%

Единственное, модель интеллекта по CHC довольно всесторонняя, поэтому авторы исключили из списка тестов физиологические навыки и оставили чисто интеллектуальные

Авторы подчеркивают отсутствие четкого опредления, и аппелируют к довольно общему: AGI это AI система, которая решает когнитивные задачи >= среднестатистический взрослый, и которая обладает такими качествами как 
- versatility (широта знания)
- profficiency (умение решать)

Они оценили существующие модели по их варианту критерия и получилось __27%__ для ChatGPT-4 и __57%__ для ChatGPT-5

## Turing Test
В 1949 году в своем эссе на тему перспектив использования компьютеров Тьюринг задавался вопросом, которое собственно вынес в название "Can machines think?". Он заменил нечеткий функциональный подход к определению интеллекта на проблемно-ориентированный: вместо ответа на заглавный вопрос он предложил рассматривать прокси-вопрос - могут ли машины имитировать человеческое поведение?

Предложил использовать известную Party Game: по текстовому взаимодействию (через записки) с двум игроками определить кто из них мужчина, кто женщина. Всего встречалось три варианта формулировки
- может ли машина сыграть в эту игру в роли одного из игроков незамеченой
- может ли ведущий отличить человека от машины
- может ли машина запутать жюри, что это человек

Из постановки предполагается, что AI владеет вербальным интерфесом. Первыми такими системами были ELIZA (по всей видимости, отсылка к Бернарду Шоу) и PARRY (аналог психотерапевта)

### Мотивация

Направление мысли в то время: если машины могут все вычислить, то достаточно формализовать когнитивный навык (например, язык) и ничего не будет мешать этот навык запрограммировать. В реальности "оцифровать" навыки океазалось сложнее и и практика показала, что функциональные подходы скорее уступили гонку статистическим

С точки зрения философии вся дискуссия после появления этого тезиса шла про то, допустимо ли считать, что имитация = мышление или это слишком грубое приближение? Актуальным вопрос остается только в домене философии, на практику особо не влияет

В качестве критики привязки к человечемкому опыту: успех в авиацию пришел только после того как начали изучать аэродинамику, а не пытаться симитировать полет птицы

Тест допускает некоторую широту постановки. Например, доля выигрыша увеличивалась, если ввести в заблуждение (сказать, что не родной зык). Также сильно влияло знание/незнание о наличии среди ипытуемых робота

У теста есть score, частота прохождения. Сам Тьюринг давал прогноз 30% к 2020. Куцвейл заключил пари, что к 2030 году тест будет пройден

Первая LLM-ка, прошедщая тест - ChatGPT-4 в 2024 году<br>
В 2025 ChatGPT прошел. Ещё прикол, оказалось, что сеть маскируется под человека лучше, чем сам человек

__Обратный тест Тьюринга__ - когда не человек пытается определить искусственность, а машина определяет "человечность". CAPTCHA - самое практическое применение теста (визуальный вариант)

Некоторые вариации теста:
- тест со специалистом по доменной области
- тест на копирование стиля
- тест на сжатие информации
- мета-тест
- MIST
- тест на сборку

## Реализации

В основном таким занимается академия. Например, в течении 30 лет проводили премию Лейбнера - ежегодный конкурс по прохождению теста. Коммерческие производители такого не проводят, им это не интересно. С какого-то момента это относительно несложно + 

Регламент везде устанавливается свой<br>
Судья может быть один, а может быть жюри<br>
Судья выбирает из двух вариантов (знает что есть один робот и один человек)

Варианты метрики:
- Win-Rate: процент разговоров, в которых судьи предпочли робота человеку<br>если 50%, то тест пройден - робот неотличим
- Accuracy, если считаем ошибки равноценными 
- 1 - Precision: среди принятых за человека доля машин
- 1 - Recall: доля роботов, принятых за человека

Минимальный Win-Rate (частота обмана), который можно считать успешным, сам Тьюринг оценивал в 30%, сейчас 50%

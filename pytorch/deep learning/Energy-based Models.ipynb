The development of neural models after 2000 highlighted the difference between __discriminatory__ and __generative__ Machine Learning. While they both model "reality" based on available data, they do it in a slightly diffrerent way. Discrimatory models focus on describing class boundaries while generative models focus on learning true data distribution

The goal of any generative model is to find appropriate data distribution

# Hopfield networks (1982)
---
Hopfield network contains one input layer and no output. It models one-level interconnections between the dimensions of the input. 

<img src="img/hopfield1.png" width=200>

HN is a smart way to learn input data distribution - if the incoming signal fits the model well, the model is awarded, otherwise penalized. Over time network starts to mimic input distirubtion

The overall fit is evaluated with an energy function (as opposed to the loss function in discriminative models)

Models input distribution by modeling inner correlations in signal

In the foundational form of Hopfield networks the neurons $h_{i} \in \{0,1\}$ are binary, and connections $w \in \mathbb{R}$

Since inference is iterative, there is also a bad visual representation as a recurrent process

<img src="img/hopfield2.png" width=300>

#### Relation to Associative memory
Moeldes one of the earliest systems of associative memory = auto-complete

#### Relation to Physics
Mimics the principles from Physics<br> Any system tends to minimize its energy: gas occupies all available space<br> Gas under pressure shows high energy (many particle collisions)

Bolzman distribution = possible energy values are distibuted as an inverse exponent

#### Energy
Energy of a network (conceptually) = level of system accordance<br>
Energy of a network (formally) = sum of dot-products of all neuron pairs

$$E=-\sum w_{i,j} h_i h_j \rightarrow \min$$

Energy landscape is defined by the values of the weights

Local energy minimum = stable state aka Memory

One or many patterns might be encoded via training, but not exceeding maximum limit.

#### Training
Training is performed by setting the connection weights to some value<br>Unlike traditional ML models where training is iterative, here training is done in one step

Suppose we have some pattern we need to encode $\zeta = (\zeta_1 ... \zeta_n)$

This encoding will corespond to local minima - change will change the sign and make accordance less

We set each weight $w_{i,j}$ of the network to a dot-product of dimensions $\zeta_i \cdot \zeta_j$<br>
When more than one patterns have to be encoded => just average them $w_{i,j}=\sum \zeta_i \zeta_j$

#### Inference
Make initial run and calculate energy
Iteratively update weights until Energy stabilizes at minimum
Comes to the closest local minima. If many patterns were encoded durnioig training, system is not guaranteed to converge to local minima, it can hallucinate


# Bolzman machines (1985)
---
Bolzman Machine is an advanced version of a Hopfield network. It shares the same architecure: binary neurons, real connections, same definition of Energy. 

But while Hopfield networks are deterministic, Bolzman machines introduce stochsticity: they evaluate the density of energy values using Bolzman distribution

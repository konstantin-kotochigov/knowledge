# Pruning Techniques

Pruning = zeroing unimportant weights or blocks of weights in a network. Usually performed as a post-training activity, but sometimes during training. 

Makes sense if we prune the majority of weights (for example 80%).

Technically implemented by multiplying layer element-wise on "mask" binary matrix

Pupose:
1. To make model smaller => lower latency, lower memory footprint, less storage, more devices available<br>why smaller? because weights are stored in memory / on disk in sparse format
2. To prevent overfitting / make model generalize better<br>

```python
mask = (weight.abs() > threshold).float()
weight.mul_(mask)
```

Pruning stage:
1. post-training
2. in-training

All pruning methods reduce model size (N) and memory usage (M) because in most modern formats weights are stored as sparse stractures

# Post-training pruning
---

### Magnitude-Based
Remove weights with smallest absolute values post-training.  Implemented by applying sparsity mask to trained weights (global or per-layer); no architectural change

### Structured
Removes entire neurons, filters, or attention heads based on their importance metrics (modifies network structure)  

#### Lottery Ticket Hypothesis (Sparse Re-init)
Finds a sparse subnetwork that performs as well as the full model. Train dense model → prune → reset surviving weights to original init → retrain

(very strange name) Lottery ticket because it relies on random weight selection

# In-Training pruning
---
#### Dynamic Sparse Training

Maintains and updates sparsity throughout training. Train with a sparse mask and dynamically regrow/prune connections<br><br>
- __SET__ Sparse Evolutianary TReaining<br>Every few steps of training remove bottom-K weights and unfreeze random K weights<br><br>
- __Regl__ Rigged Lottery<br>Every few steps of treaining remove bottom-K weights and unfreeze weights with top-K gradients<br><br>
- __SNIP__ Single-shot network pruning<br>Prior to training run a single forward-backward pass and remove bottom-K weightes by gradient by weight value<br><br>
- __GraSP__ Gradient Signal Preservation<br>

#### Regularization (L1, Elastic Net)
Encourages sparsity via loss penalties. Adds L1 (or Elastic Net) term to loss function during training

#### n:m Structured Sparsity
Enforces fixed patterns like 2 non-zero weights per 4. Trains/prunes with block constraints (e.g., 2:4) using specialized hardware support. Requires compatible hardware (e.g., NVIDIA Ampere+)  

#### Variational Dropout / Bayesian Sparsity
Learns to drop weights by modeling uncertainty. Uses stochastic weight distributions (e.g., Gaussian) and optimize uncertainty.  

#### Low-Rank Factorization
Replaces weight matrices with products of low-rank matrices. Decomposes weight matrix (e.g., W = A × B) with small inner dimension.  

#### Quantization-Induced Sparsity
Quantization causes many small weights to round to zero. Applies aggressive quantization (e.g., 2-bit); prune zeroed weights.

__Naive implementation:__ try to apply pattern to each starting point in the original string<br>If the pattern breaks move to the next character

It has quadratic complexity as it requires O(mn) operations

```
# scan the whole string in O(n)
for i in 1..n:
    
    # scan the following |W| chars if it matches the pattern
    for j in i..i+w:

        # until first non-matching char
        if s[i] = s[j]:
            continue
        if s[i] != s[j]:
            break
```

__KMP implementation:__ try to apply the pattern to each starting point, but when done (fails or succeeds) omit some redundant comparisons that have already been done. 

Why we can do it? We already matched part of the pattern. To check if we can find other prefixes in the explored substring, we dont have to parse the original string, it can be seen from the pattern itself

Necessary condition = prefix must match the suffix. If it does not - there is no reason to check the pattern it breaks

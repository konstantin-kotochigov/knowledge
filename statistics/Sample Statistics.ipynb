In this notebook I gonna explore two most important estimates in statistics: sample mean and sample variance and their properties

# Introduction
---
Sample mean<br>
$\hat{\mu} = \frac{1}{n} \sum_{i=1}^n X_i$<br>

Sample variance<br>
$\hat{s}^2 = \frac{1}{n} \sum_{i=1}^n (X_i-\hat{\mu})^2$

Notice that sample variance is based on the sample mean $\hat{\mu}$<br>Since sample mean is not perfect itself, sample variance is biased<br>$\mathbb{E}(\hat{s}^2) \leqslant \sigma^2$<br>

Intuitively, the distances are centered around the biased estimate, not the real mean => variance is underestimated

To overcome this bias a new adjusted estimate was proposed<br>
$\hat{s}_{unbiased}^2 = \frac{1}{n-1} \sum_{i=1}^n (X_i-\mu)^2$

Luckily the correction requires very subtle change - we replace the denominator $n$ with a smaller value $n-1$ (proof will be given below):

$$
\mathbb{E}[\hat{s}_{unbiased}^2] = \mathbb{E}\left[\frac{1}{n-1} \sum_{i=1}^n (X_i - \bar{X})^2\right] = \sigma^2
$$


## Sample Mean
---

### Mean
The estimate misses symmetrically around the true value<br>
$\mathbb{E}[\overline{X}] = \mu$<br>

<u>Proof</u><br>$\mathbb{E}[\bar{X}] = \mathbb{E}\left[\frac{1}{n} \sum_{i=1}^n X_i\right]
= \frac{1}{n} \sum_{i=1}^n \mathbb{E}[X_i]
= \frac{1}{n} \cdot n \cdot \mu = \mu$

Important corrollary: sample mean is unbiased!

Not to confuse with Law of Large Numbers: estimate converges to $\mu$ as sample size goes to $\infty$<br>
$\overline{X} \rightarrow \mu$

### Variance
Variance of the sample average (aka standard error of the mean) is proportional to the variance of the variable iteself $X$ (scaled to the sample size)<br>
$\text{Var}(\overline{X})$ $=\frac{\sigma^2}{n}$

<u>Proof</u><br>
$\text{Var}(\hat{\mu})
= \text{Var}\left(\frac{1}{n} \sum_{i=1}^n X_i\right)
= \frac{1}{n^2} \sum_{i=1}^n \text{Var}(X_i)
= \frac{1}{n^2} \cdot n \cdot \sigma^2
= \frac{\sigma^2}{n}$

The proof is based on the fact that scaling factors inside the variance are extracted as squares

### Consistency
The estimate converges towards the real value as sample size grows<br>
$\bar{X} \xrightarrow{p} \mu \,\, \text{as} \,\, n \to \infty$<br>

This repeats the formulation of the Weak Law of Large Numbers (wLLN)

Let's apply Chebyshev inequality to evaluate the probability of extreme values:
$$P(|\bar{X}_n - \mu| \geq \varepsilon) \leq \frac{\text{Var}(\bar{X}_n)}{\varepsilon^2} = \frac{\sigma^2}{n \varepsilon^2}$$

See the probability depends on the sample size. Let's write down the limit:
$$\lim_{n \to \infty} P(|\bar{X}_n - \mu| \geq \varepsilon) \leq \lim_{n \to \infty} \frac{\sigma^2}{n \varepsilon^2} = 0$$


### Efficiency
$\overline{X}(x)$ is the <u>best estimator</u> in class of linear unbiased estimators (BLUE)<br>

First notice that sample average $\hat{\mu}(x)$ is an OLS estimator for a set of data - it minimizes the variance of this data (aka squared error)<br>This can be easily shown by minimizing the sum of squares:<br>$\sum_{i=1}^n
  (y_i - \hat{y})^2 \rightarrow \mathrm{min}$<br>$\hat{y} = \frac{1}{n} \sum_{i=1}^n y_i$

Gauss-Markov theorem states that under certain conditions OLS estimator is BLUE in class of unbiased linear estimators. This means it has minimum possible variance

Those conditions include: linearity, exogeneity, homoskedastiity, full rank. They are all met in our setting<br><br>

If X is normally distributed then it is also MVUE <u>efficient</u> in terms of Rao-Cramer boundary - best in the class of unbiased estimators

Let's write down the loglikelihood function of the normal distribution
$\log L(\mu) = \sum_{i=1}^n \log \left[ \frac{1}{\sqrt{2\pi \sigma^2}} \exp\left( -\frac{(X_i - \mu)^2}{2\sigma^2} \right) \right]
= -\frac{n}{2} \log(2\pi \sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^n (X_i - \mu)^2
$

Now let's take the second derivative with respect to parameter $\mu$:<br>
$\frac{\partial^2}{\partial \mu^2} \log L(\mu) = -\frac{n}{\sigma^2}
$

Notice it is a constant. Let's take an expectation over X:<br>
$\mathcal{I}_n(\mu) = - \mathbb{E} \left[ \frac{\partial^2}{\partial \mu^2} \log L(\mu) \right] = \frac{n}{\sigma^2}
$

Then by Rao-Cramer theorem the variance of the estimate is limited by:
$$\text{Var}(\hat{\mu}) \geq \frac{1}{\mathcal{I}_n(\mu)} = \frac{\sigma^2}{n}
$$

It can be clearly seen that this boundary matches our estimator variance $\mathrm{Var}(\bar{X})$


### Asymptotic normality
---
Let's recall the definition.
Asymptotic Normality = property of some estimator $\hat{\theta}_n$ which states that its distribution converges to a Normal distribution $\mathbb{N}(\mu, \sigma)$ as sample size grows $n \rightarrow \infty$

Most basic statistics (including sample mean and sample variance) hold this property, which directly follows from LLN and CLT theorems

Suppose we compare some estimate $\hat{\theta}_n$ to the real value $\theta$<br>If its error is distributed as $(\hat{\theta}_n - \theta) \rightarrow \frac{1}{\sqrt{n}} N(0,\sigma^2)$ then we consider $\theta$ <u>asymptotically normal</u>. Notice that for larger sample sizes estimate variance gonna reduce thus it must include scaling factor $\sqrt{n}$

---

Let's show that sample mean asymptotically follows the normal distribution<br>
$\sqrt{n}(\bar{X} - \mu) \xrightarrow{d} \mathcal{N}(0, \sigma^2)$

Write down the standardized version of the sample mean $\hat{X}$. Notice the variance of the average incliudes square root of n<br>
$Z = \frac{\sqrt{n}(\bar{X}_n - \mu)}{\sigma} = \frac{1}{\sqrt{n}} \sum_{i=1}^n \frac{X_i - \mu}{\sigma} = \frac{1}{\sqrt{n}} \sum_{i=1}^n Y_i$<br>

CLT directly states that the sample average of some variable is asymptotically normal<br>
$\frac{1}{\sqrt{n}} \sum_{i=1}^n X_i \xrightarrow{d} \mathcal{N}(0,\sigma^2)$<br>

Multiplying by $\sigma$ we get the original statement:
$(\bar{X} - \mu) \xrightarrow{d} \mathcal{N}(0, \frac{\sigma^2}{\sqrt{n}})$

If dataset is small and normal the distribution of $\bar{X}$ is Student's t

## Sample Variance
---

In the following paragpaphs we'll use the adjusted version of the sample variance $s^2=s^2_{adj}$

### Mean
The mean of the estimate coincides with the real variance:<br>
$E[s^2] = \sigma^2$

Importanct corollary: estimate is unbiased!

<u>Proof</u>

Let's decompose the sample variance to its simpler form (see Consistency block for details):<br>
$s^2 = \sum \big[\frac{1}{n} (X - \mu)^2 \big] - (\overline{X}-\mu)^2$

Important corollary: Variance of the $s^2$ can be decomposed to the variance with respect to real value - squared shift (standard error)

Now take expectation<br>
$\mathbb{E}[s^2] = \mathbb{E} \big[ \frac{1}{n} \sum (X - \mu)^2 \big] - \mathbb{E}(\overline{X}-\mu)^2$

First argument is an expectation of the sum = sum of the expectations = sum of the variances<br>
$\mathbb{E} \big[ \frac{1}{n} \sum (X - \mu)^2 \big] = \frac{1}{n} \big[ \sum \mathbb{E} (X - \mu)^2 \big] = \frac{1}{n} n \cdot \sigma^2 = \sigma^2$

Second argument is a variance of the sample average<br>
$\mathbb{E}(\overline{X}-\mu)^2 = \frac{\sigma^2}{n}$

Thus
$\mathbb{E}[s^2] = \frac{n}{n-1} \bigg[\sigma^2 - \frac{\sigma^2}{n} \bigg] = \sigma^2$

### Variance
The variance of $s^2$ is more complex - it is based on the 4th moment<br>
$
\operatorname{Var}(S^2)
  = \frac{1}{n}\left( \mu_4 - \frac{n-3}{n-1}\sigma^4 \right)
$ where
$
\mu_4 = \mathbb{E}[(X - \mu)^4]
$

For normal distributions of the data it can be simplified:<br>
$
\operatorname{Var}(S^2)
  = \frac{2\sigma^4}{n-1}
$


### Consistency
The estimate converges to the real variance $\sigma^2$ as $n \rightarrow \infty$:<br>
$s^2_{adj} \rightarrow \sigma^2$

We'll use LLN to show the convergence

<u>Proof</u>

Let's take unbiased sample variance<br>
$s^2 = \frac{1}{n-1} \sum (x - \hat{x})$

Subtract the real mean:<br>
$s^2 = \frac{1}{n-1} \bigg[ \sum \big( (x-\mu) - (\hat{x}-\mu) \big)^2\bigg]$

Open the square:<br>
$s^2 = \frac{1}{n-1} \bigg[ \sum (x-\mu)^2 - \sum 2(x-\mu)(\hat{x}-\mu) +\sum(\hat{x}-\mu)^2 \bigg]$

Notice the third summand is constant:<br>
$s^2 = \frac{1}{n-1} \bigg[ \sum (x-\mu)^2 - \sum 2(x-\mu)(\hat{x}-\mu) + n (\hat{x}-\mu)^2 \bigg]$

Extract sample mean from the second summand:<br>
$\sum 2(x-\mu)(\hat{x}-\mu) = 2 (\hat{x}-\mu) \cdot \sum (x-\mu) = 2 (\hat{x}-\mu) \big[ n \frac{1}{n} \sum x - n \mu \big] = 2 (\hat{x}-\mu) (\frac{1}{n} n \hat{x} - n \mu) = 2n(\hat{x} - \mu)$

Now we can sum the second and the third arguments to simplify:<br>
$s^2 = \frac{1}{n-1} \bigg[ \sum (x-\mu)^2 - 2n (\hat{x}-\mu)^2 + n (\hat{x}-\mu)^2 \bigg] = \frac{1}{n-1} \bigg[ \sum (x-\mu)^2 - n (\hat{x}-\mu)^2\bigg]$

Rewite:<br>
$s^2 = \frac{n}{n-1} \bigg[ \frac{1}{n} \sum (x-\mu)^2 - (\hat{x}-\mu)^2 \bigg]$

LLN tells us that any sample average converges to its real mean => we can replace the first summand with its limit<br>
$\frac{1}{n} \sum (x - \mu)^2 \rightarrow \mathrm{E} (x-\mu)^2 = \sigma^2$

The same LLN applied to the sample average $\overline{X}$ tells us that $\overline{X} \rightarrow \mu$. Thus<br>
$(\overline{x} - \mu)^2 \rightarrow 0$

$s^2 \rightarrow \frac{n}{n-1} \cdot \sigma^2 \rightarrow \sigma^2$



### Asymptotic Normality

---
Let's recall the definition.
Asymptotic Normality = property of some estimator $\hat{\theta}_n$ which states that its distribution converges to a Normal distribution $\mathbb{N}(\mu, \sigma)$ as sample size grows $n \rightarrow \infty$

Most basic statistics (including sample mean and sample variance) hold this property, which directly follows from LLN and CLT theorems

Suppose we compare some estimate $\hat{\theta}_n$ to the real value $\theta$<br>If its error is distributed as $(\hat{\theta}_n - \theta) \rightarrow \frac{1}{\sqrt{n}} N(0,\sigma^2)$ then we consider $\theta$ <u>asymptotically normal</u>. Notice that for larger sample sizes estimate variance gonna reduce thus it must include scaling factor $\sqrt{n}$

---

Let's show that $s^2$ is distributed normally around the real value $\sigma$:<br> that $\sqrt{n} (s^2 - \sigma^2) \rightarrow N(0,\tau)$

Write down the decomposed version of $s^2$ (see details from the preovous paragraph):<br>
$s^2 = \frac{1}{n-1} \sum (X - \hat{X})^2 = \frac{1}{n-1} \bigg[ \sum (X-\mu)^2 - (\hat{X}-\mu) \bigg]$

We can neglect the second argument since it gets elimiated with $n \rightarrow 0$ and focus on the main part - a sample average of the squared error with respect to the real value $\mu$:<br>
$s^2 \approx \frac{1}{n} \sum (X - \mu)^2$

The error is gonna be:<br>
$\sqrt{n} (s^2 - \sigma^2) \approx \frac{1}{n} \sum (X-\mu)^2 - \sigma^2$

Recall the CLT: sample average of any random variable is asymptotically normal for large enough $n$<br>$\sqrt{n}(\overline{X}_n - \mu) \rightarrow N(0,\sigma^2)$

We can apply this theorem to :<br>
$\sqrt{n} (s^2 - \sigma^2) \rightarrow \mathbb{N}(0, Var(X-\,u))$

Thus $s^2$ is asymptotically normal around the true value $\sigma^2$

### Bias

Let's show that sample statistic $s^2$ is biased<br>
$s^2 = \mathbb{E}\left[\sum_{i=1}^n (X_i - \bar{X})^2\right] \neq \sigma^2$

After opening the brackets:
$$\sum_{i=1}^n (X_i - \bar{X})^2 = \sum_{i=1}^n X_i^2 - n \bar{X}^2$$

and taking expectations
$$\mathbb{E}\left[\sum_{i=1}^n (X_i - \bar{X})^2 \right] = \mathbb{E}\left[\sum_{i=1}^n X_i^2\right] - \mathbb{E}\left[n \bar{X}^2\right]
$$

The first argument becomes:
$$\mathbb{E}\left[\sum_{i=1}^n X_i^2\right] = \sum_{i=1}^n \mathbb{E}[X_i^2] = n \mathbb{E}[X_1^2]
$$
$$\mathbb{E}[X^2] = \sigma^2 + \mu^2
$$
$$\Rightarrow \quad \mathbb{E}\left[\sum_{i=1}^n X_i^2\right] = n(\sigma^2 + \mu^2)
$$

The second argument becomes:
$$\mathbb{E}[n \bar{X}^2] = n \mathbb{E}[\bar{X}^2]
$$
$$\mathbb{E}[\bar{X}^2] = \mathrm{Var}(\bar{X}) + (\mathbb{E}[\bar{X}])^2 = \frac{\sigma^2}{n} + \mu^2
$$
$$\Rightarrow \quad \mathbb{E}[n \bar{X}^2] = n \left(\frac{\sigma^2}{n} + \mu^2 \right) = \sigma^2 + n \mu^2
$$

Overall,
$$\mathbb{E}\left[\sum_{i=1}^n (X_i - \bar{X})^2\right] = n(\sigma^2 + \mu^2) - (\sigma^2 + n \mu^2) = (n - 1) \sigma^2
$$

This proves that
$$\mathbb{E}(\hat{\sigma}^2) = (n - 1) \sigma^2 \neq \sigma^2
$$

Putting $(n-1)$ to the left side we get the unbiased version of the estimator
$$\mathbb{E}\left[\frac{1}{n-1}\sum_{i=1}^n (X_i - \bar{X})^2\right] =  \sigma^2
$$

$$\mathbb{E}\hat{\sigma} = \sigma$$

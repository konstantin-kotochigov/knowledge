$$X_n \xrightarrow{P} X \quad \text{if} \quad \forall \varepsilon > 0, \ \lim_{n \to \infty} P(|X_n - X| > \varepsilon) = 0$$

$$X_n \xrightarrow{a.s.} X \quad \text{if} \quad P\left( \lim_{n \to \infty} X_n = X \right) = 1$$

$$X_n \xrightarrow{d} X \quad \text{if} \quad \lim_{n \to \infty} F_{X_n}(x) = F_X(x), \ \text{at all continuity points of } F_X$$

$$X_n \xrightarrow{L^1} X \quad \text{if} \quad \lim_{n \to \infty} \mathbb{E}[|X_n - X|] = 0$$

$$X_n \xrightarrow{L^2} X \quad \text{if} \quad \lim_{n \to \infty} \mathbb{E}[(X_n - X)^2] = 0$$

$$X_n \xrightarrow{a.s.} X \Rightarrow X_n \xrightarrow{P} X \Rightarrow X_n \xrightarrow{d} X$$

## Терминология

Функция $\varphi: I \to \mathbb{R}$ называется **выпуклой**, если для всех $x, y \in I$ и $\lambda \in [0,1]$:

$$\varphi\big(\lambda x + (1-\lambda)y\big) \leq \lambda \varphi(x) + (1-\lambda)\varphi(y).$$

Геометрически: хорда, соединяющая две точки графика, лежит не ниже самого графика между этими точками.

<img src="img/convex.png" width=550>

Эквивалентное условие (для дважды дифференцируемых функций): $\varphi''(x) \geq 0$ везде на $I$.

Примеры выпуклых функций:<br> $x^2$, $e^x$, $-\ln x$ (на $x > 0$), $|x|$, $x \ln x$ (на $x > 0$).

Примеры вогнутых функций:<br> $\ln x$, $\sqrt{x}$, $\sin x$ (на $[0, \pi]$)

---

### Формулировка Неравенства

Пусть $\varphi: I \to \mathbb{R}$ — <u>выпуклая</u> функция на интервале $I \subseteq \mathbb{R}$, а $X$ — случайная величина, принимающая значения в $I$, такая что математические ожидания $\mathbb{E}[X]$ и $\mathbb{E}[\varphi(X)]$ существуют,

$$\text{тогда} \, \varphi\big(\mathbb{E}[X]\big) \leq \mathbb{E}\big[\varphi(X)\big].$$

То есть матожидание преобразованного набора точек Y всегда > преобразованного матожидания точек X

Иллюстрация данного утверждения на графике<br>
<img src="img/jensen1.png" width=250>

Если функция $\varphi$ <u>строго выпукла</u>, то равенство достигается тогда и только тогда, когда $X$ почти наверное постоянна (то есть $X = \mathbb{E}[X]$ п.н.).

Для <u>вогнутых</u> функций неравенство меняет знак на противоположный:

$\varphi\big(\mathbb{E}[X]\big) \geq \mathbb{E}\big[\varphi(X)\big]$

Если расписать расчет матожидания $\mathbb{E}$, то получим две формулировки: дискретную и более общую - неперерывную 

<u>Дискретная форма</u>

Подходит для конечного набора точек $\{X\}$. В матожидания $\mathbb{E}(x) \, \text{и} \, \mathbb{E}(\phi(x))$ подставляем их дискретную формулу: взвешенная сумма этих точек. При равных весах $\frac{1}{n}$ (читай, равномерном распределении аргумента) это просто выборочное среднее.

Если $x_1, \ldots, x_n \in I$ и $\lambda_1, \ldots, \lambda_n \geq 0$ с $\sum_{i=1}^n \lambda_i = 1$, то для выпуклой $\varphi$:

$$\varphi\!\left(\sum_{i=1}^n \lambda_i x_i\right) \leq \sum_{i=1}^n \lambda_i \, \varphi(x_i).$$

В частности, при $\lambda_i = \tfrac{1}{n}$:

$$\varphi\!\left(\frac{x_1 + \cdots + x_n}{n}\right) \leq \frac{\varphi(x_1) + \cdots + \varphi(x_n)}{n}.$$

<u>Интегральная форма</u>

Если $(\Omega, \mathcal{F}, \mu)$ — вероятностное пространство (т. е. $\mu(\Omega) = 1$) и $f: \Omega \to I$ — интегрируемая функция, то

$$\varphi\!\left(\int_\Omega f\, d\mu\right) \leq \int_\Omega \varphi \circ f \, d\mu.$$

### Интерпретация

Неравенство Йенсена утверждает, что

> *«функция от среднего не превосходит среднее от функции»* (для выпуклых $\varphi$).

**Геометрический смысл.** Возьмём две точки $(x_1, \varphi(x_1))$ и $(x_2, \varphi(x_2))$ на графике выпуклой функции. Точка $(\bar{x}, \varphi(\bar{x}))$, где $\bar{x} = \tfrac{x_1+x_2}{2}$, лежит **ниже** середины хорды $\big(\bar{x}, \tfrac{\varphi(x_1)+\varphi(x_2)}{2}\big)$. Это и есть неравенство Йенсена для двух точек с равными весами.

**Вероятностный смысл.** Если $X$ — случайная величина с разбросом, то выпуклая функция «штрафует» крайние значения сильнее, чем вознаграждает противоположные. Поэтому $\mathbb{E}[\varphi(X)]$ больше, чем $\varphi$ от детерминированного среднего.

**Связь с дисперсией.** Для $\varphi(x) = x^2$ получаем

$$\big(\mathbb{E}[X]\big)^2 \leq \mathbb{E}[X^2],$$

что есть ни что иное, как $\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 \geq 0$.

---

### Доказательство

#### Ключевая лемма (опорная прямая)

Для любой выпуклой функции $\varphi: I \to \mathbb{R}$ и любой внутренней точки $x_0 \in I$ существует число $c \in \mathbb{R}$ такое, что

$$\varphi(x) \geq \varphi(x_0) + c(x - x_0) \quad \text{для всех } x \in I.$$

Геометрически: через каждую точку графика выпуклой функции можно провести прямую, лежащую целиком под графиком (опорную прямую). Если $\varphi$ дифференцируема, то $c = \varphi'(x_0)$. В общем случае подойдёт любое значение между левой и правой производной в точке $x_0$ (которые существуют для выпуклых функций).

#### Основное доказательство

Положим $\mu = \mathbb{E}[X]$. По условию $X$ принимает значения в $I$, и (при разумных предположениях) $\mu$ — внутренняя точка $I$.

Применим лемму при $x_0 = \mu$: существует $c$ такое, что

$$\varphi(x) \geq \varphi(\mu) + c(x - \mu) \quad \forall x \in I.$$

Подставим $x = X(\omega)$ (это допустимо, поскольку $X \in I$ почти наверное):

$$\varphi(X) \geq \varphi(\mu) + c(X - \mu).$$

Возьмём математическое ожидание от обеих частей (монотонность матожидания):

$$\mathbb{E}[\varphi(X)] \geq \varphi(\mu) + c \cdot \mathbb{E}[X - \mu] = \varphi(\mu) + c \cdot 0 = \varphi(\mathbb{E}[X]).$$

Что и требовалось доказать. $\blacksquare$

#### Случай равенства

Если $\varphi$ строго выпукла, то опорное неравенство $\varphi(x) \geq \varphi(\mu) + c(x-\mu)$ строго для всех $x \neq \mu$. Если $X$ не является п.н. константой, то $\mathbb{P}(X \neq \mu) > 0$, и при взятии матожидания получим строгое неравенство.

## Примеры

#### Пример 1<br>Неравенство "средних"

Используется для доказательства знаменитого соотношения между средним арифметическим и средним геометрическим

Пусть $a_1, \ldots, a_n > 0$. Применим неравенство Йенсена к вогнутой функции $\varphi(x) = \ln x$ с равными весами $\lambda_i = \tfrac{1}{n}$:

$$\ln\!\left(\frac{a_1 + \cdots + a_n}{n}\right) \geq \frac{\ln a_1 + \cdots + \ln a_n}{n} = \ln\!\big(\sqrt[n]{a_1 \cdots a_n}\big).$$

Поскольку $\ln$ монотонна,

$$\frac{a_1 + \cdots + a_n}{n} \geq \sqrt[n]{a_1 a_2 \cdots a_n}.$$



---

#### Пример 2<br>Неравенство Коши–Буняковского

Применяя $\varphi(x) = x^2$ (выпуклая) к случайной величине $|X|$:

$$\big(\mathbb{E}|X|\big)^2 \leq \mathbb{E}[X^2].$$

Более общо, применяя неравенство Йенсена к $\varphi(x) = x^p$ при $p \geq 1$, получаем неравенство Ляпунова:

$$\big(\mathbb{E}|X|^q\big)^{1/q} \leq \big(\mathbb{E}|X|^p\big)^{1/p}, \quad 0 < q \leq p.$$

---

#### Пример 3<br>Энтропия и информационная теория

Пусть $p_1, \ldots, p_n$ — распределение вероятностей ($p_i \geq 0$, $\sum p_i = 1$), а $q_1, \ldots, q_n$ — другое распределение. Дивергенция Кульбака–Лейблера:

$$D(p \| q) = \sum_{i=1}^n p_i \ln \frac{p_i}{q_i}.$$

Покажем, что $D(p \| q) \geq 0$. Применим неравенство Йенсена к вогнутой функции $\ln$:

$$-D(p \| q) = \sum_i p_i \ln \frac{q_i}{p_i} \leq \ln\!\left(\sum_i p_i \cdot \frac{q_i}{p_i}\right) = \ln\!\left(\sum_i q_i\right) = \ln 1 = 0.$$

Таким образом $D(p \| q) \geq 0$ — фундаментальный факт теории информации (**неравенство Гиббса**).

---

#### Пример 4 <br>Финансы: вогнутость функции полезности

В экономике поведение «несклонного к риску» агента моделируется **вогнутой** функцией полезности $u$. Неравенство Йенсена даёт:

$$u\big(\mathbb{E}[W]\big) \geq \mathbb{E}\big[u(W)\big],$$

где $W$ — случайное богатство. Это значит, что гарантированно получить $\mathbb{E}[W]$ предпочтительнее, чем играть в лотерею с тем же средним выигрышем — формальное определение **неприятия риска**.

---

#### Пример 5<br>Машинное обучение: ELBO и EM-алгоритм

В вариационном выводе (variational inference) для модели с скрытыми переменными $z$:

$$\ln \underbracket{p(x)}_{вер-ть } = \underbracket{\ln \int p(x,z)\, dz}_{\text{взвешенно по z}} = \ln \int q(z) \frac{p(x,z)}{q(z)}\, dz \geq \int q(z) \ln \frac{p(x,z)}{q(z)}\, dz.$$

Последнее неравенство — это применение Йенсена к вогнутой $\ln$. Правая часть называется **ELBO** (Evidence Lower BOund) и является основой EM-алгоритма и вариационных автоэнкодеров.

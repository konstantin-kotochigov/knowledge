# Парадокс Монти-Холла

Правила игры:
- за одной из 3 дверей приз
- игрок выбирает случайную дверь и демонстрирует ведущему свой выбор
- в качестве реакции ведущий открывает среди двух других дверей ту, которая точно без приза

Вопрос: что выгоднее для игрока, поменять выбор или оставить изначальный<br>
Ответ: выгоднее поменять (аж в 2x раза больше вероятность выиграть)

#### Интуитивное обяснение
С вероятностью 1/3 приз будет за выбранной дверью, с вероятностью 2/3 в паре других. Ведущий выкидыает точно неправильный. Кол-во доступных для выбора вариантов в этой паре уменьшается до одного, но вот вероятность при этом никуда не девается, с вероятностью 2/3 приз все также в этой паре, но уже только во второй двери

#### Более формальное обяхснение

Немного формализуем правила игры<br>$P_A,P_B,P_C$ - выбор игрока<br>$H_A, H_B, H_C$ - действие ведущего, как реакция на выбор игрока. По правилам игры оно детерминированное

Машина распределена по дверям равномерно, поэтому априорные вероятности $(1/3 \quad 1/3 \quad 1/3)$<br>Итого у нас 27 возможных вариантов игры

$
car = A \quad \begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_A & 0 & 0.5 & 0.5 \\
  P_B & 0 & 0 & 1 \\
  P_C & 0 & 1 & 0
\end{array}\quad\quad
$
$car=B \quad
\begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_A & 0 & 0 & 1 \\
  P_B & 0.5 & 0 & 0.5 \\
  P_C & 1 & 0 & 0
\end{array}\quad\quad
$
$car = C \quad
\begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_1 & 0 & 1 & 0 \\
  P_2 & 1 & 0 & 0 \\
  P_3 & 0.5 & 0.5 & 0
\end{array}
$

В таблицах выше вероятность нормирована по строкe, то есть это вероятность действия ведущего в ответ на выбор игрока

С точки зрения игрока тут все симметрично, поэтому можно оставить по одной строчке. Допустим, игрок выбирает дверь A

$
car = A \quad \begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_1 & 0 & 0.5 & 0.5
\end{array}\quad\quad
$
$car=B \quad
\begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_1 & 0 & 0 & 1
\end{array}\quad\quad
$
$car = C \quad
\begin{array}{c|ccc}
     & H_A & H_B & H_C \\ \hline
  P_1 & 0 & 1 & 0
\end{array}
$

Далее игрок видит реакцию ведущего. Пусть она $H_B$. После conditioning-а по наблюдаемым действиям у нас остается вектор из 3х вероятностей, которые надо просто сравнить между собой

$
car = A \quad \begin{array}{c|ccc}
  & H_B \\ \hline
  P_A & 0.5
\end{array}\quad\quad
$
$car=B \quad
\begin{array}{c|ccc}
     & H_B \\ \hline
  P_1 & 0
\end{array}\quad\quad
$
$car = C \quad
\begin{array}{c|ccc}
     & H_B \\ \hline
  P_1 & 1
\end{array}
$

По формуле Байеса оценим апостериорные вероятности расположения машины зная дверь открытую ведущим $P(A|H_B)$

$$P(A|H_B) = \frac{P(H_B|A) P(A)}{P(H_B|A) P(A) + P(H_B|B) P(B) + P(H_B|B) P(C)} = \frac{1/3 \cdot 1/2}{1/3 \cdot 1/2 + 0 \cdot 1/3 + 1 \cdot 1/3} = 1/3$$

$$P(B|H_B) = 0$$

$$P(C|H_B) = \frac{P(H_B|C) P(C)}{P(H_B|A) P(A) + P(H_B|B) P(B) + P(H_B|B) P(C)} = \frac{1/3 \cdot 1}{1/3 \cdot 1/2 + 0 \cdot 1/3 + 1 \cdot 1/3} = 2/3$$

__Интерпретация:__ делая первоначальный выбор, игрок выигрывает с вер-тью 1/3 (приз за выборанной дверью) и проигрывает с вер-тью 2/3 (приз в невыбранной паре). Убирая один неверный вариант, ведущий не меняет расположение выигрыша, то есть он остается с вероятностью 2/3 в невыбранной паре (представленной теперь одним вариантом). Поэтому выгоднее поменять решение и выбрать пару

### Вариация с картами

Более наглядный пример

- Загадываем карту
- Выбираем случайную карту из колоды
- Ведущий убирает из колоды 50 карт, которые точно не загаданная
- Вопрос: какую из оставшихся двух карт выбрать - которую изначально выбрали или которая осталась

__Ответ:__ выгоднее выбирать ту, котрая осталась (1/52 vs 51/52)

__Интерпретация:__ вероятность угадать карту сразу $\frac{1}{52}$. С вероятностью $\frac{51}{52}$ карта в колоде. Когда выкидываем точно неправильные карты, тут случайности нет, вероятности этих двух кейсов остаются теми же самыми. То есть у второй карты вероятность $\frac{51}{52}$

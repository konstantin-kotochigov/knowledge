# Convergence Types
---


#### Convergence in Probability<br>
$X_n \xrightarrow{P} X \quad \text{if} \quad \forall \varepsilon > 0, \ \lim_{n \to \infty} P(|X_n - X| > \varepsilon) = 0$<br><br>

#### Almost Sure Convergence<br>
$X_n \xrightarrow{a.s.} X \quad \text{if} \quad P\left( \lim_{n \to \infty} X_n = X \right) = 1$<br><br>

#### Convergence in Distribution<br>
$X_n \xrightarrow{d} X \quad \text{if} \quad \lim_{n \to \infty} F_{X_n}(x) = F_X(x), \ \text{at all continuity points of } F_X$<br><br>

#### Convergence in Mean (L¹ and L²)<br>
$X_n \xrightarrow{L^1} X \quad \text{if} \quad \lim_{n \to \infty} \mathbb{E}[|X_n - X|] = 0$<br>
$X_n \xrightarrow{L^2} X \quad \text{if} \quad \lim_{n \to \infty} \mathbb{E}[(X_n - X)^2] = 0$<br>


$$X_n \xrightarrow{a.s.} X \Rightarrow X_n \xrightarrow{P} X \Rightarrow X_n \xrightarrow{d} X$$



# The Law of Large Numbers
---

__Idea:__ sample average converges to expectation as number of samples grow

$$\frac{1}{N} \big[ X_1 + X_2 + ... + X_N \big] \longrightarrow \mathbb{E}(X)$$

Never trust small data estimates, always look at scale

History<br>
- Cardano formulated the idea 1500<br>
- Bernoulli proved for binary variables 1700<br>
- Poisson introduced the name<br>
- Generalized by Kolomogorov, Markov etc

There are 2 formulations (weak and strong) that slightly differ in how estimates converge

<u>__Weak__</u><br>
The sample average converges in probability to the expected value as the sample size grows<br>
$\overline{X} \longrightarrow \mu$

With more samples, it becomes unlikely that the sample mean is far from the true mean<br>
$\quad \lim_{n \to \infty} P(|X_n - X| > \varepsilon) = 0$

<u>__Strong__</u><br>
The sample average converges almost surely to the expected value as the sample size grows<br>
$\overline{X} \longrightarrow \mu$

With probability 1, the sample mean converges to the true mean for nearly every infinite sequence of samples<br>
$\quad P\left( \lim_{n \to \infty} X_n = X \right) = 1$

"Almost surely" is a special case of "in probability" => the second formulation is a more strong statement<br>
But the assumptions for the Strong LLN are weaker: it does not require finite variance

# Central Limit Theorem
---
LLN states that sample average converges to the expectation. Now we need a theorem that would describe how exactly it converges

__Idea:__ distribution of sample mean converges a <u>normal distribution</u> as the sample size increases. This property holds no matter what is the original distribution (only finite variance is required)

$\sqrt{n} (\hat{X} - \mu) \rightarrow N(0,\sigma^2)$

CLT proves the asymptotic normality of the sample mean. AN however can be proved for other estimates, for example sample variance as well (see notebook with sample statistics properties). In general many estimators are averages of something => CLT can be applied, so they are also asymptotically normal

There are other versions of the Theorem

# Slutsky’s Theorem
---

If X converges in distribution and Y converges (in probability) to a constant, then the elements of their sum/product/ratio can be substituted with corresponding limits

Let $X_n \xrightarrow{d} X$ and $Y_n \xrightarrow{P} c\,\,\,$  Then:

$\,\,\,X_n + Y_n \xrightarrow{d} X + c$<br>

$\,\,\,X_n \cdot Y_n \xrightarrow{d} X \cdot c$<br>

$\,\,\,\frac{X_n}{Y_n} \xrightarrow{d} \frac{X}{c}$<br>

<br>Here is a practical example<br><br>
One sample T-statistic is a normalized sample mean where unknown deviation $\sigma$ is repalced with a known sample mean $\frac{S_n}{\sqrt{n}}$<br> Asymptotically it behaves like normal $N(0,1)$ => a. we can compute confidence intervals for the mean b. calculate p-value for the hypothesis<br>Thus it justifies using T-statistic for inference

# Continuous Mapping Theorem
---
If X converges and there is a transformation $g(x)$ which is continuous, then $g(X)$ also converges

Let $\{X_n\}_{n=1}^\infty$ be a sequence of random variables and let $X$ be a random variable such that
$X_n \xrightarrow{d} X$
and let $g: \mathbb{R} \to \mathbb{R}$ be a function that is continuous at all points in the support of the distribution of $X$. Then
$g(X_n) \xrightarrow{d} g(X)$

Similarly, if
$X_n \xrightarrow{p} X$ and $g(x)$ is continuous at $X$, then
$g(X_n) \xrightarrow{p} g(X)$


# Neyman–Pearson Lemma
---
Let $X$ be a random variable with probability density $f(x; \theta)$. Suppose we are testing the <u>simple</u> null hypothesis
$H_0: \theta = \theta_0 \quad \text{versus} \quad H_1: \theta = \theta_1$

Among all tests of size $\alpha$, the most powerful test for testing $H_0$ against $H_1$ is the test that rejects $H_0$ in favor of $H_1$ when the likelihood ratio

$$
\Lambda(x) = \frac{f(x; \theta_1)}{f(x; \theta_0)}
$$

exceeds a threshold $k$, where $k$ is chosen such that the test has significance level $\alpha$:

$$
P_{H_0}(\Lambda(X) > k) = \alpha
$$








# Rao–Blackwell Theorem
---
Any unbiased estimator can be improved (in terms of variance) by conditioning it on a sufficient statistic

Let $X$ be a random variable with density $f(x;\theta)$, and let $T(X)$ be a sufficient statistic for $\theta$<br> If $\hat{\theta}(X)$ is an unbiased estimator of $\theta$, then the estimator
$$\tilde{\theta}(X) = \mathbb{E}[\hat{\theta}(X) \mid T(X)]$$

satisfies the following properties:
- $\tilde{\theta}(X)$ is unbiased
- $\operatorname{Var}_\theta(\tilde{\theta}(X)) \leqslant \operatorname{Var}_\theta(\hat{\theta}(X)) \,\, \text{for all } \theta$,
- Equality holds if and only if $\hat{\theta}(X) = \tilde{\theta}(X)$ almost surely (i.e., $\hat{\theta}$ is already a function of $T(X)$).

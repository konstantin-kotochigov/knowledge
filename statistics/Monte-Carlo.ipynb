Метод __Monte-carlo__ = повторяемое сэмплирование. По сути любая вероятностная симуляция, но не любая, а обязательно повторяющаяся для инференса каких-то свойств распределения 

Применение 

1) оптимизация<br><br>
пусть надо оптимизировать сложную f(x). она для нас черный ящик в том смысле, что ее формулу мы не знаем: можем только применить для каого-то x<br><br>
путем многократного случайного сэмплирования мы строим выборочный вариант этой функции - получаем на выходе некий аппроксимирующий её каркас. Далее решаем задачу оптимизации уже на нем<br><br>
если заменить случайное сэмплирование на адаптивное (постепенно смещаемся из тех областей, где плотность больше в те области, где значение меньше/больше)<br><br>
2) интегрирование<br><br>
пусть надо проинтегрировать сложную функцию f(x). аналитически она не описывается. единственная возможность насэмплировать случайных точек и просуммировать их<br><br>
3) сэмплирование<br><br>

#### История создания
Автор метода - польский математик Улам, который вместе с Фон-Нейманом работал над разработкой атомной бомбы в Лос-Аламосе. Название из-за того, что проект был секретным, надо было как-то назвать. По одной из версий название предложил ещё один коллега, Metropolis. По другой он сам вспомнил про своего дядю заядлого игрока

Он изучал вероятность выиграть расклад пасьянса. Для аналитического решения формула слишком сложна, а повторных экспериментов потребовалось бы слишком много, поэтому он запрограммировал на ENIAC-е

Прикладная задача, которую решал Улам - моделирование ядерной реакции. 
Принцип: нейтрон сталкивается с ядром урана, от него "откалывается" несколько нейтронов и выделяется энергия. Далее деление происходит по экспоненте. 

Однако с некоторой вероятностью нейтрон может также отскочить или поглотиться. Эти вероятности зависят от входных условий и влияют на ход процесса. 

Главный показатель процесса - коэффициент деления k: сколько новых нейтронов выделяется в среднем на один свободный нейтрон. Если k > 1, идет неконтроируемая реакция. Задача - понять распределение коэффиуциента деления при разных вводных


Самый простой пример Monte Carlo imputation - подсчет sample average.

Поскольку оригинально это скорее подход, чем математически строгое концепция, все дают свое определение что такое метод Monte-Carlo. 

Похоже на inferential statistics - определение параметров популяции по имеющегося сэмплу. Но в MC во-первых тут есть еще симуляция - то есть сам процесс генерации сэмплов, во-вторых у MC цель пошире.

Bootstrap - цель та же самое, но главное отличие - MC генерирует данные из популяции, bootstrap сэмплирует подвыборку из уже существующих данных

#### Задача Буффона
Вопрос: на полу натянута сетка с шагом l. Какова вероятность, что упавшая на пол иголка длины t будет пересекать сетку?
Ответ: $P = 2/pi * l/t$

Поставлена еще в 18 веке. Практическая полезность в том что через нее можно вычислить pi с любой точностью. В начала 20 века например так нашли самую удобную апроксимацию 355/113

##### Решение
тут две степени свободы: положение - куда упала иголка - x, поворот - theta. Они очевидно независимы. Кроме того два кейса, простой - иголка короче шага сетки, сложный - длинее

Решение 1
собираем совместное распределение (x,theta) и интегрируем двойным интегралом по устраивающему нас интервалу

Решение 2
сначала оценить вероятности а потом перемножить
P = P(попасть центром в допустимую область) x P(развернуть так, чтобы было пересечение)
P1 определяется просто из геометрии l/t
для P2 считаем долю угла theta от всего разворота (pi) - для этого интегрируем по расположениям [0,1] 2/pi
